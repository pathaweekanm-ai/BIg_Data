## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่


## แบบฟอร์ม Insight Card

ทุก insight เขียนในรูปแบบนี้ 5 บรรทัด ไม่ต้องยาว

> **Insight:** ประโยคเดียวที่บอกสิ่งที่พบ ต้องมีตัวเลขอยู่ในประโยค \
> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \
> **So what:** กระทบธุรกิจอย่างไร ถ้าประมาณขนาดผลกระทบเป็นตัวเลขได้ยิ่งดี \
> **Action:** ทีมไหนควรทำอะไร และจะรู้ได้อย่างไรว่าได้ผล \
> **ความมั่นใจ:** สูง กลาง หรือ ต่ำ พร้อมเหตุผลสั้น ๆ

**ตัวอย่างที่ยังไม่ใช่ insight**
> ลูกค้าพูดถึงเรื่องราคาบ่อยที่สุด คิดเป็น 32% ของรีวิว

เป็นเพียงข้อสังเกต ไม่ได้บอกว่าใครต้องทำอะไร และ 32% นั้นมากหรือน้อยก็ไม่รู้เพราะไม่มีอะไรให้เทียบ

**ตัวอย่างที่ใช้ได้**
> **Insight:** รีวิวที่พูดถึงบรรจุภัณฑ์รั่วมีเพียง 6.2% ของทั้งหมด (n = 3,100) แต่ได้ดาวเฉลี่ย 2.1 เทียบกับค่าเฉลี่ยรวม 4.0 และกระจุกอยู่ในสินค้า 12 ชิ้นที่ใช้ขวดแบบปั๊ม \
> **หลักฐาน:** ตารางค่าเฉลี่ยดาวแยกตามด้าน และกราฟแท่งสินค้า 12 ชิ้นในผลวิเคราะห์ \
> **So what:** สินค้า 12 ชิ้นนี้คิดเป็น 18% ของยอดรีวิวทั้งหมวด ถ้าแก้ปัญหาฝาขวดได้ คาดว่าคะแนนเฉลี่ยของหมวดจะขยับขึ้นราว 0.15 ดาว \
> **Action:** ทีม Product ตรวจสอบซัพพลายเออร์ฝาขวดของสินค้า 12 รายการนี้ภายในไตรมาสหน้า แล้วติดตามสัดส่วนรีวิวที่มีคำว่า leak, spill, broken cap ซึ่งตอนนี้อยู่ที่ 14% ต่อเดือน \
> **ความมั่นใจ:** กลาง ยืนยันได้จากข้อความ แต่ยังไม่มีข้อมูลการคืนสินค้ามาช่วยยืนยัน

* คำถามที่เราจะตอบให้เขา: ปัญหาใดที่ทำให้เกิดรีวิวแย่ มีรีวิวแย่มากในหมวดสินค้าไหน และสินค้า ID ไหนต้องแก้ก่อน
* สมมติฐานตั้งต้นของเรา (จะถูกหรือผิดก็ได้): รีวิวแย่ของเครื่องประดับมาจากคุณภาพวัสดุมากกว่าการจัดส่ง

### 2.2 เตรียมข้อมูลที่เชื่อถือได้

ทำเพียงสิ่งจำเป็นก่อนเริ่มวิเคราะห์

* สุ่มรีวิวอย่างน้อย **20,000 รายการ** และอย่าใช้เพียง N บรรทัดแรก
* เชื่อม review กับ product metadata ด้วย `parent_asin`
* รายงานสั้น ๆ ว่า join สำเร็จกี่เปอร์เซ็นต์ และมี missing/duplicate สำคัญหรือไม่
* เขียน 2–3 บรรทัดว่า **ข้อมูลนี้เป็นตัวแทนของใคร และไม่เป็นตัวแทนของใคร**

> จุดประสงค์คือให้รู้ว่าข้อมูลที่ใช้เชื่อถือได้พอสำหรับคำถามของทีม ไม่ใช่ทำ data cleaning ทุกอย่างที่เป็นไปได้


````python
def reservoir_sample(path, k, seed):
    rng = random.Random(seed)
    sample, total = [], 0
    with open(path, encoding="utf-8") as f:
        for i, line in enumerate(f):
            total += 1
            if i < k:
                sample.append(line)
            else:
                j = rng.randint(0, i)
                if j < k:
                    sample[j] = line
    return sample, total

lines, TOTAL_REVIEWS = reservoir_sample(review_path, SAMPLE_N, RANDOM_SEED)
rev = pd.DataFrame([json.loads(l) for l in lines]).drop(columns=["images"], errors="ignore")
del lines
print(f"[Sampling] สุ่ม {len(rev):,} จาก {TOTAL_REVIEWS:,} รีวิว ({len(rev)/TOTAL_REVIEWS:.1%}) seed={RANDOM_SEED}")

# ---------- 2) เชื่อม review กับ product metadata ด้วย parent_asin ----------
META_COLS = ["parent_asin", "main_category", "title", "store", "categories",
             "average_rating", "rating_number", "price"]
with open(meta_path, encoding="utf-8") as f:
    meta = pd.DataFrame([{c: d.get(c) for c in META_COLS} for d in map(json.loads, f)])
meta = meta.rename(columns={"title": "product_title"})     # กันชนกับ title ของรีวิว

meta_dup = int(meta["parent_asin"].duplicated().sum())
meta = meta.drop_duplicates("parent_asin")                   # กันรีวิวถูกนับซ้ำตอน join

df = rev.merge(meta, on="parent_asin", how="left", indicator=True)
join_rate = (df["_merge"] == "both").mean()

# ---------- 3) รายงาน join / missing / duplicate ----------
print(f"\n[Join] สำเร็จ {join_rate:.2%} | หาสินค้าไม่เจอ {(df['_merge']=='left_only').sum():,} รีวิว "
      f"| parent_asin ซ้ำใน meta {meta_dup:,} | แถวก่อน/หลัง join {len(rev):,}/{len(df):,}")

blank = lambda s: s.isna() | s.astype(str).str.strip().isin(["", "None", "nan", "[]"])
missing = pd.DataFrame({
    "field": ["text", "rating", "product_title", "main_category", "categories", "store", "price"],
    "pct_missing": [blank(df["text"]).mean(), df["rating"].isna().mean(),
                    blank(df["product_title"]).mean(), blank(df["main_category"]).mean(),
                    blank(df["categories"]).mean(), blank(df["store"]).mean(),
                    pd.to_numeric(df["price"], errors="coerce").isna().mean()]})
print("\n[Missing]")
display(missing.style.format({"pct_missing": "{:.1%}"}).hide(axis="index"))

dup_n = int(df.duplicated(subset=["user_id", "parent_asin", "text"]).sum())
print(f"[Duplicate] รีวิวซ้ำ (user + สินค้า + ข้อความเดียวกัน) = {dup_n:,} ({dup_n/len(df):.2%})")

# ตัดเฉพาะที่กระทบการวิเคราะห์: รีวิวซ้ำ, ไม่มีข้อความ, join ไม่ได้ (field อื่นที่ว่างแค่รายงาน)
n0 = len(df)
df = (df.drop_duplicates(subset=["user_id", "parent_asin", "text"])
        .loc[lambda d: ~blank(d["text"]) & (d["_merge"] == "both")]
        .drop(columns="_merge")
        .reset_index(drop=True))
print(f"[Clean] ตัดไป {n0-len(df):,} แถว เหลือใช้ต่อ {len(df):,} รีวิว")

# ---------- 4) ตัวเลขสำหรับเขียน "เป็นตัวแทนของใคร" ----------
dates = pd.to_datetime(df["timestamp"], unit="ms")
per_asin = df["parent_asin"].value_counts()
print(f"\n[Representativeness] ช่วงเวลา {dates.min():%Y-%m} ถึง {dates.max():%Y-%m} "
      f"| verified purchase {df['verified_purchase'].mean():.1%} "
      f"| % 1–2 ดาว {(df['rating'] <= 2).mean():.1%}")
print(f"   สินค้า {len(per_asin):,} ตัว | median {per_asin.median():.0f} รีวิว/สินค้า "
      f"| สินค้า top 1% ครองรีวิว {per_asin.head(max(1, len(per_asin)//100)).sum()/len(df):.1%}")
print("   main_category:", df["main_category"].value_counts(normalize=True).head(3).round(3).to_dict())
````

### 2.3 สำรวจข้อมูลเท่าที่จำเป็น

เลือกดูเฉพาะสิ่งที่ช่วยให้เข้าใจ brief ของทีม เช่น

* การแจกแจงดาว
* verified purchase / helpful vote
* ราคา หรือกลุ่มสินค้า
* เวลา
* ความยาวรีวิว

ทำ **2–4 ตารางหรือกราฟ** ก็พอ แล้วเขียนสั้น ๆ ว่า

> “สิ่งใดจาก EDA ทำให้ทีมอยากวิเคราะห์ต่อ และเพราะอะไร”

ไม่ต้องทำ EDA ทุกตัวแปร


````python
import matplotlib.pyplot as plt
from cycler import cycler
from scipy.stats import norm, chi2_contingency

MIN_N_GROUP = 200   # กลุ่มสินค้าต้องมีรีวิวอย่างน้อยเท่านี้จึงจะสรุป
MIN_N_YEAR  = 200   # ปีที่มีรีวิวน้อยกว่านี้ไม่นำมาสรุป

# ---------- พาเลตหลัก 4 สี (ไล่จากรุนแรงไปปกติ) ----------
RED    = "#C62828"   # ปัญหา / รีวิวแย่ / สูงกว่าเฉลี่ย / หัวตาราง
ORANGE = "#FF8F00"   # สีเน้นรอง
YELLOW = "#FBC02D"   # ค่ากลาง ๆ / 3 ดาว / ไม่ต่างจากเฉลี่ย
BEIGE  = "#F5F5DC"   # ปกติ / รีวิวดี / ต่ำกว่าเฉลี่ย

# ---------- สีเสริม ----------
TEXT      = "#3E2723"   # ตัวอักษร แกน เส้นค่าเฉลี่ย (พาเลตไม่มีสีเข้ม จึงเพิ่มเพื่อให้อ่านง่าย)
RED_LIGHT = "#F4B4B4"   # แถบ CI / heatmap ระดับกลาง
RED_WHITE = "#FDECEA"   # ไฮไลต์แถวในตาราง

# ---------- ชื่อตัวแปรเดิม (คงไว้ให้โค้ดทุกเซลล์ใช้ได้เหมือนเดิม) ----------
NAVY       = TEXT
ROSE       = RED
BLUE_GREY  = ORANGE
LIGHT_BLUE = BEIGE
PEACH      = YELLOW
PINK_LIGHT = RED_LIGHT
PINK_WHITE = RED_WHITE
CREAM      = "#FFFFFF"

COLOR_BAD, COLOR_NEUTRAL, COLOR_GOOD = RED, YELLOW, BEIGE
VS_AVG_COLORS = {"สูงกว่าเฉลี่ย": RED, "ไม่ต่างจากเฉลี่ย": YELLOW, "ต่ำกว่าเฉลี่ย": BEIGE}
TABLE_HEADER = [{"selector": "th", "props": [("background-color", RED), ("color", "white")]}]

plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": TEXT,     "axes.labelcolor": TEXT,  "axes.titlecolor": TEXT,
    "xtick.color": TEXT,        "ytick.color": TEXT,      "text.color": TEXT,
    "axes.titleweight": "bold", "axes.titlesize": 12,
    "axes.spines.top": False,   "axes.spines.right": False,
    "axes.grid": True, "grid.color": "#EAEAEA", "grid.linewidth": 0.8, "axes.axisbelow": True,
    "axes.prop_cycle": cycler(color=[RED, ORANGE, YELLOW, TEXT]),
    # ให้ทุกแท่งมีเส้นขอบ (สีครีมจะได้ไม่กลืนกับพื้นขาว)
    "patch.force_edgecolor": True, "patch.edgecolor": "#8D6E63", "patch.linewidth": 0.6,
})

# ---------- ฟังก์ชันและตัวแปรที่ใช้ทุก EDA ----------
def wilson_ci(k, n, alpha=0.05):
    # ช่วงความเชื่อมั่น 95% ของสัดส่วน k/n แบบ Wilson (แม่นกว่าแบบปกติเมื่อ n เล็กหรือ % ต่ำ)
    k, n = np.asarray(k, float), np.asarray(n, float)
    z = norm.ppf(1 - alpha / 2); p = k / n
    den = 1 + z**2 / n
    c = (p + z**2 / (2 * n)) / den
    h = z * np.sqrt(p * (1 - p) / n + z**2 / (4 * n**2)) / den
    return c - h, c + h

df["is_bad"] = (df["rating"] <= 2).astype(int)
df["year"]   = pd.to_datetime(df["timestamp"], unit="ms").dt.year
BAD_AVG = df["is_bad"].mean()
print(f"% รีวิว 1–2 ดาวทั้งหมวด = {BAD_AVG:.2%} (n = {len(df):,})")
````

#### EDA 1 การแจกแจงดาว
* **ทำอะไร:** นับว่ารีวิวแต่ละดาว (1–5) มีกี่ % และเช็กว่ารีวิวแย่ (1–2 ดาว) มีมากพอให้วิเคราะห์ต่อไหม
* **ทำอย่างไร:** นับจำนวนรีวิวแต่ละดาว แล้วทำกราฟแท่ง แยกสี 1–2 ดาว (แดง) / 3 ดาว (เหลือง) / 4–5 ดาว (ครีม)
* **ทำไปเพื่ออะไร:** นับรีวิว 1–2 ดาวเป็น "รีวิวแย่" ถ้าลูกค้าส่วนใหญ่ให้ 5 ดาว คนที่ให้ 1–2 ดาวก็คือคนที่ไม่พอใจชัดเจน
  และ % ของรีวิวแย่ที่ได้ จะใช้เป็น "ค่าปกติ" ไว้เทียบในขั้นต่อไป


````python
dist = df["rating"].value_counts().sort_index()
pct  = dist / dist.sum()

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(dist.index.astype(int).astype(str), pct * 100,
              color=[COLOR_BAD if r <= 2 else COLOR_NEUTRAL if r == 3 else COLOR_GOOD for r in dist.index],
              edgecolor=NAVY, linewidth=0.6)
for b, p, n in zip(bars, pct, dist):
    ax.text(b.get_x() + b.get_width() / 2, b.get_height(), f"{p:.1%}\n(n={n:,})",
            ha="center", va="bottom", fontsize=9)
ax.set_ylim(0, pct.max() * 118)
ax.grid(axis="x", visible=False)
ax.set_xlabel("Star rating"); ax.set_ylabel("% of reviews")
ax.set_title(f"EDA 1: Rating distribution (n={len(df):,}) | bad (1-2 stars) = {BAD_AVG:.1%}")
plt.tight_layout(); plt.show()
````

#### EDA 2 % รีวิวแย่รายกลุ่มสินค้า
* **ทำอะไร:** ดูว่าแต่ละกลุ่มสินค้า (เช่น Jewelry, Home & Kitchen) มีรีวิวแย่กี่ % แล้วเทียบกับค่าปกติของทั้งหมวด
* **ทำอย่างไร:** คำนวณ % รีวิวแย่ของแต่ละกลุ่ม พร้อมช่วงความคลาดเคลื่อน (95% CI) เฉพาะกลุ่มที่มีรีวิวอย่างน้อย 200 รีวิว
  ถ้าทั้งช่วงอยู่เหนือค่าปกติ = "สูงกว่าเฉลี่ย" จริง แล้วใช้ chi-square เช็กว่ากลุ่มต่าง ๆ ต่างกันจริงไหม และ Cramér's V ดูว่าต่างกันมากแค่ไหน
* **ทำไปเพื่ออะไร:** หาว่าควรเจาะกลุ่มสินค้าไหนก่อน


````python
# กลุ่มสินค้ามาจาก categories ของข้อมูลโดยตรง (ไม่ได้จัดกลุ่มเอง)
print("ตัวอย่าง categories:", df["categories"].head(3).tolist())
GROUP_LEVEL = 1   # ตำแหน่งใน categories ที่เป็นชื่อกลุ่มสินค้า ถ้าโครงสร้างต่างให้แก้เลขนี้
df["sub_category"] = df["categories"].apply(
    lambda c: c[GROUP_LEVEL] if isinstance(c, (list, np.ndarray)) and len(c) > GROUP_LEVEL else "(not specified)")

g = df.groupby("sub_category").agg(n=("is_bad", "size"), n_bad=("is_bad", "sum")).reset_index()
n_all_groups = len(g)
g = g[g["n"] >= MIN_N_GROUP].copy()
g["bad_rate"] = g["n_bad"] / g["n"]
g["ci_low"], g["ci_high"] = wilson_ci(g["n_bad"], g["n"])
g["vs_avg"] = np.select([g["ci_low"] > BAD_AVG, g["ci_high"] < BAD_AVG],
                        ["สูงกว่าเฉลี่ย", "ต่ำกว่าเฉลี่ย"], "ไม่ต่างจากเฉลี่ย")
g["share_of_all_neg"] = g["n_bad"] / df["is_bad"].sum()
g = g.sort_values("bad_rate", ascending=False)

print(f"กลุ่มทั้งหมด {n_all_groups} | ผ่านเกณฑ์ n >= {MIN_N_GROUP}: {len(g)} กลุ่ม "
      f"(ครอบคลุม {g['n'].sum()/len(df):.1%} ของรีวิว) | ค่าเฉลี่ยทั้งหมวด = {BAD_AVG:.2%}")
display(g.style.format({"bad_rate": "{:.2%}", "ci_low": "{:.2%}", "ci_high": "{:.2%}",
                        "share_of_all_neg": "{:.1%}"})
         .apply(lambda r: [f"background-color: {PINK_WHITE}" if r["vs_avg"] == "สูงกว่าเฉลี่ย" else ""] * len(r), axis=1)
         .set_table_styles(TABLE_HEADER).hide(axis="index"))

# chi-square: กลุ่มสินค้าสัมพันธ์กับการเป็นรีวิวลบหรือไม่ + Cramér's V วัดขนาดผล
sub = df[df["sub_category"].isin(g["sub_category"])]
tab = pd.crosstab(sub["sub_category"], sub["is_bad"])
chi2, p_chi, dof, _ = chi2_contingency(tab)
cramers_v = np.sqrt(chi2 / (tab.values.sum() * (min(tab.shape) - 1)))
print(f"Chi-square = {chi2:,.1f}, df = {dof}, p = {p_chi:.2g} | Cramér's V = {cramers_v:.3f}")
print("ตีความ:", "p < 0.05 → % รีวิวลบต่างกันระหว่างกลุ่มจริง" if p_chi < 0.05 else "p ≥ 0.05 → ยังไม่พบความต่างระหว่างกลุ่ม",
      "| V < 0.1 → กลุ่มสินค้าอธิบายได้น้อย ต้องดูระดับสินค้าต่อ" if cramers_v < 0.1 else "| V ≥ 0.1 → กลุ่มสินค้ามีผลพอสมควร")

from matplotlib.patches import Patch
from matplotlib.lines import Line2D

# สีที่แยกกันชัด: ชมพู = ปัญหา, เทาอ่อน = ไม่ต่างจากเฉลี่ย, ฟ้า = ดีกว่าเฉลี่ย
PLOT_COLORS = {"สูงกว่าเฉลี่ย": ROSE, "ไม่ต่างจากเฉลี่ย": "#D6D6D6", "ต่ำกว่าเฉลี่ย": LIGHT_BLUE}

gp = g[g["sub_category"] != "(not specified)"].sort_values("bad_rate")   # ไม่ใช่กลุ่มจริง จึงไม่ใส่ในกราฟ
labels = [f"{s}  (n={n:,})" for s, n in zip(gp["sub_category"], gp["n"])]

fig, ax = plt.subplots(figsize=(10, max(3.5, 0.5 * len(gp))))
ax.barh(labels, gp["bad_rate"] * 100, color=gp["vs_avg"].map(PLOT_COLORS), height=0.65)
ax.errorbar(gp["bad_rate"] * 100, range(len(gp)),
            xerr=[(gp["bad_rate"] - gp["ci_low"]) * 100, (gp["ci_high"] - gp["bad_rate"]) * 100],
            fmt="none", ecolor="#8A8A8A", elinewidth=1, capsize=3)
ax.axvline(BAD_AVG * 100, ls="--", c=NAVY, lw=1.3)

# ตัวเลข % ท้ายแท่ง (วางหลังปลาย CI จะได้ไม่ทับเส้น)
for i, (r, hi, v) in enumerate(zip(gp["bad_rate"], gp["ci_high"], gp["vs_avg"])):
    ax.text(hi * 100 + 0.3, i, f"{r:.1%}", va="center", fontsize=9,
            fontweight="bold" if v == "สูงกว่าเฉลี่ย" else "normal")

ax.grid(axis="y", visible=False)
ax.set_xlim(0, gp["ci_high"].max() * 100 * 1.15)
ax.set_xlabel("% bad reviews (1-2 stars)  |  error bar = 95% Wilson CI")
ax.set_title("EDA 2: Which product groups get more bad reviews than average?", loc="left")
ax.legend(handles=[Patch(color=ROSE, label="Above average (CI entirely above)"),
                   Patch(color="#D6D6D6", label="Not different from average"),
                   Patch(color=LIGHT_BLUE, label="Below average (CI entirely below)"),
                   Line2D([0], [0], color=NAVY, ls="--", label=f"Category average {BAD_AVG:.1%}")],
          loc="lower right", frameon=True, facecolor="white", edgecolor="#DDDDDD", fontsize=9)
plt.tight_layout(); plt.show()
````

#### EDA 3 รีวิวแย่แยกตามรหัสสินค้า (parent_asin)
* **ทำอะไร:** หาว่าสินค้าตัวไหน (รหัสสินค้า) มีรีวิวแย่มากที่สุด 15 อันดับ และดูว่ารีวิวแย่กระจุกอยู่ในสินค้าไม่กี่ตัวหรือไม่
* **ทำอย่างไร:** นับรีวิวทั้งหมดและรีวิวแย่ของแต่ละสินค้า แล้วเรียงตาม "จำนวน" รีวิวแย่
* **ทำไปเพื่ออะไร:** ทำเพื่อตอบว่า "สินค้า ID ไหนแย่ และควรแก้แบบเจาะจงรายสินค้า หรือแก้ทั้งหมวด

````python
from matplotlib.patches import Patch

PLOT_COLORS = {"above": ROSE, "same": "#D6D6D6", "below": LIGHT_BLUE}

# ================= ตาราง: สรุปรายสินค้า (parent_asin) =================
asin = (df.groupby("parent_asin")
          .agg(n=("is_bad", "size"), n_bad=("is_bad", "sum"),
               product_title=("product_title", "first"), store=("store", "first"),
               sub_category=("sub_category", "first"),
               avg_rating_all=("average_rating", "first"), n_rating_all=("rating_number", "first"))
          .reset_index())
asin["bad_rate"] = asin["n_bad"] / asin["n"]

print(f"สินค้าใน sample = {len(asin):,} ตัว | median รีวิวต่อสินค้า = {asin['n'].median():.0f} "
      f"| สินค้าที่มีรีวิวลบอย่างน้อย 1 อัน = {(asin['n_bad'] > 0).sum():,} ตัว")

# เรียงตาม "จำนวน" รีวิวลบ ไม่ใช่ % (กันสินค้า 1 รีวิวแย่ 1 อัน = 100% ขึ้นอันดับหนึ่ง)
top_asin = asin.sort_values(["n_bad", "n_rating_all"], ascending=False).head(15).copy()
top_asin["ci_low"], top_asin["ci_high"] = wilson_ci(top_asin["n_bad"], top_asin["n"])
top_asin["vs_avg"] = np.select([top_asin["ci_low"] > BAD_AVG, top_asin["ci_high"] < BAD_AVG],
                               ["above", "below"], "same")
top_asin["product_title"] = top_asin["product_title"].astype(str).str[:55]

display(top_asin[["parent_asin", "product_title", "store", "sub_category", "n", "n_bad", "bad_rate",
                  "avg_rating_all", "n_rating_all"]]
        .style.format({"bad_rate": "{:.0%}", "avg_rating_all": "{:.1f}", "n_rating_all": "{:,.0f}"})
        .set_table_styles(TABLE_HEADER).hide(axis="index"))

# ================= กราฟ: Top 15 สินค้าที่มีรีวิวลบมากที่สุด =================
t = top_asin.iloc[::-1]   # ให้อันดับ 1 อยู่บนสุด

fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(t["parent_asin"], t["n_bad"], color=t["vs_avg"].map(PLOT_COLORS), height=0.65)
for i, (k, n, r, v) in enumerate(zip(t["n_bad"], t["n"], t["bad_rate"], t["vs_avg"])):
    ax.text(k + 0.5, i, f"{k} of {n} reviews = {r:.0%}", va="center", fontsize=9,
            fontweight="bold" if v == "above" else "normal")
ax.grid(axis="y", visible=False)
ax.set_xlim(0, t["n_bad"].max() * 1.45)
ax.set_xlabel("Number of bad reviews (1-2 stars) in sample")
ax.set_ylabel("Product ID (parent_asin)")
ax.set_title("EDA 3: Top 15 products by number of bad reviews", loc="left")
ax.legend(handles=[Patch(color=ROSE, label=f"Bad-review rate above avg {BAD_AVG:.1%} (95% CI)"),
                   Patch(color="#D6D6D6", label="Not different from average"),
                   Patch(color=LIGHT_BLUE, label="Below average (many reviews, few bad)")],
          loc="lower right", facecolor="white", edgecolor="#DDDDDD", fontsize=9)
plt.tight_layout(); plt.show()

# ================= ตัวเลขความกระจุก (ข้อความ ไม่มีกราฟ) =================
a = asin.sort_values("n_bad", ascending=False).reset_index(drop=True)
cum_bad = a["n_bad"].cumsum() / a["n_bad"].sum() * 100
cum_all = a["n"].cumsum() / a["n"].sum() * 100
for p in [1, 5, 10]:
    i = max(0, int(round(p / 100 * len(a))) - 1)
    print(f"สินค้า top {p}% ({i+1:,} ตัว): ครองรีวิวลบ {cum_bad[i]:.1f}% แต่ครองรีวิวทั้งหมด {cum_all[i]:.1f}%")
````

#### EDA 4 ความยาวรีวิวแยกตามดาว
* **ทำอะไร:** ดูว่ารีวิวแต่ละดาวเขียนยาวกี่คำ และรีวิวแย่ (1–2 ดาว) เขียนยาวกว่ารีวิวดี (4–5 ดาว) หรือไม่
* **ทำอย่างไร:** นับคำจากหัวรีวิว + เนื้อรีวิว แล้วแสดงความยาวตรงกลาง (median) และช่วงความยาวของรีวิวส่วนใหญ่ (box plot)
  จากนั้นทดสอบด้วย Mann–Whitney U ว่ารีวิวแย่ยาวกว่าจริงไหม
* **ทำไปเพื่ออะไร:** ไม่ได้ใช้เป็น insight เพราะความยาวไม่ได้บอกว่าต้องแก้สินค้าอย่างไร แต่ใช้เช็กว่า ถ้ารีวิวแย่ยาวกว่า
  ก็จะมีโอกาสเจอคำที่บอกปัญหามากกว่าโดยอัตโนมัติ


````python
from scipy.stats import mannwhitneyu

# จำนวนคำ = หัวรีวิว + เนื้อรีวิว (เก็บไว้ใช้ต่อใน 2.6)
df["n_words"] = (df["title"].fillna("") + " " + df["text"].fillna("")).str.split().str.len()

# ---------- ตาราง: ความยาวรีวิวแยกตามดาว ----------
L = (df.groupby("rating")["n_words"]
       .agg(n="size", median="median",
            q25=lambda s: s.quantile(0.25), q75=lambda s: s.quantile(0.75),
            pct_short=lambda s: (s < 10).mean())
       .reset_index())
display(L.style.format({"median": "{:.0f}", "q25": "{:.0f}", "q75": "{:.0f}", "pct_short": "{:.1%}", "n": "{:,}"})
         .set_table_styles(TABLE_HEADER).hide(axis="index"))

# ---------- ทดสอบ: รีวิวลบยาวกว่ารีวิวดีจริงไหม ----------
bad_len  = df.loc[df["rating"] <= 2, "n_words"]
good_len = df.loc[df["rating"] >= 4, "n_words"]
u, p_mw = mannwhitneyu(bad_len, good_len, alternative="greater")
print(f"median คำ: รีวิวลบ = {bad_len.median():.0f} (n={len(bad_len):,}) | รีวิวดี = {good_len.median():.0f} (n={len(good_len):,})")
print(f"Mann–Whitney U: H0 = ความยาวรีวิวลบกับรีวิวดีไม่ต่างกัน | p = {p_mw:.2g}")
print("ตีความ:", f"รีวิวลบยาวกว่าประมาณ {bad_len.median()/good_len.median():.1f} เท่า → ต้องคุมความยาวใน Reality Check"
      if p_mw < 0.05 else "ความยาวไม่ต่างกันชัดเจน → ผลของความยาวไม่น่ากังวล")

# ---------- Box plot: ความยาวรีวิวต่อระดับดาว ----------
ratings = sorted(df["rating"].dropna().unique())
data    = [df.loc[df["rating"] == r, "n_words"].values for r in ratings]
colors  = [COLOR_BAD if r <= 2 else COLOR_NEUTRAL if r == 3 else COLOR_GOOD for r in ratings]

fig, ax = plt.subplots(figsize=(8, 4.5))
bp = ax.boxplot(data, patch_artist=True, showfliers=False, widths=0.55,
                medianprops=dict(color=NAVY, linewidth=2),
                whiskerprops=dict(color="#8A8A8A"), capprops=dict(color="#8A8A8A"),
                boxprops=dict(edgecolor=NAVY, linewidth=0.8))
for box, c in zip(bp["boxes"], colors):
    box.set_facecolor(c)

# ตัวเลข median และ n เหนือหนวดบน
for i, (r, d_) in enumerate(zip(ratings, data), start=1):
    q1, q3 = np.percentile(d_, [25, 75])
    top = min(d_.max(), q3 + 1.5 * (q3 - q1))
    ax.text(i, top + 2, f"median {np.median(d_):.0f}\n(n={len(d_):,})", ha="center", va="bottom", fontsize=8.5)

ax.set_xticks(range(1, len(ratings) + 1)); ax.set_xticklabels([f"{int(r)}" for r in ratings])
ax.set_ylim(0, ax.get_ylim()[1] * 1.15)
ax.grid(axis="x", visible=False)
ax.set_xlabel("Star rating"); ax.set_ylabel("Words per review")
ax.set_title("EDA 4: Review length by star rating (outliers hidden)", loc="left")
plt.tight_layout(); plt.show()
````


### 2.4 เตรียมข้อความและสร้าง Text Features

ก่อนนำข้อความไปวิเคราะห์ จำเป็นต้องมีการ **preprocessing** เพื่อทำให้ข้อความอยู่ในรูปที่เหมาะสมกับเทคนิคที่จะใช้ อย่างไรก็ตาม ในงานนี้ไม่จำเป็นต้องทำ preprocessing ทุกขั้นตอน ให้เลือกทำ **เท่าที่จำเป็น** ตามวัตถุประสงค์ของการวิเคราะห์

ตัวอย่างเช่น การเปลี่ยนข้อความเป็นตัวพิมพ์เล็ก (lower case) การนำ URL หรือ HTML ออก การแบ่งข้อความเป็นคำ (tokenization) หรือการตัด stopwords เช่น `the`, `and`, `is` ออก ในกรณีที่คำเหล่านี้ไม่ได้ช่วยในการวิเคราะห์

**ไม่บังคับให้ใช้ stemming หรือ lemmatization** หากไม่มีเหตุผลว่าขั้นตอนดังกล่าวจะช่วยตอบคำถามของทีมได้ดีขึ้น

หลังจากเตรียมข้อความแล้ว ให้สร้าง **Text Features อย่างน้อย 2 ชนิด** เพื่อนำข้อความมาเปลี่ยนให้อยู่ในรูปข้อมูลที่สามารถนำไปสำรวจ เปรียบเทียบ หรือวิเคราะห์ต่อได้

---

#### Text Feature คืออะไร?

Text Feature คือ **ข้อมูลเชิงตัวเลขที่สกัดหรือสร้างขึ้นจากข้อความ** เพื่อแทนลักษณะหรือความหมายบางอย่างของข้อความนั้น

Text Feature อาจอยู่ในรูปของ

* **ค่าตัวเลขค่าเดียว** เช่น จำนวนคำ หรือ sentiment score
* **ตัวแปรหลายตัว** เช่น `price=1`, `quality=0`, `delivery=1`
* **เวกเตอร์หลายมิติ** เช่น TF-IDF vector หรือ Text Embedding

ตัวอย่างเช่น รีวิว

> `"The product is easy to use but the delivery was very slow."`

สามารถเปลี่ยนเป็น feature ได้หลายรูปแบบ

| Feature                    |                 ตัวอย่างค่าที่ได้ |
| -------------------------- | --------------------------------: |
| จำนวนคำในรีวิว             |                                12 |
| Sentiment score            |                             -0.15 |
| กล่าวถึง usability หรือไม่ |                                 1 |
| กล่าวถึง delivery หรือไม่  |                                 1 |
| กล่าวถึง price หรือไม่     |                                 0 |
| TF-IDF vector              |   `[0.00, 0.41, 0.00, 0.72, ...]` |
| Embedding vector           | `[0.21, -0.08, 0.63, ..., -0.14]` |

ดังนั้นข้อความหนึ่งข้อความสามารถถูกแทนได้ทั้งด้วย **feature ที่ตีความได้โดยตรง** และ **feature แบบเวกเตอร์ที่ใช้แทนเนื้อหาหรือความหมายของข้อความ**

---

## ตัวอย่าง Text Features ที่สามารถเลือกใช้

ผมจะแก้ตารางเดิมเป็นแบบนี้:

| Text Feature          | คืออะไร                                     | ตัวอย่าง                     | ใช้ตอบคำถามอะไรได้                                    |
| --------------------- | ------------------------------------------- | ---------------------------- | ----------------------------------------------------- |
| **Review Length**     | จำนวนคำหรือตัวอักษร                         | 120 words                    | รีวิวที่ยาวกว่าได้รับ helpful vote มากกว่าหรือไม่     |
| **Sentiment Score**   | คะแนนความเป็นบวก/ลบ                         | `-0.72`, `+0.85`             | ความรู้สึกลูกค้าสัมพันธ์กับ rating หรือไม่            |
| **Aspect Indicators** | ระบุว่าพูดถึงหัวข้อใด                       | `delivery=1`, `price=0`      | ลูกค้าบ่นเรื่องใดมากที่สุด                            |
| **Lexical Diversity** | ความหลากหลายของคำ                           | unique words / total words   | รีวิวที่เขียนละเอียดมีประโยชน์มากกว่าหรือไม่          |
| **TF-IDF Vector**     | เวกเตอร์ที่ให้น้ำหนักคำเด่นในเอกสาร         | `[0, .35, 0, .81, ...]`      | คำใดแยกกลุ่มรีวิวบวกกับลบ                             |
| **Text Embedding**    | dense vector ที่แทนความหมายโดยรวมของข้อความ | `[.12, -.44, .08, ...]`      | รีวิวใดมีความหมายคล้ายกัน หรือมี customer themes อะไร |
| **Custom Feature**    | feature ที่ทีมสร้างตามโจทย์                 | `refund=1`, `defect_count=3` | ตอบ business question เฉพาะของทีม                     |

#### 2.4.1 Preprocessing
* **ทำอะไร:** สร้าง text_all = หัวรีวิว + เนื้อรีวิว ที่ทำความสะอาดแล้ว
* **ทำอย่างไร:** ลบ HTML และ URL, แปลงเป็นตัวพิมพ์เล็ก, ยุบช่องว่าง
  - ไม่ตัด stopwords ในขั้นนี้ เพราะ keyword ปัญหาต้องใช้คำเล็ก ๆ เช่น "too small", "not as described", "fell out" (ตัด stopwords เฉพาะตอนทำ TF-IDF)
  - ไม่ทำ stemming/lemmatization เพราะ regex ใน dictionary ครอบคลุมรูปคำที่ต้องการแล้ว (เช่น tarnish\w* จับ tarnished, tarnishes)
* **ทำไปเพื่ออะไร:** หัวรีวิวมักสรุปปัญหาไว้ชัด เช่น "Broke after 2 days" จึงรวมเข้าไปด้วย และข้อความที่สะอาดทำให้ keyword จับได้ตรง
* **ดูผลตรงไหน:** ตัวอย่างข้อความก่อน–หลัง ต้องไม่มี HTML เหลือ และหัวรีวิวต้องต่ออยู่ด้านหน้า

````python
import re

def clean_text(s):
    s = str(s or "")
    s = re.sub(r"<[^>]+>", " ", s)            # ลบ HTML
    s = re.sub(r"http\S+|www\.\S+", " ", s)   # ลบ URL
    s = s.lower()
    return re.sub(r"\s+", " ", s).strip()

df["text_all"] = (df["title"].fillna("") + ". " + df["text"].fillna("")).map(clean_text)
df["n_words"]  = df["text_all"].str.split().str.len()

ex = df[df["text"].str.contains("<br", na=False)].sample(2, random_state=RANDOM_SEED)
for _, r in ex.iterrows():
    print("ก่อน:", (str(r["title"]) + " | " + str(r["text"]))[:160])
    print("หลัง:", r["text_all"][:160], "\n")
````

#### 2.4.2 Feature 1: Aspect Indicators (feature หลัก)
* **ทำอะไร:** สร้างคอลัมน์ 0/1 ต่อประเภทปัญหา เช่น asp_durability = 1 ถ้ารีวิวพูดถึงของหัก/หลุด
* **ทำอย่างไร:** ใช้ keyword dictionary แบบ regex 9 ด้านที่ออกแบบให้เข้ากับสินค้า handmade และผลจาก EDA (จิว/ห่วงจมูก, สร้อยสลักชื่อ) ด้าน shipping ใส่ไว้เพื่อ "แยกออก" เพราะเป็นงานของทีม Operations ไม่ใช่ Product
* **ทำไปเพื่ออะไร:** เปลี่ยนคำถามจาก "ลูกค้าไม่พอใจ" เป็น "ลูกค้าไม่พอใจเรื่องอะไร" ซึ่งทีม Product ลงมือแก้ได้ และนำไปนับรายกลุ่ม/รายสินค้าได้ใน 2.5
* **ดูผลตรงไหน:** กราฟ % ของรีวิวลบ เทียบรีวิวดี ต่อด้าน

````python
ASPECTS = {
    "durability":       r"\b(?:broke|broken|breaks|fell apart|falling apart|came apart|snapp?ed|flimsy|cheaply made|fragile|fell (?:off|out)|came (?:off|out)|clasp)\b",
    "tarnish_material": r"\b(?:tarnish\w*|turn(?:ed|s)? (?:green|black|brown)|rust\w*|discolou?r\w*|fad(?:ed|ing)|peel(?:ed|ing|s)?|not (?:real|sterling|gold)|fake)\b",
    "skin_reaction":    r"\b(?:allerg\w*|itch\w*|rash|irritat\w*|infect\w*|swell\w*|green (?:skin|mark)|hurt\w*|pain\w*)\b",
    "size_fit":         r"\b(?:too (?:small|big|large|short|long|tight|loose|thick|thin)|tiny|smaller than|bigger than|larger than|(?:does ?n[o']?t|did ?n[o']?t|did not|does not|won'?t) fit|wrong size|sizing|gauge)\b",
    "not_as_pictured":  r"\b(?:not as (?:pictured|described|shown|advertised)|looks? (?:nothing|different)|than (?:the )?(?:picture|photo|pictured|image)|misleading|colou?r (?:is|was) (?:off|different|wrong)|not what i (?:expected|ordered))\b",
    "personalization":  r"\b(?:misspell\w*|spelled wrong|wrong (?:name|date|initials?|letters?|font)|engrav\w*|personali[sz]\w*|custom name)\b",
    "craftsmanship":    r"\b(?:glue\w*|crooked|uneven|sloppy|poorly made|poor quality|bad quality|low quality|stitch\w*|messy|scratch\w*|chipp?ed|defect\w*|sharp edge\w*)\b",
    "scent":            r"\b(?:smell\w*|scent\w*|fragrance|odou?r)\b",
    "shipping":         r"\b(?:arrived (?:broken|damaged|late|crushed)|never (?:arrived|received|came)|shipping|delivery|packag\w*|took (?:weeks|forever|a month))\b",
}
OWNER = {a: ("Operations" if a == "shipping" else "Product") for a in ASPECTS}

for a, pat in ASPECTS.items():
    df[f"asp_{a}"] = df["text_all"].str.contains(pat, regex=True).astype(int)
ASP_COLS = [f"asp_{a}" for a in ASPECTS]
df["any_aspect"] = df[ASP_COLS].max(axis=1)

bad, good = df[df["rating"] <= 2], df[df["rating"] >= 4]
cover = pd.DataFrame({
    "aspect":      list(ASPECTS),
    "owner":       [OWNER[a] for a in ASPECTS],
    "pct_all":     df[ASP_COLS].mean().values,
    "pct_in_bad":  bad[ASP_COLS].mean().values,
    "pct_in_good": good[ASP_COLS].mean().values,
})
cover["lift"] = cover["pct_in_bad"] / cover["pct_in_good"].clip(lower=1e-9)
cover = cover.sort_values("pct_in_bad", ascending=False)

display(cover.style.format({"pct_all": "{:.1%}", "pct_in_bad": "{:.1%}", "pct_in_good": "{:.1%}", "lift": "{:.1f}x"})
             .set_table_styles(TABLE_HEADER).hide(axis="index"))
print(f"coverage: รีวิวลบที่ติดอย่างน้อย 1 ด้าน = {bad['any_aspect'].mean():.1%} (n bad = {len(bad):,})")
````

#### 2.4.3 Feature 2: TF-IDF คำที่รีวิวลบใช้ต่างจากรีวิวดี
* **ทำอะไร:** หาคำ/วลี 2 คำที่โดดเด่นในรีวิวลบเมื่อเทียบกับรีวิวดี
* **ทำอย่างไร:** TF-IDF (unigram + bigram, ตัด stopwords ภาษาอังกฤษ, คำต้องปรากฏ ≥ 10 รีวิว) แล้วเทียบค่า TF-IDF เฉลี่ยของกลุ่มรีวิวลบลบด้วยกลุ่มรีวิวดี
* **ทำไปเพื่ออะไร:** dictionary ใน 2.4.2 มาจากสิ่งที่ทีมคาดไว้ TF-IDF ช่วย "ค้นพบ" ปัญหาที่ยังไม่ได้คิด แล้วนำคำไปเติม dictionary (เช่น ถ้าเจอ "fell out", "backing" ให้เติมใน durability)
* **ดูผลตรงไหน:** กราฟ Top 20 คำเด่นในรีวิวลบ แท่งสีชมพู = คำที่ยังไม่อยู่ใน dictionary ถ้าเป็นคำที่บอกปัญหา ให้เติมเข้า ASPECTS ส่วนตารางคำเด่นในรีวิวดีใช้เทียบให้เห็นภาพ

````python
from sklearn.feature_extraction.text import TfidfVectorizer

bg = df[(df["rating"] <= 2) | (df["rating"] >= 4)]
is_bad_arr = (bg["rating"] <= 2).values

vec = TfidfVectorizer(stop_words="english", ngram_range=(1, 2), min_df=10, max_features=30_000)
X = vec.fit_transform(bg["text_all"])
terms = np.array(vec.get_feature_names_out())

diff = np.asarray(X[is_bad_arr].mean(axis=0)).ravel() - np.asarray(X[~is_bad_arr].mean(axis=0)).ravel()
all_patterns = "|".join(ASPECTS.values())

TOP_K = 20
top_bad  = pd.DataFrame({"term": terms[np.argsort(-diff)[:TOP_K]], "score": np.sort(-diff)[:TOP_K] * -1})
top_good = pd.DataFrame({"term": terms[np.argsort(diff)[:TOP_K]],  "score": np.sort(diff)[:TOP_K] * -1})
top_bad["in_dictionary"] = top_bad["term"].apply(lambda w: bool(re.search(all_patterns, w)))

print(f"รีวิวที่ใช้ = {len(bg):,} (ลบ {is_bad_arr.sum():,} / ดี {(~is_bad_arr).sum():,}) | คำศัพท์ = {len(terms):,}")
````

#### 2.4.4 Feature 3: Sentiment Score (VADER) — ตรวจสอบไขว้
* **ทำอะไร:** ให้คะแนนความรู้สึกของข้อความ (−1 ลบ ถึง +1 บวก) แล้วเทียบกับเกณฑ์ rating ≤ 2
* **ทำอย่างไร:** ใช้ VADER compound จัดเป็น negative (< −0.05) / neutral / positive (> 0.05) แล้วทำตารางไขว้กับกลุ่มดาว
* **ทำไปเพื่ออะไร:** ยืนยันว่าเกณฑ์ดาวที่ทีมเลือกสอดคล้องกับน้ำเสียงจริง และแสดงหลักฐานว่าทำไมไม่ใช้ sentiment เป็นเกณฑ์หลัก เพราะ VADER อ่านรีวิวผสม เช่น "pretty but broke in a week" หรือรีวิวประชดผิดได้
* **ดูผลตรงไหน:** แถว bad (1-2) — % ที่ VADER ให้ positive คือขนาดความผิดพลาดถ้าใช้ sentiment เป็นเกณฑ์ และตัวอย่างรีวิว 1–2 ดาวที่ VADER ให้คะแนนบวก

````python
import nltk
nltk.download("vader_lexicon", quiet=True)
from nltk.sentiment import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()
df["vader"] = df["text_all"].map(lambda s: sia.polarity_scores(s)["compound"])
df["vader_label"] = pd.cut(df["vader"], [-1.01, -0.05, 0.05, 1.01], labels=["negative", "neutral", "positive"])
df["star_label"]  = np.select([df["rating"] <= 2, df["rating"] >= 4], ["bad (1-2)", "good (4-5)"], "neutral (3)")

ct = pd.crosstab(df["star_label"], df["vader_label"], normalize="index")
display(ct.style.format("{:.1%}").set_table_styles(TABLE_HEADER))

agree = (df.loc[df["rating"] <= 2, "vader_label"] == "negative").mean()
miss  = (df.loc[df["rating"] <= 2, "vader_label"] == "positive").mean()
print(f"รีวิว 1–2 ดาวที่ VADER ให้ negative = {agree:.1%} | ให้ positive (อ่านผิด) = {miss:.1%}")
````

#### 2.4.6 Feature 4 : Text Embedding + Clustering — ธีมของรีวิวแย่
* **ทำอะไร:** จัดกลุ่มรีวิว 1–2 ดาวตามความหมาย แล้วดูว่าแต่ละกลุ่ม (ธีม) ลูกค้าบ่นเรื่องอะไร
* **ทำอย่างไร:**
  * แปลงรีวิวแย่แต่ละอันเป็นเวกเตอร์ความหมาย 384 มิติด้วยโมเดล all-MiniLM-L6-v2 (รีวิวที่ความหมายคล้ายกันจะได้เวกเตอร์ใกล้กัน แม้ใช้คำต่างกัน เช่น "fell apart" กับ "broke in two days")
  * จัดกลุ่มด้วย K-Means โดยเลือกจำนวนกลุ่ม (k = 4–8) จาก silhouette score (ยิ่งสูง กลุ่มยิ่งแยกกันชัด)
  * ตั้งชื่อธีมจากคำ TF-IDF ที่เด่นในกลุ่มนั้นเมื่อเทียบกับรีวิวแย่ทั้งหมด และ aspect ที่พบบ่อยที่สุด
* **ทำไปเพื่ออะไร:** Aspect dictionary จับได้เฉพาะคำที่ทีมคิดไว้ล่วงหน้า Embedding จัดกลุ่มตามความหมายโดยไม่ต้องกำหนดคำ จึงช่วย (1) ยืนยันว่าปัญหาที่ dictionary จับได้เป็นธีมหลักจริง และ (2) ค้นพบปัญหาที่ dictionary ยังไม่ครอบคลุม
* **ดูผลตรงไหน:** คอลัมน์ top_terms และ example ใช้ตั้งชื่อธีม, top_aspect บอกว่าธีมนั้นตรงกับ aspect ไหน, %_no_aspect ถ้าสูง (เช่น > 50%) แปลว่าเป็นปัญหาที่ dictionary ยังจับไม่ได้ ให้เติมคำเข้า ASPECTS

````python
RUN_EMBEDDING = True     # ตั้งเป็น False ถ้าไม่ต้องการส่วน Bonus (ใช้เวลาประมาณ 1–3 นาทีบน CPU)
if RUN_EMBEDDING:
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "sentence-transformers"])
    from sentence_transformers import SentenceTransformer
    from sklearn.cluster import KMeans
    from sklearn.metrics import silhouette_score

    # ---------- 1) Embedding ของรีวิวแย่ ----------
    neg_df = df[df["is_bad"] == 1].copy()
    model = SentenceTransformer("all-MiniLM-L6-v2")
    E = model.encode(neg_df["text_all"].tolist(), batch_size=64, show_progress_bar=True,
                     normalize_embeddings=True)
    print("Embedding shape:", E.shape)   # (จำนวนรีวิวแย่, 384)

    # ---------- 2) เลือกจำนวนธีมด้วย silhouette score ----------
    scores = {k: silhouette_score(E, KMeans(k, n_init=10, random_state=RANDOM_SEED).fit_predict(E),
                                  sample_size=min(2000, len(E)), random_state=RANDOM_SEED)
              for k in range(4, 9)}
    best_k = max(scores, key=scores.get)
    print("silhouette:", {k: round(v, 3) for k, v in scores.items()}, "→ เลือก k =", best_k)
    neg_df["theme"] = KMeans(best_k, n_init=10, random_state=RANDOM_SEED).fit_predict(E)

    # ---------- 3) ตั้งชื่อธีม: คำ TF-IDF เด่น + aspect ที่พบบ่อย ----------
    Xn = X[bg.index.get_indexer(neg_df.index)]            # แถว TF-IDF ของรีวิวแย่ (X สร้างจาก bg ใน 2.4.4)
    neg_mean = np.asarray(Xn.mean(axis=0)).ravel()        # ค่าเฉลี่ยของรีวิวแย่ทั้งหมด

    rows = []
    for c in range(best_k):
        mk = (neg_df["theme"] == c).to_numpy()
        distinct = np.asarray(Xn[mk].mean(axis=0)).ravel() - neg_mean   # เด่นในธีมนี้เมื่อเทียบกับรีวิวแย่ทั้งหมด
        top = terms[np.argsort(-distinct)[:6]]
        asp = neg_df.loc[mk, ASP_COLS].mean().sort_values(ascending=False)
        rows.append({"theme": c, "n": int(mk.sum()), "%_of_neg": mk.mean() * 100,
                     "top_terms": ", ".join(top),
                     "top_aspect": asp.index[0].replace("asp_", ""), "%_top_aspect": asp.iloc[0] * 100,
                     "%_no_aspect": (1 - neg_df.loc[mk, "any_aspect"].mean()) * 100,
                     "example": neg_df.loc[mk, "text_all"].sample(1, random_state=RANDOM_SEED).iloc[0][:110]})
    themes = pd.DataFrame(rows).sort_values("n", ascending=False)

    print(f"\nFeature 4 (Bonus) — ธีมของรีวิว 1–2 ดาวจาก Embedding (n = {len(neg_df):,})")
    display(themes.style.format({"%_of_neg": "{:.1f}", "%_top_aspect": "{:.1f}", "%_no_aspect": "{:.1f}"})
                  .set_table_styles(TABLE_HEADER).hide(axis="index"))

    gap = themes[themes["%_no_aspect"] > 50]
    if len(gap):
        print("ธีมที่ dictionary ยังจับไม่ได้เกินครึ่ง → ดู top_terms แล้วพิจารณาเติมคำเข้า ASPECTS:")
        for _, r in gap.iterrows():
            print(f"  theme {r['theme']} (n={r['n']:,}): {r['top_terms']}")
````

* VADER ยืนยันว่าการใช้ดาว 1-2 เป็นนิยามรีวิวแย่เหมาะกว่าการใช้ sentiment เพราะ sentiment อ่านรีวิวผสมคำชมกับคำติผิดได้

#### 2.5.2 ปัญหาไหนหนักในสินค้าย่อยไหน (Segmentation)
* **ทำอะไร:** heatmap สินค้าย่อย × ประเภทปัญหา ค่าในช่องคือ excess bad reviews ภายในกลุ่มนั้น
* **ทำอย่างไร:** ทำ 2.5.1 ซ้ำแยกทีละ sub_category (เฉพาะกลุ่ม n ≥ 200 และไม่รวม "not specified") เทียบรีวิวที่พูดถึงกับไม่พูดถึงปัญหา **ภายในกลุ่มเดียวกัน** ช่องที่ถูกพูดถึงน้อยกว่า 30 รีวิวเว้นว่าง
* **ทำไปเพื่ออะไร:** ปัญหาเดียวกันหนักไม่เท่ากันในแต่ละกลุ่ม เช่น skin_reaction สำคัญกับจิว แต่ scent สำคัญกับสบู่/เทียน การเทียบภายในกลุ่มยังกันไม่ให้กลุ่มที่แย่อยู่แล้วทำให้ปัญหาดูหนักเกินจริง
* **ดูผลตรงไหน:** ช่องสีชมพูเข้มที่สุด คือคู่ (กลุ่ม, ปัญหา) ที่ทำให้เกิดรีวิวลบเกินมามากที่สุด

````python
groups_ok = g.loc[(g["sub_category"] != "(not specified)"), "sub_category"].tolist()
cell_rows = []
for sc in groups_ok:
    d = df[df["sub_category"] == sc]
    for a in ASPECTS:
        m = d[f"asp_{a}"] == 1
        if m.sum() == 0 or (~m).sum() == 0:
            continue
        br_m, br_n = d.loc[m, "is_bad"].mean(), d.loc[~m, "is_bad"].mean()
        cell_rows.append({"sub_category": sc, "aspect": a, "owner": OWNER[a], "n_group": len(d),
                          "n_mention": int(m.sum()), "n_bad_mention": int(d.loc[m, "is_bad"].sum()),
                          "bad_rate_if_mention": br_m, "bad_rate_if_not": br_n,
                          "excess_bad": m.sum() * (br_m - br_n)})
cell = pd.DataFrame(cell_rows)
cell["excess_bad_full"] = cell["excess_bad"] * SCALE
cell_ok = cell[cell["n_mention"] >= MIN_N_MENTION].copy()

heat = cell_ok.pivot(index="sub_category", columns="aspect", values="excess_bad").reindex(columns=list(ASPECTS))
heat = heat.loc[heat.clip(lower=0).sum(axis=1).sort_values(ascending=False).index]

cmap = LinearSegmentedColormap.from_list("rose", ["#FFFFFF", PINK_LIGHT, ROSE, "#C9607C"])
fig, ax = plt.subplots(figsize=(11, max(3.5, 0.5 * len(heat))))
im = ax.imshow(heat.clip(lower=0).fillna(0).values, cmap=cmap, aspect="auto")
ax.set_xticks(range(heat.shape[1])); ax.set_xticklabels(heat.columns, rotation=30, ha="right")
ax.set_yticks(range(heat.shape[0])); ax.set_yticklabels(heat.index)
for i in range(heat.shape[0]):
    for j in range(heat.shape[1]):
        v = heat.values[i, j]
        ax.text(j, i, "" if np.isnan(v) else f"{v:.0f}", ha="center", va="center", fontsize=8, color=NAVY)
ax.grid(False)
plt.colorbar(im, ax=ax, label="Excess bad reviews (sample)")
ax.set_title(f"2.5.2 Excess bad reviews: sub-category x issue (blank = mentions < {MIN_N_MENTION})", loc="left")
plt.tight_layout(); plt.show()
````

````python
TOP3 = (cell_ok.sort_values("excess_bad_full", ascending=False)
               .head(3)[["sub_category", "aspect"]]
               .itertuples(index=False, name=None))
TOP3 = list(TOP3) # Convert to list of tuples
print(f"Top 3 issues for drill-down: {TOP3}")

# ผลรวมรีวิวลบส่วนเกินของ 3 เรื่อง และรีวิวลบทั้งหมวด (ขยายเป็นทั้งหมวดแล้ว) ใช้ใน Insight Card
fix3 = cell_ok.set_index(["sub_category", "aspect"]).loc[TOP3, "excess_bad_full"].sum()
total_bad_full = df["is_bad"].sum() * SCALE
print(f"แก้ครบ 3 เรื่อง ลดรีวิวลบได้สูงสุดราว {fix3:,.0f} จาก {total_bad_full:,.0f} รีวิวลบทั้งหมวด ({fix3/total_bad_full:.0%})")
````

#### 2.5.3 สินค้ารหัสไหน (parent_asin) ต้องแก้ก่อน
* **ทำอะไร:** จัดอันดับสินค้ารหัสต่าง ๆ ภายในแต่ละเรื่องของ TOP3 ว่าสินค้าตัวไหนมีรีวิวแย่เรื่องนั้นมากที่สุด
* **ทำอย่างไร:** เอารีวิวใน sample ที่อยู่ในกลุ่มสินค้าของ TOP3 มานับต่อสินค้า ได้ n (รีวิวทั้งหมด), n_bad (รีวิวแย่) และ n_bad_issue (รีวิวแย่ที่พูดถึงปัญหานั้น) เก็บเฉพาะสินค้าที่มีรีวิวอย่างน้อย 20 ฉบับ แล้วเรียงตาม n_bad_issue ตัวไหนที่ขอบล่างของ Wilson CI 95% ยังสูงกว่าค่าเฉลี่ยกลุ่ม จะถือว่าแย่กว่ากลุ่มจริง
* **ทำไปเพื่ออะไร:** ให้ตอบลูกค้าได้ว่า "สินค้ารหัสไหน ร้านไหน ต้องแก้เรื่องอะไร" ไม่ใช่แค่บอกว่าทั้งกลุ่มมีปัญหา
* **ดูผลตรงไหน:** ตาราง Top 10 ต่อเรื่อง และกราฟ แท่งสีชมพูคือ % รีวิวแย่สูงกว่าค่าเฉลี่ยกลุ่มจริง คอลัมน์ share_of_issue คือสินค้านี้ครองรีวิวแย่เรื่องนั้นกี่ % ของกลุ่ม

````python
MIN_N_ASIN = 20   # สินค้าต้องมีรีวิวใน sample อย่างน้อยเท่านี้ (ถ้าตารางมีสินค้าน้อยเกินไป ลดเป็น 10 ได้)

target_subs = {sc for sc, _ in TOP3}

# ใช้ sample ที่สุ่มมา แทนการอ่านไฟล์เต็ม
full = df[df["sub_category"].isin(target_subs)].copy()
if "year" not in full:
    full["year"] = pd.to_datetime(full["timestamp"], unit="ms").dt.year
print(f"ข้อมูลจาก sample ของ {sorted(target_subs)} = {len(full):,} รีวิว จาก {full['parent_asin'].nunique():,} สินค้า "
      f"| สินค้าที่มีรีวิว >= {MIN_N_ASIN}: {(full['parent_asin'].value_counts() >= MIN_N_ASIN).sum():,}")

# ================= ตารางของแต่ละเรื่อง =================
asin_rank, grp_rates = {}, {}
for rank, (sc, a) in enumerate(TOP3, start=1):
    d = full[full["sub_category"] == sc]
    grp_rate = d["is_bad"].mean(); grp_rates[(sc, a)] = grp_rate
    total_issue = int((d["is_bad"] * d[f"asp_{a}"]).sum())
    t = (d.assign(bad_issue=d["is_bad"] * d[f"asp_{a}"])
          .groupby("parent_asin")
          .agg(product_title=("product_title", "first"), store=("store", "first"),
               n=("is_bad", "size"), n_bad=("is_bad", "sum"), n_bad_issue=("bad_issue", "sum"))
          .reset_index())
    t = t[t["n"] >= MIN_N_ASIN].copy()
    t["bad_rate"] = t["n_bad"] / t["n"]
    t["ci_low"], t["ci_high"] = wilson_ci(t["n_bad"], t["n"])
    t["group_bad_rate"] = grp_rate
    t["vs_group"] = np.where(t["ci_low"] > grp_rate, "สูงกว่ากลุ่ม", "ไม่ต่าง")
    t["share_of_issue"] = t["n_bad_issue"] / max(total_issue, 1)
    t = t.sort_values(["n_bad_issue", "bad_rate"], ascending=False).reset_index(drop=True)
    asin_rank[(sc, a)] = t

    print(f"\n=== เรื่องที่ {rank}: {sc} | {a} | % แย่ของกลุ่ม = {grp_rate:.1%} | "
          f"รีวิวแย่เรื่องนี้ทั้งกลุ่ม = {total_issue:,} | สินค้าที่ n >= {MIN_N_ASIN}: {len(t):,} ===")
    show = t.head(10).copy(); show["product_title"] = show["product_title"].astype(str).str[:50]
    display(show[["parent_asin", "product_title", "store", "n", "n_bad", "n_bad_issue", "share_of_issue",
                  "bad_rate", "group_bad_rate", "vs_group"]]
            .style.format({"share_of_issue": "{:.1%}", "bad_rate": "{:.1%}", "group_bad_rate": "{:.1%}"})
            .set_table_styles(TABLE_HEADER).hide(axis="index"))

# ================= กราฟ 3 เรื่องเรียงกัน =================
fig, axes = plt.subplots(1, len(TOP3), figsize=(6.2 * len(TOP3), 5.5))
axes = np.atleast_1d(axes)
for rank, ((sc, a), ax) in enumerate(zip(TOP3, axes), start=1):
    p = asin_rank[(sc, a)].head(10).iloc[::-1]
    if p.empty:
        ax.text(0.5, 0.5, "No product passes min n", ha="center", va="center", transform=ax.transAxes)
        ax.set_title(f"#{rank}: {a}", loc="left"); continue
    ax.barh(p["parent_asin"], p["n_bad_issue"],
            color=np.where(p["vs_group"] == "สูงกว่ากลุ่ม", ROSE, GREY), height=0.65)
    for i, (k, r, n) in enumerate(zip(p["n_bad_issue"], p["bad_rate"], p["n"])):
        ax.text(k + p["n_bad_issue"].max() * 0.02, i, f"{k} | {r:.0%} of {n}", va="center", fontsize=8)
    ax.set_xlim(0, p["n_bad_issue"].max() * 1.55)
    ax.grid(axis="y", visible=False)
    ax.tick_params(axis="y", labelsize=8.5)
    ax.set_xlabel(f"Bad reviews mentioning '{a}'")
    ax.set_title(f"#{rank}: {sc} | {a}\n(group avg bad rate {grp_rates[(sc, a)]:.1%})", loc="left", fontsize=11)
axes[0].set_ylabel("Product ID (parent_asin)")

fig.legend(handles=[Patch(color=ROSE, label="Bad rate above group average (95% CI)"),
                    Patch(color=GREY, label="Not different from group")],
           loc="lower center", ncol=2, frameon=False, fontsize=9.5, bbox_to_anchor=(0.5, -0.02))
fig.suptitle("2.5.3 Products to fix first for each of the top 3 issues (sample)  |  label = bad on issue | bad rate of n reviews",
             x=0.01, ha="left", fontsize=12.5, fontweight="bold", color=NAVY)
plt.tight_layout(rect=[0, 0.04, 1, 0.95]); plt.show()
````

#### 2.5.4 ร้านไหนมีรีวิวแย่มากกว่าค่าเฉลี่ย และแย่ในกลุ่มสินค้าไหน
* **ทำอะไร:** ดูภาพรวมระดับร้านว่าร้านไหนมีรีวิวแย่มากกว่าค่าเฉลี่ยของทั้งหมวด แล้วลงไปดูว่าในร้านนั้น กลุ่มสินค้าไหนแย่กว่าค่าเฉลี่ยของกลุ่มนั้น ๆ
* **ทำอย่างไร:** รวมชื่อร้านที่สะกดต่างกัน (เช่น ALOME กับ Alome) เป็นร้านเดียว เลือกร้านที่มีรีวิวอย่างน้อย 30 ฉบับ คำนวณ % รีวิวแย่พร้อมช่วงความเชื่อมั่น Wilson 95% ถ้าขอบล่างของช่วงยังสูงกว่าค่าเฉลี่ยหมวด ถือว่าแย่กว่าค่าเฉลี่ยจริง ไม่ใช่เพราะสุ่มได้น้อย
* **ทำไปเพื่ออะไร:** รู้ว่าปัญหากระจุกอยู่ที่ร้านไหน เพื่อตัดสินใจว่าจะส่งทีม Seller Success ไปคุยกับร้านใดก่อน และจุดที่ต้องดูต่อคือร้านกับกลุ่มสินค้าใด
* **ดูผลตรงไหน:** กราฟซ้ายคือ % รีวิวแย่ของแต่ละร้านเทียบเส้นค่าเฉลี่ย กราฟขวาคือ heatmap ร้านกับกลุ่มสินค้า ช่องที่มีเครื่องหมาย * คือแย่กว่าค่าเฉลี่ยของกลุ่มนั้นจริง

````python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from IPython.display import display, Markdown
from matplotlib.colors import LinearSegmentedColormap

MIN_STORE_N = 30   # ร้านต้องมีรีวิวอย่างน้อยเท่านี้จึงจะสรุป
MIN_CELL_N  = 10   # ช่อง ร้าน × กลุ่มสินค้า ต้องมีรีวิวอย่างน้อยเท่านี้
TOP_STORES  = 15   # แสดงร้านที่มีรีวิวแย่ส่วนเกินมากที่สุดกี่ร้าน
TOP_GROUPS  = 10   # แสดงกลุ่มสินค้ากี่กลุ่มใน heatmap
TEXT_C = TEXT if "TEXT" in globals() else NAVY

# ---------- ตารางมีเส้นขอบ ----------
BORDER = "1px solid #D7CCC8"
BORDER_STYLE = [
    {"selector": "th", "props": [("border", BORDER), ("padding", "6px 10px"),
                                 ("background-color", ROSE), ("color", "white"), ("text-align", "center")]},
    {"selector": "td", "props": [("border", BORDER), ("padding", "6px 10px")]},
    {"selector": "tbody tr:nth-child(even) td", "props": [("background-color", "#FDF6E3")]},
]
def bordered(sty):
    return (sty.set_table_styles(BORDER_STYLE, overwrite=True)
               .set_table_attributes('style="border-collapse:collapse; font-size:13px;"')
               .hide(axis="index"))

# ---------- เตรียมข้อมูล: รวมชื่อร้านที่สะกดต่างกัน ----------
d = df.copy()
d["store_key"] = d["store"].astype(str).str.strip().str.upper()
d = d[~d["store_key"].isin(["", "NONE", "NAN"])]

# ================= 1) ระดับร้าน =================
st = (d.groupby("store_key")
        .agg(n=("is_bad", "size"), n_bad=("is_bad", "sum"), n_products=("parent_asin", "nunique"),
             n_groups=("sub_category", "nunique"), main_group=("sub_category", lambda s: s.mode().iat[0]))
        .query("n >= @MIN_STORE_N").copy())
st["bad_rate"] = st["n_bad"] / st["n"]
st["ci_low"], st["ci_high"] = wilson_ci(st["n_bad"], st["n"])
st["vs_avg"] = np.select([st["ci_low"] > BAD_AVG, st["ci_high"] < BAD_AVG],
                         ["สูงกว่าเฉลี่ย", "ต่ำกว่าเฉลี่ย"], "ไม่ต่างจากเฉลี่ย")
st["excess_bad"] = st["n_bad"] - st["n"] * BAD_AVG          # รีวิวแย่ส่วนเกิน
st["share_of_all_neg"] = st["n_bad"] / d["is_bad"].sum()
top_st = st[st["vs_avg"] == "สูงกว่าเฉลี่ย"].sort_values("excess_bad", ascending=False).head(TOP_STORES)

# ================= 2) ร้าน × กลุ่มสินค้า =================
group_avg = d.groupby("sub_category")["is_bad"].mean()      # ค่าเฉลี่ยของแต่ละกลุ่มสินค้า (ทุกร้าน)
sub = d[d["store_key"].isin(top_st.index)]

cell = (sub.groupby(["store_key", "sub_category"])
          .agg(n=("is_bad", "size"), n_bad=("is_bad", "sum")).reset_index())
cell = cell[cell["n"] >= MIN_CELL_N].copy()
cell["bad_rate"] = cell["n_bad"] / cell["n"]
cell["group_avg"] = cell["sub_category"].map(group_avg)
cell["ci_low"], cell["ci_high"] = wilson_ci(cell["n_bad"], cell["n"])
cell["vs_group"] = np.where(cell["ci_low"] > cell["group_avg"], "แย่กว่ากลุ่มนี้", "ไม่ต่าง")

asp_share = sub[sub["is_bad"] == 1].groupby(["store_key", "sub_category"])[ASP_COLS].mean()
top_issue = pd.DataFrame({"main_issue": asp_share.idxmax(axis=1).str.replace("asp_", ""),
                          "%_bad_mention": asp_share.max(axis=1)}).reset_index()
top_issue.loc[top_issue["%_bad_mention"] == 0, "main_issue"] = "-"
cell = cell.merge(top_issue, on=["store_key", "sub_category"], how="left")

hot = cell[cell["vs_group"] == "แย่กว่ากลุ่มนี้"].assign(excess=lambda x: x["n_bad"] - x["n"] * x["group_avg"])
order_map = {s: i for i, s in enumerate(top_st.index)}
hot = hot.assign(o=hot["store_key"].map(order_map)).sort_values(["o", "excess"], ascending=[True, False]).drop(columns="o")

# ================= 3) ตาราง =================
print(f"ร้านที่มีรีวิว ≥ {MIN_STORE_N}: {len(st):,} ร้าน | สูงกว่าค่าเฉลี่ยหมวด ({BAD_AVG:.1%}) จริง: "
      f"{(st['vs_avg'] == 'สูงกว่าเฉลี่ย').sum():,} ร้าน | แสดง {len(top_st)} ร้านที่มีรีวิวแย่ส่วนเกินมากที่สุด")
display(bordered(top_st.reset_index()[["store_key", "main_group", "n_groups", "n_products", "n", "n_bad",
                                       "bad_rate", "ci_low", "excess_bad", "share_of_all_neg"]]
                 .style.format({"n": "{:,}", "bad_rate": "{:.1%}", "ci_low": "{:.1%}",
                                "excess_bad": "{:.0f}", "share_of_all_neg": "{:.1%}"})))

print(f"\nจุดที่ต้องแก้: ร้าน × กลุ่มสินค้าที่แย่กว่าค่าเฉลี่ยของกลุ่มนั้นจริง ({len(hot)} จุด จาก {hot['store_key'].nunique()} ร้าน)")
display(bordered(hot[["store_key", "sub_category", "n", "n_bad", "bad_rate", "group_avg", "main_issue", "%_bad_mention"]]
                 .style.format({"bad_rate": "{:.1%}", "group_avg": "{:.1%}", "%_bad_mention": "{:.0%}"}, na_rep="-")))

# ================= 4) สรุปเป็นข้อความ =================
lines = []
for s in top_st.index[:5]:
    h = hot[hot["store_key"] == s]
    r = top_st.loc[s]
    if len(h):
        w = h.iloc[0]
        lines.append(f"- **{s}** รีวิวแย่ {r['bad_rate']:.0%} (n={int(r['n']):,}) แย่สุดในกลุ่ม **{w['sub_category']}** "
                     f"({w['bad_rate']:.0%} เทียบค่าเฉลี่ยกลุ่ม {w['group_avg']:.0%}) ส่วนใหญ่บ่นเรื่อง **{w['main_issue']}**")
    else:
        lines.append(f"- **{s}** รีวิวแย่ {r['bad_rate']:.0%} (n={int(r['n']):,}) แต่ยังไม่พบกลุ่มที่แย่กว่าค่าเฉลี่ยชัดเจน (ข้อมูลต่อกลุ่มน้อย)")
display(Markdown(f"**สรุป 2.5.4: ร้านที่มีรีวิวแย่ส่วนเกินมากที่สุด 5 ร้าน และกลุ่มสินค้าที่ควรแก้**\n" + "\n".join(lines)))

# ================= 5) กราฟ 2 รูปข้างกัน =================
groups = (cell.groupby("sub_category")["n"].sum().sort_values(ascending=False)
              .drop("(not specified)", errors="ignore").head(TOP_GROUPS).index)
heat   = cell.pivot(index="store_key", columns="sub_category", values="bad_rate").reindex(index=top_st.index, columns=groups)
heat_n = cell.pivot(index="store_key", columns="sub_category", values="n").reindex(index=top_st.index, columns=groups)
heat_s = cell.pivot(index="store_key", columns="sub_category", values="vs_group").reindex(index=top_st.index, columns=groups)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(22, 7.5),
                               gridspec_kw={"width_ratios": [1.1, 0.35 * len(groups) + 0.6]})

# --- ซ้าย: กราฟแท่งแนวตั้ง % รีวิวแย่ของร้าน ---
p = top_st.reset_index()
x = np.arange(len(p))
ax1.bar(x, p["bad_rate"] * 100, color=ROSE, width=0.65)
ax1.errorbar(x, p["bad_rate"] * 100,
             yerr=[(p["bad_rate"] - p["ci_low"]) * 100, (p["ci_high"] - p["bad_rate"]) * 100],
             fmt="none", ecolor="#8A8A8A", elinewidth=1, capsize=3)
ax1.axhline(BAD_AVG * 100, ls="--", c=NAVY, lw=1.3)
ax1.text(len(p) - 0.5, BAD_AVG * 100 + 0.8, f"category average {BAD_AVG:.1%}", ha="right", fontsize=9, color=NAVY)
for xi, r, hi, n in zip(x, p["bad_rate"], p["ci_high"], p["n"]):
    ax1.text(xi, hi * 100 + 1, f"{r:.0%}\nn={n:,}", ha="center", va="bottom", fontsize=8)
ax1.set_xticks(x)
ax1.set_xticklabels([s[:18] for s in p["store_key"]], rotation=45, ha="right", fontsize=9)
ax1.set_ylim(0, p["ci_high"].max() * 100 * 1.25)
ax1.grid(axis="x", visible=False)
ax1.set_ylabel("% bad reviews (1-2 stars), 95% Wilson CI")
ax1.set_title(f"Stores with more bad reviews than average\n(top {len(p)} by excess bad reviews)", loc="left")

# --- ขวา: heatmap ร้าน × กลุ่มสินค้า ---
cmap = LinearSegmentedColormap.from_list("bad", ["#FFFFFF", "#F5F5DC", "#FBC02D", "#FF8F00", "#C62828"])
im = ax2.imshow((heat * 100).fillna(0).values, cmap=cmap, aspect="auto", vmin=0, vmax=max(40, heat.max().max() * 100))
for i in range(heat.shape[0]):
    for j in range(heat.shape[1]):
        v = heat.values[i, j]
        if np.isnan(v):
            continue
        star = "*" if heat_s.values[i, j] == "แย่กว่ากลุ่มนี้" else ""
        ax2.text(j, i, f"{v:.0%}{star}\n(n={int(heat_n.values[i, j])})", ha="center", va="center", fontsize=7.5,
                 fontweight="bold" if star else "normal", color="white" if v >= 0.3 else TEXT_C)
ax2.set_xticks(range(len(groups))); ax2.set_xticklabels(groups, rotation=30, ha="right", fontsize=9)
ax2.set_yticks(range(len(heat))); ax2.set_yticklabels([s[:22] for s in heat.index], fontsize=9)
ax2.grid(False)
fig.colorbar(im, ax=ax2, label="% bad reviews", format=lambda v, _: f"{v:.0f}%", fraction=0.05, pad=0.03)
ax2.set_title(f"Store x product group: % bad reviews\n(* = worse than that group's average, blank = n < {MIN_CELL_N})", loc="left")

fig.suptitle("2.5.4 Which stores get more bad reviews, and in which product group?",
             x=0.01, ha="left", fontsize=13, fontweight="bold", color=NAVY)
plt.tight_layout(rect=[0, 0, 1, 0.95]); plt.show()

top_st.to_csv("stores_bad_reviews.csv", encoding="utf-8-sig")
hot.to_csv("stores_by_group.csv", encoding="utf-8-sig", index=False)
````

#### 2.5.6 ควรแก้ที่ตัวสินค้า หรือที่ร้าน
* **ทำอะไร:** ตัดสินว่าสินค้าใน Watchlist มีรีวิวแย่เพราะ "ตัวสินค้ารุ่นนั้น" หรือเพราะ "ร้านทำสินค้าทุกชิ้นออกมาแย่เหมือนกัน"
* **ทำอย่างไร:** เทียบสินค้าชิ้นนั้นกับสินค้าอื่นในร้านเดียวกัน (ไม่นับชิ้นที่กำลังดู) ด้วย Z-test 2 ข้อ แล้วรวมผลตามตารางด้านล่าง ส่วนโค้ดแบ่งเป็น 2.5.6.1 ถึง 2.5.6.7 เรียงตามลำดับ
  - **ทดสอบ A (ร้านแย่กว่าค่าเฉลี่ยหรือไม่):** H0: p_ร้าน = p₀ H1: p_ร้าน > p₀ โดย p_ร้าน คือสัดส่วนรีวิวแย่ของสินค้าอื่นในร้านเดียวกัน และ p₀ คือสัดส่วนรีวิวแย่เฉลี่ยของ Jewelry (One-sample Z-test ทางเดียว, α = 0.05)
  - **ทดสอบ B (สินค้าแย่กว่าสินค้าอื่นของร้านหรือไม่):** H0: p_สินค้า = p_ร้าน H1: p_สินค้า > p_ร้าน โดย p_สินค้า คือสัดส่วนรีวิวแย่ของสินค้าชิ้นนี้ (Two-proportion Z-test ทางเดียว, α = 0.05)
  - ถ้าสินค้าอื่นของร้านมีรีวิวรวมไม่ถึง 30 ฉบับ ข้อมูลร้านน้อยเกินไป ให้แก้ที่ตัวสินค้าไปก่อน
* **ทำไปเพื่ออะไร:** สองกรณีนี้ใช้คนละทีมแก้ ถ้าปัญหาอยู่ที่ร้าน การโค้ชร้านครั้งเดียวช่วยได้ทุกสินค้า (Seller Success) แต่ถ้าอยู่ที่สินค้า ต้องแก้สเปกทีละรุ่น (Product)
* **ดูผลตรงไหน:** ตารางรวมผล A และ B ด้านล่าง และตารางสรุปคำตัดสินท้าย 2.5.6.5

| | B: ไม่ปฏิเสธ H0 (สินค้าไม่ได้แย่กว่าร้าน) | B: ปฏิเสธ H0 (สินค้าแย่กว่าร้าน) |
|---|---|---|
| **A: ปฏิเสธ H0 (ร้านแย่)** | **แก้ที่ร้าน** โค้ชทั้งร้าน | **แก้ทั้งร้านและสินค้า** |
| **A: ไม่ปฏิเสธ H0 (ร้านปกติ)** | **ยังสรุปไม่ได้** ติดตามต่อ | **แก้ที่สินค้า** แก้สเปกเฉพาะรุ่น |

##### 2.5.6.1 ตั้งค่าและฟังก์ชัน
* **ทำอะไร:** กำหนด α = 0.05 และเขียนฟังก์ชัน Z-test ที่ใช้ซ้ำในข้อ 2.5.6 และ 2.6
* **ทำอย่างไร:** `one_prop_ztest` ใช้ทดสอบ A (เทียบกับค่าคงที่ p₀) ส่วน `two_prop_ztest` ใช้ทดสอบ B (เทียบสองกลุ่ม) จากนั้นสร้างคอลัมน์ `store_key` เพื่อรวมชื่อร้านที่สะกดต่างกันแค่ตัวพิมพ์เล็ก/ใหญ่
* **ทำไปเพื่ออะไร:** ให้ทุกการทดสอบใช้สูตรและเกณฑ์เดียวกัน และไม่แยกร้านเดียวกันเป็นสองร้านเพราะการสะกด
* **ดูผลตรงไหน:** บรรทัดสุดท้ายที่พิมพ์ว่า α = 0.05 และค่าวิกฤต Z = 1.645

````python
from scipy.stats import norm
from IPython.display import display, Markdown
from matplotlib.patches import Patch
from matplotlib.lines import Line2D

ALPHA          = 0.05
Z_CRIT         = norm.ppf(1 - ALPHA)   # ค่าวิกฤตทางเดียว = 1.645
MIN_STORE_N    = 30                    # สินค้าอื่นของร้านต้องมีรีวิวรวมอย่างน้อยเท่านี้จึงจะทดสอบ
TOP_PER_ISSUE  = 10                    # ดู Top 10 สินค้าของแต่ละเรื่องจาก 2.5.3

def one_prop_ztest(k, n, p0):
    # One-sample Z-test ทางเดียว  H0: p = p0  vs  H1: p > p0
    z = (k / n - p0) / np.sqrt(p0 * (1 - p0) / n)
    return z, 1 - norm.cdf(z)

def two_prop_ztest(k1, n1, k2, n2):
    # Two-proportion Z-test ทางเดียว  H0: p1 = p2  vs  H1: p1 > p2
    p1, p2 = k1 / n1, k2 / n2
    p_pool = (k1 + k2) / (n1 + n2)                               # สัดส่วนรวมของสองกลุ่ม
    se = np.sqrt(p_pool * (1 - p_pool) * (1 / n1 + 1 / n2))      # ความคลาดเคลื่อนมาตรฐาน
    z = (p1 - p2) / se if se > 0 else 0.0
    return z, 1 - norm.cdf(z)

# ชื่อร้านสะกดไม่เหมือนกัน เช่น "ALOME PIERCINGS" กับ "Alome Piercings" → แปลงเป็นตัวพิมพ์เล็กก่อนจับกลุ่ม
full["store_key"] = full["store"].fillna("").astype(str).str.strip().str.lower()
full.loc[full["store_key"].isin(["none", "nan"]), "store_key"] = ""    # ไม่มีชื่อร้าน
print("ตั้งค่าเสร็จ | α =", ALPHA, "| ค่าวิกฤต Z =", round(Z_CRIT, 3))
````

##### 2.5.6.2 สร้าง Watchlist
* **ทำอะไร:** รวบรวมสินค้าที่ต้องตรวจว่าควรแก้ที่ร้านหรือที่สินค้า
* **ทำอย่างไร:** จากตาราง 2.5.3 เอา Top 10 ของแต่ละเรื่อง เก็บเฉพาะสินค้าที่ % รีวิวแย่สูงกว่ากลุ่มจริง (Wilson CI ล่างสูงกว่าค่าเฉลี่ย) ถ้าสินค้าเดียวติดหลายเรื่อง ใช้เรื่องที่มีรีวิวแย่มากที่สุด แล้วหาค่าเฉลี่ยกลุ่ม p₀
* **ทำไปเพื่ออะไร:** ไม่ต้องทดสอบสินค้าทั้งหมด ทดสอบเฉพาะที่น่าสงสัยจริง และได้ p₀ ที่ใช้เป็นเกณฑ์ในทดสอบ A
* **ดูผลตรงไหน:** บรรทัดที่บอกจำนวนสินค้าใน Watchlist จำนวนร้าน และค่า p₀

````python
wl_rows = []
for (sc, a), t in asin_rank.items():
    top = t.head(TOP_PER_ISSUE)
    top = top[top["vs_group"] == "สูงกว่ากลุ่ม"]      # เก็บเฉพาะสินค้าที่แย่กว่ากลุ่มจริง (Wilson CI)
    for _, r in top.iterrows():
        wl_rows.append({"parent_asin": r["parent_asin"], "store": r["store"], "sub_category": sc,
                        "main_issue": a, "n": r["n"], "n_bad": r["n_bad"],
                        "n_bad_issue": r["n_bad_issue"], "%_bad": r["bad_rate"] * 100})

wl = (pd.DataFrame(wl_rows)
        .sort_values("n_bad_issue", ascending=False)
        .drop_duplicates("parent_asin")              # สินค้าที่ติดหลายเรื่อง ใช้เรื่องที่มีรีวิวแย่มากที่สุด
        .set_index("parent_asin"))
wl["store_key"] = wl["store"].fillna("").astype(str).str.strip().str.lower()
wl.loc[wl["store_key"].isin(["none", "nan"]), "store_key"] = ""

P0 = full["is_bad"].mean()                           # ค่าเฉลี่ยรีวิวแย่ของกลุ่มเป้าหมาย (Jewelry)
print(f"Watchlist = {len(wl)} สินค้า จาก {wl['store_key'].nunique()} ร้าน | ค่าเฉลี่ยกลุ่ม p0 = {P0:.1%}")
````

##### 2.5.6.3 ทดสอบ A ร้านแย่กว่าค่าเฉลี่ยหรือไม่
* **ทำอะไร:** ทดสอบว่าสินค้าอื่นของร้านเดียวกัน มี % รีวิวแย่สูงกว่าค่าเฉลี่ยกลุ่ม p₀ หรือไม่
* **ทำอย่างไร:** ต่อสินค้า 1 ชิ้นใน Watchlist ดึงรีวิวของสินค้าอื่นในร้านเดียวกันโดยตัดชิ้นนั้นออก (leave-one-out) แล้วใช้ One-sample Z-test ทางเดียว ถ้าสินค้าอื่นมีรีวิวไม่ถึง 30 ฉบับ จะไม่ทดสอบ
* **ทำไปเพื่ออะไร:** ตัดชิ้นที่กำลังสงสัยออกก่อน เพื่อไม่ให้สินค้าเสียชิ้นเดียวทำให้ร้านดูแย่ไปด้วย
* **ดูผลตรงไหน:** คอลัมน์ `A` (ร้านแย่ / ร้านปกติ / ข้อมูลไม่พอ) และ `z_A` เทียบกับ 1.645

````python
rows = []
for asin, r in wl.iterrows():
    # สินค้าอื่นของร้านเดียวกัน โดยไม่นับสินค้าชิ้นนี้ (leave-one-out)
    others = full[(full["store_key"] == r["store_key"]) & (full["parent_asin"] != asin)]
    if r["store_key"] == "":
        others = others.iloc[0:0]                    # ไม่รู้ชื่อร้าน → ทดสอบไม่ได้
    n_o, k_o = len(others), int(others["is_bad"].sum())

    if n_o >= MIN_STORE_N:
        zA, pA = one_prop_ztest(k_o, n_o, P0)
        A = "ร้านแย่" if zA > Z_CRIT else "ร้านปกติ"
    else:
        zA, pA, A = np.nan, np.nan, "ข้อมูลไม่พอ"

    rows.append({"parent_asin": asin, "store": r["store"], "store_key": r["store_key"], "main_issue": r["main_issue"],
                 "n_product": r["n"], "n_bad_product": r["n_bad"], "%_bad_product": r["%_bad"],
                 "n_other": n_o, "n_bad_other": k_o, "%_bad_other": k_o / n_o * 100 if n_o else np.nan,
                 "z_A": zA, "p_A": pA, "A": A})
res = pd.DataFrame(rows).set_index("parent_asin")

print(f"ทดสอบ A  H0: p_ร้าน = {P0:.3f}  vs  H1: p_ร้าน > {P0:.3f}  (One-sample Z-test, α = {ALPHA})")
print("p_ร้าน = สัดส่วนรีวิวแย่ของสินค้าอื่นในร้านเดียวกัน\n")
display(bordered(res.reset_index()[["parent_asin", "store", "n_other", "%_bad_other", "z_A", "A"]]
                 .style.format({"%_bad_other": "{:.1f}%", "z_A": "{:.2f}"}, na_rep="-")))
````

##### 2.5.6.4 ทดสอบ B สินค้าแย่กว่าสินค้าอื่นของร้านหรือไม่
* **ทำอะไร:** ทดสอบว่าสินค้าชิ้นนี้ มี % รีวิวแย่สูงกว่าสินค้าอื่นของร้านตัวเองหรือไม่
* **ทำอย่างไร:** ใช้ Two-proportion Z-test ทางเดียว เทียบรีวิวของสินค้าชิ้นนี้กับรีวิวสินค้าอื่นของร้าน ร้านที่ข้อมูลไม่พอจากเซลล์ 3 จะข้ามไป
* **ทำไปเพื่ออะไร:** แยกให้ออกว่าสินค้าชิ้นนี้แย่เพราะตัวมันเอง หรือแย่พอ ๆ กับทุกชิ้นในร้าน
* **ดูผลตรงไหน:** คอลัมน์ `B` และ `z_B` เทียบกับ 1.645

````python
zB_list, pB_list, B_list = [], [], []
for asin, r in res.iterrows():
    if r["A"] == "ข้อมูลไม่พอ":
        zB_list.append(np.nan); pB_list.append(np.nan); B_list.append("ข้อมูลไม่พอ")
        continue
    zB, pB = two_prop_ztest(r["n_bad_product"], r["n_product"], r["n_bad_other"], r["n_other"])
    zB_list.append(zB); pB_list.append(pB)
    B_list.append("สินค้าแย่กว่าร้าน" if zB > Z_CRIT else "ไม่ต่างจากร้าน")
res["z_B"], res["p_B"], res["B"] = zB_list, pB_list, B_list

print("ทดสอบ B  H0: p_สินค้า = p_ร้าน  vs  H1: p_สินค้า > p_ร้าน  (Two-proportion Z-test, α = 0.05)")
print("p_สินค้า = สัดส่วนรีวิวแย่ของสินค้าชิ้นนี้ | p_ร้าน = สัดส่วนรีวิวแย่ของสินค้าอื่นในร้านเดียวกัน\n")
display(bordered(res.reset_index()[["parent_asin", "store", "%_bad_product", "%_bad_other", "z_B", "B"]]
                 .style.format({"%_bad_product": "{:.1f}%", "%_bad_other": "{:.1f}%", "z_B": "{:.2f}"}, na_rep="-")))
````

##### 2.5.6.5 รวมผล A และ B เป็นคำตัดสิน
* **ทำอะไร:** ตัดสินสินค้าแต่ละชิ้นว่าควร "แก้ที่ร้าน" "แก้ที่สินค้า" "แก้ทั้งคู่" หรือ "ยังสรุปไม่ได้"
* **ทำอย่างไร:** เอาผล A และ B มาเทียบกับตาราง 2×2 ด้านบน แล้วนับจำนวนสินค้าในแต่ละคำตัดสิน และรวมรายชื่อร้านที่ควรโค้ชทั้งร้าน
* **ทำไปเพื่ออะไร:** แปลงผลสถิติเป็นการตัดสินใจว่าจะส่งงานให้ทีมไหน (Seller Success หรือ Product)
* **ดูผลตรงไหน:** ตารางสรุปรายสินค้า ตารางจำนวนสินค้าต่อคำตัดสิน และตาราง "ร้านที่ควรโค้ชทั้งร้าน"

````python
def decide(A, B):
    if A == "ข้อมูลไม่พอ":                              return "แก้ที่สินค้า (ข้อมูลร้านไม่พอ)"
    if A == "ร้านแย่"  and B == "ไม่ต่างจากร้าน":         return "แก้ที่ร้าน"
    if A == "ร้านแย่"  and B == "สินค้าแย่กว่าร้าน":      return "แก้ทั้งร้านและสินค้า"
    if A == "ร้านปกติ" and B == "สินค้าแย่กว่าร้าน":      return "แก้ที่สินค้า"
    return "ยังสรุปไม่ได้"                              # ร้านปกติ และสินค้าไม่ได้แย่กว่าร้าน

res["decision"] = [decide(a, b) for a, b in zip(res["A"], res["B"])]

ORDER = ["แก้ที่ร้าน", "แก้ทั้งร้านและสินค้า", "แก้ที่สินค้า", "แก้ที่สินค้า (ข้อมูลร้านไม่พอ)", "ยังสรุปไม่ได้"]
res["_o"] = res["decision"].map({d_: i for i, d_ in enumerate(ORDER)})
show = res.sort_values(["_o", "%_bad_other"], ascending=[True, False]).reset_index()
display(bordered(show[["parent_asin", "store", "main_issue", "%_bad_product", "%_bad_other",
                       "z_A", "A", "z_B", "B", "decision"]]
                 .style.format({"%_bad_product": "{:.0f}%", "%_bad_other": "{:.0f}%",
                                "z_A": "{:.2f}", "z_B": "{:.2f}"}, na_rep="-")))

level_counts = res["decision"].value_counts().reindex(ORDER).fillna(0).astype(int)
store_fix = (res[res["decision"].isin(["แก้ที่ร้าน", "แก้ทั้งร้านและสินค้า"])]
               .groupby("store_key")
               .agg(store=("store", "first"), watchlist_products=("main_issue", "size"),
                    main_issues=("main_issue", lambda s: ", ".join(sorted(set(s)))),
                    pct_bad_other=("%_bad_other", "max"))
               .sort_values("pct_bad_other", ascending=False))
print("\nจำนวนสินค้าในแต่ละคำตัดสิน"); display(level_counts.to_frame("จำนวนสินค้า"))
print("ร้านที่ควรโค้ชทั้งร้าน"); display(bordered(store_fix.style.format({"pct_bad_other": "{:.0f}%"})))
````

##### 2.5.6.6 กราฟ สินค้านี้เทียบกับสินค้าอื่นของร้าน
* **ทำอะไร:** วาดกราฟให้เห็นว่าสินค้าแต่ละชิ้นแย่กว่าสินค้าอื่นของร้านตัวเองหรือไม่
* **ทำอย่างไร:** แท่งสีคือ % รีวิวแย่ของสินค้าชิ้นนั้น (สีตามคำตัดสิน) แท่งเทาคือ % รีวิวแย่ของสินค้าอื่นในร้าน เส้นประคือค่าเฉลี่ยกลุ่ม
* **ทำไปเพื่ออะไร:** ใช้ประกอบการนำเสนอ ให้เห็นภาพว่า ร้านที่แท่งเทาสูง = ปัญหาทั้งร้าน ส่วนร้านที่แท่งเทาต่ำแต่แท่งสีสูง = ปัญหาเฉพาะรุ่น
* **ดูผลตรงไหน:** เทียบแท่งสีกับแท่งเทาของสินค้าแต่ละแถว ตัวเลข n ของสินค้าอื่นอยู่ท้ายแท่งเทา

````python
DEC_EN    = {"แก้ที่ร้าน": "Fix STORE", "แก้ทั้งร้านและสินค้า": "Fix STORE + PRODUCT",
             "แก้ที่สินค้า": "Fix PRODUCT", "แก้ที่สินค้า (ข้อมูลร้านไม่พอ)": "Fix PRODUCT (little store data)",
             "ยังสรุปไม่ได้": "Inconclusive"}
DEC_COLOR = {"แก้ที่ร้าน": RED, "แก้ทั้งร้านและสินค้า": ORANGE, "แก้ที่สินค้า": YELLOW,
             "แก้ที่สินค้า (ข้อมูลร้านไม่พอ)": "#D6D6D6", "ยังสรุปไม่ได้": "#EFEFEF"}

v = show.iloc[::-1].reset_index(drop=True)             # อันดับแรกอยู่บนสุด
y, h = np.arange(len(v)), 0.38
fig, ax = plt.subplots(figsize=(11, 0.55 * len(v) + 2))
ax.barh(y + h/2, v["%_bad_product"], height=h, color=v["decision"].map(DEC_COLOR))   # สินค้านี้
ax.barh(y - h/2, v["%_bad_other"].fillna(0), height=h, color="#BDBDBD")               # สินค้าอื่นของร้าน
for yi, a_, b_, n_o in zip(y, v["%_bad_product"], v["%_bad_other"], v["n_other"]):
    ax.text(a_ + 0.8, yi + h/2, f"{a_:.0f}%", va="center", fontsize=8)
    ax.text((0 if pd.isna(b_) else b_) + 0.8, yi - h/2,
            "store data too small" if pd.isna(b_) or n_o < MIN_STORE_N else f"{b_:.0f}% (other products, n={n_o})",
            va="center", fontsize=8, color="#555555")
ax.axvline(P0 * 100, ls="--", c=NAVY, lw=1.2)
ax.set_yticks(y, [f"{a}  ({str(s)[:18]})" for a, s in zip(v["parent_asin"], v["store"])], fontsize=9)
ax.set_xlabel("% bad reviews (1-2 stars)")
ax.set_title("Fix the product or the store?  This product vs the store's OTHER products", loc="left")
ax.grid(axis="y", visible=False)
ax.legend(handles=[Patch(color=DEC_COLOR[d_], label=f"This product: {DEC_EN[d_]}") for d_ in ORDER if d_ in set(v["decision"])]
                  + [Patch(color="#BDBDBD", label="Store's other products"),
                     Line2D([0], [0], color=NAVY, ls="--", label=f"Group average {P0:.1%}")],
          loc="lower right", fontsize=8.5, facecolor="white")
plt.tight_layout(); plt.show()
````

### 2.6 Reality Check

ก่อนเอา insight ไปบอกลูกค้า ให้ลองหักล้างมันด้วยตัวเองก่อน อย่างน้อย 1 insight ต้องผ่านขั้นนี้

คำอธิบายทางเลือกที่พบบ่อย
* **ผลของขนาด:** กลุ่มที่ดูโดดเด่นอาจมีตัวอย่างน้อยมาก ตรวจ n ของทุกกลุ่ม
* **ผลของส่วนผสม:** ความต่างอาจมาจากสัดส่วนสินค้าในแต่ละกลุ่มไม่เหมือนกัน ไม่ใช่จากพฤติกรรมลูกค้า ลองคุมให้เทียบกันเฉพาะสินค้าประเภทเดียวกัน
* **ผลของเวลา:** ของเก่ามีเวลาสะสม vote และรีวิวมากกว่า
* **ผลของความยาว:** ข้อความยาวมีโอกาสมีคำที่เราค้นหามากกว่าโดยอัตโนมัติ
* **ผลของคนไม่กี่คน:** ผู้ใช้หรือสินค้าไม่กี่รายอาจครองสัดส่วนใหญ่ของกลุ่ม

เลือกอย่างน้อยหนึ่งคำอธิบาย แล้ว **ทดสอบด้วยตัวเลขจริง** สรุปว่า insight ยังยืนอยู่ได้ หรือต้องปรับถ้อยคำให้แคบลง

#### แผนการหักล้าง insight ของทีม

ก่อนนำ insight ไปเสนอลูกค้า ทีมลองหักล้างมันด้วยคำอธิบายทางเลือกทีละข้อ ถ้าหักล้างไม่ได้ แปลว่า insight ยังยืนอยู่ได้ ถ้าหักล้างได้ ต้องเขียน insight ใหม่ให้แคบลงเฉพาะส่วนที่ยังจริง

**Insight 1 ที่ทดสอบ (จาก 2.5):** ในกลุ่ม Jewelry รีวิวที่พูดถึงปัญหาของพัง (durability) เป็นรีวิวแย่ (1-2 ดาว) 50.1% ส่วนรีวิวที่ไม่พูดถึงเป็นรีวิวแย่ 9.5%

**ทุกข้อ T1-T5 ใช้การทดสอบแบบเดียวกัน**
* p₁ คือสัดส่วนรีวิวแย่ของรีวิว Jewelry ที่พูดถึงของพัง และ p₂ คือสัดส่วนรีวิวแย่ของรีวิว Jewelry ที่ไม่พูดถึงของพัง
* **H0:** p₁ = p₂ **H1:** p₁ > p₂
* สถิติทดสอบคือ Two-proportion Z-test ทางเดียว ที่ α = 0.05 ค่าวิกฤต Z = 1.645 สูตร Z = (p̂₁ − p̂₂) / √[ p̂(1 − p̂)(1/n₁ + 1/n₂) ] โดย p̂ คือสัดส่วนรีวิวแย่รวมของสองกลุ่ม
* ถ้า Z > 1.645 (หรือ p-value < 0.05) ให้ปฏิเสธ H0 หมายความว่าสองกลุ่มต่างกันจริง
* ทุกครั้งรายงานความต่างเป็น percentage point (pp) ด้วย เพราะข้อมูลมีจำนวนมาก ความต่างเล็กนิดเดียวก็ผ่านการทดสอบได้ ต้องดูว่าต่างมากแค่ไหนด้วย

แต่ละข้อต่างกันที่ "ชุดข้อมูลที่ใช้เทียบ" ซึ่งเป็นตัวควบคุมคำอธิบายทางเลือกนั้น

| ข้อ | คำอธิบายทางเลือกที่ลองหักล้าง | ข้อมูลที่ใช้เทียบ | ถ้าไม่ปฏิเสธ H0 ต้องปรับ insight เป็น |
|---|---|---|---|
| **T1** | ผลของขนาด: ความต่างเกิดจากความบังเอิญของการสุ่ม | รีวิว Jewelry ทั้งหมด | ไม่ควรนำเสนอ insight นี้ |
| **T2a / T2b** | ผลของความยาว: รีวิวลบยาวกว่า จึงมีโอกาสติดคำว่าพังมากกว่า | เฉพาะรีวิวสั้น (≤ median) และเฉพาะรีวิวยาว (> median) | ระบุว่าชัดเฉพาะช่วงที่ผ่าน |
| **T3** | ผลของสินค้าไม่กี่ตัว: ปัญหากระจุกในสินค้า 5 ตัวที่แย่มาก | ตัด 5 สินค้าที่มีรีวิวแย่เรื่องนี้มากที่สุดออก | "เฉพาะสินค้า 5 ตัวนี้" ไม่ใช่ทั้งกลุ่ม |
| **T4** | ผลของส่วนผสม: บังเอิญพูดถึงของพังในสินค้าประเภทที่แย่อยู่แล้ว | ทดสอบแยกตามประเภทสินค้า (Necklaces / Earrings / Bracelets ฯลฯ) | เฉพาะประเภทที่ผ่าน |
| **T5** | ผลของเวลา: เป็นปัญหาในอดีตที่ร้านแก้ไปแล้ว | ทดสอบแยกช่วงแรก และช่วงหลัง (ตัดที่ median ของปีรีวิว) | เฉพาะช่วงเวลาที่ผ่าน |

**Insight 2 (ระดับสินค้า) ทดสอบด้วย T6**
* p₁ คือสัดส่วนรีวิวแย่ของสินค้ารหัสนั้น และ p₂ คือสัดส่วนรีวิวแย่ของสินค้าอื่นทั้งหมดใน Jewelry
* **H0:** p₁ = p₂ **H1:** p₁ > p₂ ทดสอบทีละสินค้า และเพราะทดสอบหลายตัวพร้อมกัน จึงปรับ p-value ด้วยวิธี Holm เพื่อไม่ให้ผ่านเพราะโชคดี
* หักล้างคำอธิบายว่า "สินค้านี้ติดอันดับเพราะมีรีวิวเยอะ ไม่ได้แย่กว่าสินค้าอื่นจริง" ถ้าไม่ปฏิเสธ H0 จะตัดสินค้านั้นออกจาก Card 2

**วิธีสรุป:** ถ้า T1-T5 ปฏิเสธ H0 ครบ Insight 1 ยืนอยู่ได้ ถ้าข้อไหนไม่ปฏิเสธ ให้ปรับถ้อยคำตามคอลัมน์ขวาสุดของตาราง ส่วนคำถามว่า "แก้ที่ร้านหรือที่สินค้า" ตอบไปแล้วใน 2.5.6

#### 2.6.0 ตั้งค่าการทดสอบ
* **ทำอะไร:** เลือก insight ที่จะทดสอบ (ปัญหาอันดับ 1 จาก 2.5 คือ Jewelry × durability) แล้วเตรียมฟังก์ชันที่ T1-T6 ใช้ร่วมกัน
* **ทำอย่างไร:** แบ่งรีวิวของ Jewelry เป็นกลุ่มที่พูดถึงของพัง (m) กับกลุ่มที่ไม่พูดถึง (~m) เขียน `run_test` ให้รับสองกลุ่มมานับรีวิวแย่ ทดสอบ และพิมพ์ผล พร้อมเก็บผลลงรายการ `RESULTS` นอกจากนี้ดึงประเภทสินค้า (Necklaces, Earrings, Bracelets) จาก categories เพื่อใช้ใน T4 และเตรียมคอลัมน์ปีไว้ใช้ใน T5
* **ทำไปเพื่ออะไร:** ให้ทุกข้อ T1-T6 ใช้สูตรและเกณฑ์เดียวกัน ผลเทียบกันได้ และไม่ทดสอบกลุ่มที่มีรีวิวน้อยกว่า 30 ฉบับเพราะสรุปไม่ได้
* **ดูผลตรงไหน:** บรรทัดสุดท้ายที่บอกจำนวนรีวิวของ Jewelry และจำนวนที่พูดถึง / ไม่พูดถึง ทั้งสองกลุ่มต้องมีมากกว่า 30

````python
from scipy.stats import norm

ALPHA      = 0.05
Z_CRIT     = norm.ppf(1 - ALPHA)   # ค่าวิกฤตทางเดียว = 1.645
MIN_N_TEST = 30                    # แต่ละกลุ่มต้องมีอย่างน้อยเท่านี้จึงจะทดสอบ

# ---------- เลือก insight ที่จะทดสอบ (เปลี่ยนเป็น TOP3[1] หรือ TOP3[2] ได้) ----------
SC, ASP = TOP3[0]                             # Jewelry | durability
COL = f"asp_{ASP}"
d = full[full["sub_category"] == SC].copy()   # รีวิวของกลุ่มสินค้านี้
m = d[COL] == 1                               # True = รีวิวพูดถึงปัญหา
RESULTS = []                                  # เก็บผลทุกการทดสอบไว้สรุปท้าย

# ---------- ประเภทสินค้า (ระดับถัดจาก sub_category) ใช้ใน T4 ----------
TYPE_LEVEL = GROUP_LEVEL + 1
type_map = meta.set_index("parent_asin")["categories"].apply(
    lambda c: c[TYPE_LEVEL] if isinstance(c, (list, np.ndarray)) and len(c) > TYPE_LEVEL else "(not specified)")
d["product_type"] = d["parent_asin"].map(type_map).fillna("(not specified)")

# ---------- ปีของรีวิว ใช้ใน T5 ----------
if "year" not in d:
    d["year"] = pd.to_datetime(d["timestamp"], unit="ms").dt.year

# ---------- ฟังก์ชัน ----------
def fmt_p(v):
    # p-value ที่เล็กมากแสดงเป็น "< 0.001" แทน 0
    if pd.isna(v):
        return "-"
    return "< 0.001" if v < 0.001 else f"{v:.3f}"

def run_test(data, mask_a, mask_b, test, alternative, h0, h1, label_a, label_b):
    # แบ่งกลุ่ม A/B → นับ k, n → Z-test → print ผล และเก็บลง RESULTS
    a, b = data[mask_a], data[mask_b]
    k1, n1, k2, n2 = int(a["is_bad"].sum()), len(a), int(b["is_bad"].sum()), len(b)
    print(f"H0: {h0}\nH1: {h1}")
    print(f"สถิติทดสอบ: Two-proportion Z-test ทางเดียว | α = {ALPHA} | ค่าวิกฤต Z = {Z_CRIT:.3f}\n")
    row = {"การทดสอบ": test, "หักล้างผลของ": alternative, "H0": h0, "H1": h1, "n1": n1, "n2": n2}
    if min(n1, n2) < MIN_N_TEST:
        print(f"กลุ่ม A n = {n1:,} | กลุ่ม B n = {n2:,} → n น้อยกว่า {MIN_N_TEST} ไม่ทดสอบ")
        row["ตัดสินใจ"] = "ไม่ทดสอบ"
        RESULTS.append(row); return row
    z, p_value = two_prop_ztest(k1, n1, k2, n2)
    p1, p2 = k1 / n1, k2 / n2
    decision = "ปฏิเสธ H0" if z > Z_CRIT else "ไม่ปฏิเสธ H0"
    print(f"กลุ่ม A ({label_a}): รีวิวลบ {k1:,} / {n1:,} = p̂₁ = {p1:.2%}")
    print(f"กลุ่ม B ({label_b}): รีวิวลบ {k2:,} / {n2:,} = p̂₂ = {p2:.2%}")
    print(f"ต่างกัน = {(p1 - p2) * 100:+.2f} percentage point")
    print(f"Z = {z:.2f} {'>' if z > Z_CRIT else '≤'} {Z_CRIT:.3f} | p-value = {fmt_p(p_value)}")
    print(f"ผล: {decision}")
    row.update({"p̂1": p1, "p̂2": p2, "ต่างกัน (pp)": (p1 - p2) * 100, "Z": z, "p-value": p_value, "ตัดสินใจ": decision})
    RESULTS.append(row); return row

def check_all(prefix):
    # ดูผลของทุกการทดสอบที่ชื่อขึ้นต้นด้วย prefix → (ผ่านทุกข้อไหม, ชื่อข้อที่ไม่ผ่าน, จำนวนที่ทดสอบ)
    rows = [r for r in RESULTS if r["การทดสอบ"].startswith(prefix) and r.get("ตัดสินใจ") in ("ปฏิเสธ H0", "ไม่ปฏิเสธ H0")]
    failed = [r["การทดสอบ"] for r in rows if r["ตัดสินใจ"] == "ไม่ปฏิเสธ H0"]
    return len(rows) > 0 and not failed, failed, len(rows)

print(f"Insight ที่ทดสอบ: {SC} | {ASP} | รีวิว {len(d):,} ฉบับ (พูดถึง {m.sum():,} / ไม่พูดถึง {(~m).sum():,})")
````

#### T1 ผลของขนาด / ความบังเอิญ
* **ทำอะไร:** ตรวจว่าที่รีวิวพูดถึงของพังเป็นรีวิวแย่มากกว่ารีวิวอื่น เกิดจากความบังเอิญของการสุ่มหรือไม่
* **ทำอย่างไร:** Two-proportion Z-test ทางเดียว เทียบรีวิว Jewelry ที่พูดถึงของพัง (กลุ่ม A) กับที่ไม่พูดถึง (กลุ่ม B)
  - **H0:** p₁ = p₂ **H1:** p₁ > p₂
  - p₁ = สัดส่วนรีวิวแย่ของรีวิว Jewelry ที่พูดถึงของพัง, p₂ = สัดส่วนรีวิวแย่ของรีวิว Jewelry ที่ไม่พูดถึงของพัง
* **ทำไปเพื่ออะไร:** ถ้าความต่างอาจเกิดจากโชค ไม่ควรเอา insight นี้ไปเสนอลูกค้า
* **ดูผลตรงไหน:** บรรทัด Z กับ p-value ถ้า Z > 1.645 ปฏิเสธ H0 แปลว่าความต่างมีจริง และดูความต่างเป็น pp ว่ามากแค่ไหน

````python
r = run_test(d, m, ~m, "T1", "ขนาด/ความบังเอิญ",
             h0=f"% ลบของรีวิวที่พูดถึง {ASP} = % ลบของรีวิวที่ไม่พูดถึง (p₁ = p₂)",
             h1=f"% ลบของรีวิวที่พูดถึง {ASP} > % ลบของรีวิวที่ไม่พูดถึง (p₁ > p₂)",
             label_a=f"พูดถึง {ASP}", label_b=f"ไม่พูดถึง {ASP}")
print("\nสรุป:", "ความต่างไม่ใช่ความบังเอิญ insight ผ่านข้อนี้" if r.get("ตัดสินใจ") == "ปฏิเสธ H0"
      else "ยังสรุปไม่ได้ว่าต่างจริง insight ไม่ควรนำเสนอ")
````

#### T2 ผลของความยาวรีวิว
* **ทำอะไร:** ตรวจว่ารีวิวแย่ยาวกว่า (ดู EDA 3) จึงมีโอกาสติดคำที่เราค้นหามากกว่าโดยอัตโนมัติ หรือผลใน T1 มาจากความยาวจริงหรือไม่
* **ทำอย่างไร:** แบ่งรีวิวด้วย median จำนวนคำเป็นรีวิวสั้นและรีวิวยาว แล้วทดสอบแยกทีละช่วง (เทียบกันเฉพาะรีวิวที่ยาวใกล้เคียงกัน)
  - **T2a (เฉพาะรีวิวสั้น)** H0: p₁ = p₂ H1: p₁ > p₂
  - **T2b (เฉพาะรีวิวยาว)** H0: p₁ = p₂ H1: p₁ > p₂
  - p₁, p₂ นิยามเหมือน T1 แต่ใช้เฉพาะรีวิวในช่วงความยาวนั้น
* **ทำไปเพื่ออะไร:** ถ้ายังต่างจริงในทั้งสองช่วง แสดงว่าความยาวไม่ใช่ตัวอธิบาย
* **ดูผลตรงไหน:** Z และ pp ของทั้ง T2a และ T2b ถ้าปฏิเสธ H0 ทั้งคู่ insight ผ่านข้อนี้ ถ้าช่วงใดไม่ผ่านให้ระบุ insight เฉพาะช่วงที่ผ่าน

````python
med = d["n_words"].median()
short, long_ = d["n_words"] <= med, d["n_words"] > med
print(f"median จำนวนคำ = {med:.0f} คำ\n")

print("----- T2a รีวิวสั้น -----")
r2a = run_test(d, m & short, ~m & short, "T2a รีวิวสั้น", "ความยาว",
               h0=f"เฉพาะรีวิว ≤ {med:.0f} คำ: p₁ = p₂", h1=f"เฉพาะรีวิว ≤ {med:.0f} คำ: p₁ > p₂",
               label_a=f"สั้น + พูดถึง {ASP}", label_b="สั้น + ไม่พูดถึง")

print("\n----- T2b รีวิวยาว -----")
r2b = run_test(d, m & long_, ~m & long_, "T2b รีวิวยาว", "ความยาว",
               h0=f"เฉพาะรีวิว > {med:.0f} คำ: p₁ = p₂", h1=f"เฉพาะรีวิว > {med:.0f} คำ: p₁ > p₂",
               label_a=f"ยาว + พูดถึง {ASP}", label_b="ยาว + ไม่พูดถึง")

ok, failed, _ = check_all("T2")
print("\nสรุป:", "ปฏิเสธ H0 ทั้งสองช่วง → ผลไม่ได้เกิดจากความยาวรีวิว insight ผ่านข้อนี้" if ok
      else f"ไม่ปฏิเสธ H0 ที่ {', '.join(failed)} → ความยาวมีผลบางส่วน ต้องระบุว่าชัดเฉพาะช่วงที่ปฏิเสธ H0")
````

#### T3 ผลของสินค้าไม่กี่ตัว
* **ทำอะไร:** ตรวจว่าผลใน T1 มาจากสินค้าไม่กี่ตัวที่แย่มาก หรือเป็นปัญหาของทั้งกลุ่ม
* **ทำอย่างไร:** ตัด 5 สินค้าที่มีรีวิวแย่เรื่องนี้มากที่สุดออกจากข้อมูล แล้วทดสอบใหม่ด้วย Two-proportion Z-test
  - **H0:** หลังตัด 5 สินค้าออก p₁ = p₂ **H1:** หลังตัด 5 สินค้าออก p₁ > p₂
* **ทำไปเพื่ออะไร:** ถ้าตัดออกแล้วยังต่างจริง ปัญหาไม่ได้กระจุกที่ 5 ตัวนี้ ควรแก้ระดับกลุ่ม ถ้าไม่ต่างแล้ว ปัญหาอยู่ที่ 5 ตัวนี้เท่านั้น
* **ดูผลตรงไหน:** บรรทัดที่บอกว่า 5 สินค้าครองรีวิวลบเรื่องนี้กี่ % และ Z กับ pp หลังตัดออก

````python
bad_issue  = d[m & (d["is_bad"] == 1)]
top5       = bad_issue["parent_asin"].value_counts().head(5)
share_top5 = top5.sum() / len(bad_issue) if len(bad_issue) else 0
print(f"5 สินค้าที่ตัดออก: {', '.join(top5.index)}")
print(f"ครองรีวิวลบเรื่อง {ASP} {top5.sum():,} จาก {len(bad_issue):,} = {share_top5:.1%}\n")

d3 = d[~d["parent_asin"].isin(top5.index)]
m3 = d3[COL] == 1
r3 = run_test(d3, m3, ~m3, "T3 ตัด top-5 สินค้า", "สินค้าไม่กี่ตัว",
              h0="หลังตัด 5 สินค้าออก: p₁ = p₂", h1="หลังตัด 5 สินค้าออก: p₁ > p₂",
              label_a=f"พูดถึง {ASP}", label_b=f"ไม่พูดถึง {ASP}")
print("\nสรุป:", "ยังต่างจริงหลังตัด 5 สินค้า → เป็นปัญหาระดับกลุ่ม insight ผ่านข้อนี้"
      if r3.get("ตัดสินใจ") == "ปฏิเสธ H0"
      else f"ไม่ต่างแล้ว → ปัญหากระจุกใน 5 สินค้า ({share_top5:.0%}) ปรับ insight เป็นระดับสินค้า")
````

#### T4 ผลของส่วนผสมประเภทสินค้า
* **ทำอะไร:** ตรวจว่าความต่างใน T1 เกิดจากสัดส่วนประเภทสินค้าที่ไม่เท่ากันหรือไม่ เช่น ถ้าสร้อยคอพังบ่อยและแย่อยู่แล้ว รีวิวที่พูดถึงของพังก็จะมาจากสร้อยคอเป็นส่วนใหญ่
* **ทำอย่างไร:** แยกประเภทสินค้า (Necklaces, Earrings, Bracelets ฯลฯ ที่ดึงมาจาก categories) แล้วทดสอบทีละประเภท เทียบเฉพาะรีวิวในประเภทเดียวกัน ข้ามประเภทที่มีรีวิวน้อยกว่า 30 ฉบับ
  - **H0:** ในประเภทนั้น p₁ = p₂ **H1:** ในประเภทนั้น p₁ > p₂
  - p₁, p₂ นิยามเหมือน T1 แต่ใช้เฉพาะรีวิวของประเภทนั้น
* **ทำไปเพื่ออะไร:** ถ้าทุกประเภทยังต่างจริง แสดงว่าปัญหานี้ไม่ได้เกิดเพราะประเภทสินค้า ถ้าบางประเภทไม่ผ่านให้ระบุ insight เฉพาะประเภทที่ผ่าน
* **ดูผลตรงไหน:** Z และ pp ของแต่ละประเภท และบรรทัดสรุปท้ายเซลล์

````python
# เรียงประเภทสินค้าตามจำนวนรีวิว เอา 5 ประเภทแรกที่ระบุประเภทได้
type_list = [t for t in d["product_type"].value_counts().index if t != "(not specified)"][:5]
print("ประเภทสินค้าที่ทดสอบ:", type_list, "\n")

for tp in type_list:
    s = d["product_type"] == tp
    print(f"----- T4 {tp} -----")
    run_test(d, m & s, ~m & s, f"T4 {tp}", "ส่วนผสมประเภทสินค้า",
             h0=f"เฉพาะ {tp}: p₁ = p₂", h1=f"เฉพาะ {tp}: p₁ > p₂",
             label_a=f"{tp} + พูดถึง {ASP}", label_b=f"{tp} + ไม่พูดถึง")
    print()

ok, failed, n_t = check_all("T4")
print("สรุป:", f"ปฏิเสธ H0 ครบ {n_t} ประเภท → ผลไม่ได้เกิดจากส่วนผสมประเภทสินค้า insight ผ่านข้อนี้" if ok
      else f"ไม่ปฏิเสธ H0 ที่ {', '.join(failed)} → ระบุ insight เฉพาะประเภทที่ปฏิเสธ H0")
````

#### T5 ผลของเวลา
* **ทำอะไร:** ตรวจว่าปัญหาของพังเป็นเรื่องในอดีตที่ร้านแก้ไปแล้วหรือไม่ ถ้าใช่ การใช้งบแก้ตอนนี้จะเสียเปล่า
* **ทำอย่างไร:** ตัดที่ median ของปีรีวิว แบ่งเป็นช่วงแรกและช่วงหลัง แล้วทดสอบแยกทีละช่วง
  - **T5a (ช่วงแรก)** H0: p₁ = p₂ H1: p₁ > p₂
  - **T5b (ช่วงหลัง)** H0: p₁ = p₂ H1: p₁ > p₂
  - p₁, p₂ นิยามเหมือน T1 แต่ใช้เฉพาะรีวิวในช่วงเวลานั้น
* **ทำไปเพื่ออะไร:** ถ้าช่วงหลังยังต่างจริง แปลว่าปัญหายังเกิดอยู่ ไม่ใช่เรื่องเก่า
* **ดูผลตรงไหน:** Z และ pp ของ T5b (ช่วงหลัง) เป็นหลัก ถ้า T5b ไม่ปฏิเสธ H0 ให้ระบุว่าปัญหานี้เกิดเฉพาะช่วงแรก

````python
cut = int(d["year"].median())                 # ปีที่เป็น median ของรีวิว
old, new = d["year"] <= cut, d["year"] > cut  # ช่วงแรก (ถึงปี cut) / ช่วงหลัง (หลังปี cut)
print(f"ตัดที่ปี {cut} | ช่วงแรก {old.sum():,} รีวิว | ช่วงหลัง {new.sum():,} รีวิว\n")

print(f"----- T5a ช่วงแรก (ถึงปี {cut}) -----")
r5a = run_test(d, m & old, ~m & old, f"T5a ถึงปี {cut}", "เวลา",
               h0=f"เฉพาะรีวิวถึงปี {cut}: p₁ = p₂", h1=f"เฉพาะรีวิวถึงปี {cut}: p₁ > p₂",
               label_a=f"ช่วงแรก + พูดถึง {ASP}", label_b="ช่วงแรก + ไม่พูดถึง")

print(f"\n----- T5b ช่วงหลัง (หลังปี {cut}) -----")
r5b = run_test(d, m & new, ~m & new, f"T5b หลังปี {cut}", "เวลา",
               h0=f"เฉพาะรีวิวหลังปี {cut}: p₁ = p₂", h1=f"เฉพาะรีวิวหลังปี {cut}: p₁ > p₂",
               label_a=f"ช่วงหลัง + พูดถึง {ASP}", label_b="ช่วงหลัง + ไม่พูดถึง")

ok, failed, _ = check_all("T5")
print("\nสรุป:", "ปฏิเสธ H0 ทั้งสองช่วง → ปัญหายังเกิดต่อเนื่อง ไม่ใช่เรื่องในอดีต insight ผ่านข้อนี้" if ok
      else f"ไม่ปฏิเสธ H0 ที่ {', '.join(failed)} → ระบุว่าปัญหาชัดเฉพาะช่วงที่ปฏิเสธ H0")
````

#### T6 สินค้ารหัสไหนแย่กว่าสินค้าอื่นจริง (สำหรับ Insight Card 2)
* **ทำอะไร:** ตรวจว่าสินค้า 10 อันดับแรกของเรื่องนี้ (ตารางเดียวกับ 2.5.3) แย่กว่าสินค้าอื่นใน Jewelry จริง หรือติดอันดับเพียงเพราะมีรีวิวเยอะ
* **ทำอย่างไร:** ทดสอบทีละสินค้าด้วย Two-proportion Z-test แล้วปรับ p-value ด้วยวิธี Holm เพราะทดสอบหลายตัวพร้อมกัน
  - **H0:** p₁ = p₂ **H1:** p₁ > p₂
  - p₁ = สัดส่วนรีวิวแย่ของสินค้ารหัสนั้น, p₂ = สัดส่วนรีวิวแย่ของสินค้าอื่นทั้งหมดใน Jewelry
* **ทำไปเพื่ออะไร:** เลือกเฉพาะสินค้าที่แย่จริงไปใส่ Card 2 ตัวที่ไม่ผ่านจะถูกตัดออก
* **ดูผลตรงไหน:** คอลัมน์ `ตัดสินใจ` ซึ่งดูจาก p-value หลังปรับ Holm (`p-value Holm`) ตัวที่ปฏิเสธ H0 คือแย่กว่าสินค้าอื่นจริง

````python
t6_ids = asin_rank[(SC, ASP)].head(10)["parent_asin"].tolist()
rows6 = []
for pa in t6_ids:
    is_p = d["parent_asin"] == pa                      # สินค้านี้ vs สินค้าอื่นทั้งหมดใน Jewelry
    k1, n1 = int(d.loc[is_p, "is_bad"].sum()), int(is_p.sum())
    k2, n2 = int(d.loc[~is_p, "is_bad"].sum()), int((~is_p).sum())
    row = {"การทดสอบ": f"T6 {pa}", "หักล้างผลของ": "สินค้าไม่ได้แย่จริง",
           "H0": f"{pa}: p₁ = p₂", "H1": f"{pa}: p₁ > p₂", "n1": n1, "n2": n2}
    if min(n1, n2) >= MIN_N_TEST:
        z, pv = two_prop_ztest(k1, n1, k2, n2)
        row.update({"p̂1": k1 / n1, "p̂2": k2 / n2, "ต่างกัน (pp)": (k1 / n1 - k2 / n2) * 100, "Z": z, "p-value": pv})
    rows6.append(row)
t6 = pd.DataFrame(rows6)

# ปรับ p-value แบบ Holm: เรียง p จากน้อยไปมาก คูณด้วย (จำนวนที่ทดสอบ − ลำดับ) แล้วทำให้ไม่ลดลง
ok_idx = t6.index[t6["p-value"].notna()]
order  = t6.loc[ok_idx, "p-value"].sort_values().index
mt, running = len(order), 0.0
t6["p-value Holm"] = np.nan
for rank, i in enumerate(order):
    running = max(running, (mt - rank) * t6.loc[i, "p-value"])
    t6.loc[i, "p-value Holm"] = min(1.0, running)
t6["ตัดสินใจ"] = np.where(t6["p-value"].isna(), "ไม่ทดสอบ",
                  np.where(t6["p-value Holm"] < ALPHA, "ปฏิเสธ H0", "ไม่ปฏิเสธ H0"))
t6["store"] = t6["การทดสอบ"].str.replace("T6 ", "").map(asin_rank[(SC, ASP)].set_index("parent_asin")["store"])

print(f"H0: p₁ = p₂ vs H1: p₁ > p₂ | Two-proportion Z-test ทางเดียว | ปรับ p-value ด้วย Holm ({mt} สินค้า) | α = {ALPHA}\n")
display(bordered(t6[["การทดสอบ", "store", "n1", "n2", "p̂1", "p̂2", "ต่างกัน (pp)", "Z", "p-value", "p-value Holm", "ตัดสินใจ"]]
                 .style.format({"p̂1": "{:.1%}", "p̂2": "{:.1%}", "ต่างกัน (pp)": "{:+.1f}", "Z": "{:.2f}",
                                "p-value": fmt_p, "p-value Holm": fmt_p, "n1": "{:,}", "n2": "{:,}"}, na_rep="-")))
RESULTS += t6.drop(columns=["store", "p-value Holm"]).to_dict("records")

n_pass = int((t6["ตัดสินใจ"] == "ปฏิเสธ H0").sum())
print(f"\nสรุป: {n_pass} จาก {mt} สินค้า แย่กว่าสินค้าอื่นใน {SC} จริง (หลังปรับ Holm) → ใส่ Card 2 ได้เฉพาะสินค้าที่ปฏิเสธ H0")
````

````python
from IPython.display import Markdown, display

# ---------- เช็กตัวแปรที่ต้องมีก่อน ----------
need = {"TOP3": "2.5.2", "full": "2.5.3", "asin_rank": "2.5.3", "meta": "2.2",
        "fix3": "2.5.2 (ท้ายเซลล์ TOP3)", "total_bad_full": "2.5.2 (ท้ายเซลล์ TOP3)",
        "MIN_N_ASIN": "2.5.3", "res": "2.5.6"}
missing = [f"{v} (รันเซลล์ {c})" for v, c in need.items() if v not in globals()]
if missing:
    raise NameError("ยังไม่มีตัวแปร: " + ", ".join(missing))

SC1, A1 = TOP3[0]
col1 = f"asp_{A1}"
g1 = full[full["sub_category"] == SC1].copy()
m1 = g1[col1] == 1

# ---------- Card 1: ระดับปัญหา + กลุ่ม ----------
n_m, n_n   = int(m1.sum()), int((~m1).sum())
p_m, p_n   = g1.loc[m1, "is_bad"].mean(), g1.loc[~m1, "is_bad"].mean()
bad_g      = int(g1["is_bad"].sum())
rate_g     = g1["is_bad"].mean()
excess_g   = n_m * (p_m - p_n)                       # รีวิวลบส่วนเกินในกลุ่ม
share_exc  = excess_g / bad_g
rate_after = (bad_g - excess_g) / len(g1)            # % ลบของกลุ่มถ้าแก้ปัญหานี้ได้
top3_text  = "; ".join(f"({i}) {s} | {a}" for i, (s, a) in enumerate(TOP3, start=1))

# ---------- Baseline 12 เดือนล่าสุด (ใช้ในช่องวัดผล) ----------
ts = pd.to_datetime(g1["timestamp"], unit="ms")
last12 = ts >= ts.max() - pd.DateOffset(months=12)
g_recent = g1[last12]
base_per100 = (g_recent["is_bad"] * g_recent[col1]).sum() / len(g_recent) * 100   # รีวิวลบเรื่องนี้ต่อ 100 รีวิว
base_n      = len(g_recent)
base_period = f"{ts[last12].min():%b %Y}–{ts.max():%b %Y}"

# ---------- ผล Reality Check จาก 2.6 ----------
rc_ready = all(v in globals() for v in ["SC", "ASP", "summary"]) and (SC, ASP) == (SC1, A1)
if rc_ready:
    # Card 1 ใช้ T1-T5 (หักล้าง Insight 1) ส่วน Card 2 ใช้ T6 (ระดับสินค้า)
    tested = summary[summary["ตัดสินใจ"].isin(["ปฏิเสธ H0", "ไม่ปฏิเสธ H0"]) & summary["การทดสอบ"].str.match(r"T[1-5]")]
    n_rej, n_tested = int((tested["ตัดสินใจ"] == "ปฏิเสธ H0").sum()), len(tested)
    z1 = summary.loc[summary["การทดสอบ"] == "T1", "Z"].iloc[0]
    not_rej = tested.loc[tested["ตัดสินใจ"] == "ไม่ปฏิเสธ H0", "การทดสอบ"].tolist()
    t6_all = summary[summary["การทดสอบ"].str.startswith("T6") & summary["ตัดสินใจ"].isin(["ปฏิเสธ H0", "ไม่ปฏิเสธ H0"])]
    t6_rej, t6_n = int((t6_all["ตัดสินใจ"] == "ปฏิเสธ H0").sum()), len(t6_all)
    t3_rows = tested[tested["การทดสอบ"].str.startswith("T3")]
    t3_rej = bool(len(t3_rows)) and bool((t3_rows["ตัดสินใจ"] == "ปฏิเสธ H0").all())
else:
    n_rej = n_tested = t6_rej = t6_n = 0
    z1, not_rej, t3_rej = np.nan, [], True


# ---------- Card 2: ระดับสินค้า ----------
t   = asin_rank[(SC1, A1)]
top = t.head(5).copy()
ids = top["parent_asin"].tolist()
total_issue = int((g1["is_bad"] * g1[col1]).sum())
share5  = top["n_bad_issue"].sum() / max(total_issue, 1)
rate5   = top["n_bad"].sum() / top["n"].sum()
stores  = ", ".join(pd.Series(top["store"].dropna().unique()).astype(str))
reach   = pd.to_numeric(meta.set_index("parent_asin")["rating_number"].reindex(ids), errors="coerce").sum()

# T6 เฉพาะ 5 สินค้าที่อยู่ใน Card 2
if rc_ready:
    t6_top = summary[summary["การทดสอบ"].isin([f"T6 {i}" for i in ids])]
    t6_rej5 = int((t6_top["ตัดสินใจ"] == "ปฏิเสธ H0").sum())
else:
    t6_rej5 = 0

# คำตัดสินจาก 2.5.6 (แก้ที่ร้าน หรือที่สินค้า) ของ 5 สินค้านี้
DEC = {i: (res.loc[i, "decision"] if i in res.index else "ไม่อยู่ใน Watchlist") for i in ids}
prod_ids  = [i for i in ids if DEC[i] in ("แก้ที่สินค้า", "แก้ทั้งร้านและสินค้า", "แก้ที่สินค้า (ข้อมูลร้านไม่พอ)")]
store_ids = [i for i in ids if DEC[i] in ("แก้ที่ร้าน", "แก้ทั้งร้านและสินค้า")]
store_names = ", ".join(sorted({str(res.loc[i, "store"]) for i in store_ids})) or "-"

recent5 = g1[last12 & g1["parent_asin"].isin(ids)]
rate5_recent, n5_recent = (recent5["is_bad"].mean(), len(recent5)) if len(recent5) else (np.nan, 0)

print(f"Card 1: {SC1} | {A1} | % ลบเมื่อพูดถึง {p_m:.1%} (n={n_m:,}) vs ไม่พูดถึง {p_n:.1%} (n={n_n:,})")
print(f"        excess = {excess_g:,.0f} รีวิว = {share_exc:.1%} ของรีวิวลบกลุ่ม | % ลบกลุ่ม {rate_g:.1%} → {rate_after:.1%}")
print(f"        baseline {base_period}: {base_per100:.2f} รีวิวลบเรื่องนี้ต่อ 100 รีวิว (n={base_n:,})")
print(f"Card 2: 5 สินค้าครอง {share5:.1%} ของรีวิวลบเรื่องนี้ | % ลบรวม {rate5:.1%} vs กลุ่ม {rate_g:.1%} "
      f"| 12 เดือนล่าสุด {rate5_recent:.1%} (n={n5_recent:,})")
if rc_ready:
    print(f"Reality Check: T1-T5 ปฏิเสธ H0 {n_rej}/{n_tested} การทดสอบ | T1 Z = {z1:.2f} | T6 ปฏิเสธ H0 {t6_rej}/{t6_n} (ใน 5 สินค้าของ Card 2: {t6_rej5}/5)")
print("2.5.6:", DEC)
````
