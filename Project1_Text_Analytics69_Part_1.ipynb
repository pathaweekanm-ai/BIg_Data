## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่

# Part 1: Social Listening Warm-up
*งานเดี่ยว — ทุกคนต้องทำและส่งของตนเอง ไม่ต้องนำเสนอ (10 คะแนน)*

ใช้ข้อมูลโพสต์สาธารณะจาก **Bluesky**  
https://huggingface.co/datasets/alpindale/two-million-bluesky-posts

ใช้เพียง **1–3 ไฟล์** ก็พอ ไม่จำเป็นต้องโหลดทั้ง dataset

ฟิลด์ที่มี เช่น `text`, `created_at`, `author`, `uri`, `has_images`, `reply_to`, `predicted_language`

## ตอบ 4 คำถามหลัก

1. **คนโพสต์เมื่อไร:** จำนวนโพสต์ ช่วงเวลาที่ข้อมูลครอบคลุม และชั่วโมงที่โพสต์มากที่สุด
2. **ใช้ภาษาอะไร:** รายงานภาษาหลักและสัดส่วน เพื่อบอกว่าถ้าทำคอนเทนต์ควรรองรับภาษาใด
3. **คนพูดถึงอะไร:** หา hashtag หรือคำ/วลีเด่นที่น่าสนใจ และยกตัวอย่างข้อความจริงประกอบ
4. **ข้อมูลนี้บอกอะไรได้และบอกอะไรไม่ได้:** สร้าง visualization 1 รูป แล้วเขียนข้อจำกัด 3–5 บรรทัด

> Part 1 เป็น warm-up ไม่ต้องทำ influencer analysis หรือ reply-network เว้นแต่สนใจทำเพิ่มเอง


```python
# ----------------- Your code here -----------------
#โติดตั้ง Library + โหลดข้อมูลจริง

import os
import re
import random
import warnings
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from datasets import load_dataset

warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# กำหนดค่าการวิเคราะห์
# ------------------------------------------------------------

RANDOM_SEED = 42
SAMPLE_N = 20_000

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

DATA_DIR = "./data/"
os.makedirs(DATA_DIR, exist_ok=True)


# ------------------------------------------------------------
# โหลด Dataset จริงจาก Hugging Face
# ------------------------------------------------------------

print("=" * 60)
print("LOADING BLUESKY DATASET")
print("=" * 60)

dataset = load_dataset(
    "alpindale/two-million-bluesky-posts",
    split="train"
)

print(f"จำนวนข้อมูลทั้งหมดใน Dataset: {len(dataset):,} โพสต์")


# ------------------------------------------------------------
# สุ่ม 20,000 โพสต์จากข้อมูลจริง
# ------------------------------------------------------------

sample_size = min(
    SAMPLE_N,
    len(dataset)
)

dataset_sample = (
    dataset
    .shuffle(seed=RANDOM_SEED)
    .select(range(sample_size))
)

df = dataset_sample.to_pandas()


# ------------------------------------------------------------
# ตรวจสอบข้อมูล
# ------------------------------------------------------------

print(f"จำนวนข้อมูลที่นำมาวิเคราะห์: {len(df):,} โพสต์")
print(f"จำนวนตัวแปร: {len(df.columns)} ตัวแปร")

print("\nตัวแปรที่มีใน Dataset:")
print(df.columns.tolist())


# ------------------------------------------------------------
# ตรวจสอบ Missing และ Duplicate
# ------------------------------------------------------------

print("\nจำนวนข้อมูลซ้ำ:",
      df.duplicated().sum())

print("\nMissing Values:")
display(
    pd.DataFrame({
        "Missing": df.isna().sum(),
        "Missing (%)":
            df.isna().mean() * 100
    })
)
```

```python
# ----------------- Your code here -----------------
# Q1 คนโพสต์เมื่อไร
print("=" * 60)
print("Q1: คนโพสต์เมื่อไร?")
print("=" * 60)


# ------------------------------------------------------------
# เตรียมข้อมูลวันที่และเวลา
# ------------------------------------------------------------

df["created_at"] = pd.to_datetime(
    df["created_at"],
    errors="coerce",
    utc=True
)

# จำนวนวันที่แปลงไม่ได้
invalid_dates = df["created_at"].isna().sum()

print(f"\nวันที่ที่ไม่สามารถแปลงได้: {invalid_dates:,} รายการ")


# ------------------------------------------------------------
# สร้างตัวแปรเวลา
# ------------------------------------------------------------

df["hour"] = df["created_at"].dt.hour
df["date"] = df["created_at"].dt.date


# ------------------------------------------------------------
# จำนวนโพสต์
# ------------------------------------------------------------

total_posts = len(df)

print(f"\nจำนวนโพสต์ที่นำมาวิเคราะห์: {total_posts:,} โพสต์")


# ------------------------------------------------------------
# ช่วงเวลาของข้อมูล
# ------------------------------------------------------------

start_date = df["created_at"].min()
end_date = df["created_at"].max()

print("\nช่วงเวลาของข้อมูล")
print(f"เริ่มต้น : {start_date}")
print(f"สิ้นสุด  : {end_date}")
print(f"ครอบคลุม : {end_date - start_date}")


# ------------------------------------------------------------
# จำนวนโพสต์ในแต่ละชั่วโมง
# ------------------------------------------------------------

hourly_posts = (
    df["hour"]
    .dropna()
    .value_counts()
    .sort_index()
)


# ------------------------------------------------------------
# ชั่วโมงที่โพสต์มากที่สุด
# ------------------------------------------------------------

peak_hour = hourly_posts.idxmax()
peak_count = hourly_posts.max()

peak_percentage = (
    peak_count / total_posts * 100
)

print("\nชั่วโมงที่มีการโพสต์มากที่สุด")
print(f"เวลา       : {int(peak_hour):02d}:00 UTC")
print(f"จำนวนโพสต์ : {peak_count:,} โพสต์")
print(f"สัดส่วน    : {peak_percentage:.2f}%")


# ------------------------------------------------------------
# กราฟ
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

sns.barplot(
    x=hourly_posts.index,
    y=hourly_posts.values
)

plt.title(
    "Number of Bluesky Posts by Hour",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Hour of Day (UTC)")
plt.ylabel("Number of Posts")

plt.xticks(range(24))

plt.tight_layout()

plt.savefig(
    os.path.join(
        DATA_DIR,
        "Q1_posts_by_hour.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ----------------- Your code here -----------------
# Q2: ใช้ภาษาอะไร?
# ============================================================

print("=" * 60)
print("Q2: ใช้ภาษาอะไร?")
print("=" * 60)

# ตรวจสอบชื่อคอลัมน์ทั้งหมดก่อน
print("\nคอลัมน์ที่มีอยู่ในข้อมูล:")
print(df.columns.tolist())

# ------------------------------------------------------------
# ตรวจสอบว่ามีคอลัมน์สำหรับภาษาหรือไม่
# ------------------------------------------------------------

language_candidates = [
    "predicted_language",
    "language",
    "lang"
]

language_col = next(
    (col for col in language_candidates if col in df.columns),
    None
)

if language_col is None:
    print("\nไม่พบคอลัมน์ภาษาที่คาดไว้ในข้อมูล")
    print("ดังนั้น Q2 จะวิเคราะห์ภาษาจากข้อความจริงในคอลัมน์ text แทน")

    # --------------------------------------------------------
    # ตรวจสอบว่ามี text หรือไม่
    # --------------------------------------------------------
    if "text" not in df.columns:
        raise KeyError(
            "ไม่พบทั้งคอลัมน์ภาษาและคอลัมน์ text "
            "กรุณาตรวจสอบโครงสร้างข้อมูลใน Cell 1"
        )

    # --------------------------------------------------------
    # ใช้ langdetect วิเคราะห์ภาษาจากข้อความจริง
    # --------------------------------------------------------
    !pip -q install langdetect

    from langdetect import detect, DetectorFactory
    from langdetect.lang_detect_exception import LangDetectException

    DetectorFactory.seed = RANDOM_SEED

    def detect_language(text):
        try:
            text = str(text).strip()

            # ข้อความสั้นเกินไปไม่เหมาะกับการตรวจภาษา
            if len(text) < 10:
                return "unknown"

            return detect(text)

        except LangDetectException:
            return "unknown"
        except Exception:
            return "unknown"

    df["detected_language"] = df["text"].apply(detect_language)

    language_col = "detected_language"

else:
    print(f"\nพบคอลัมน์ภาษา: {language_col}")

# ------------------------------------------------------------
# นับจำนวนแต่ละภาษา
# ------------------------------------------------------------

language_counts = (
    df[language_col]
    .fillna("unknown")
    .astype(str)
    .str.lower()
    .value_counts()
)

# ------------------------------------------------------------
# ตารางสรุป
# ------------------------------------------------------------

language_summary = pd.DataFrame({
    "Language": language_counts.index,
    "Count": language_counts.values
})

language_summary["Percentage"] = (
    language_summary["Count"]
    / language_summary["Count"].sum()
    * 100
)

print("\n10 ภาษาที่พบมากที่สุด:")
display(language_summary.head(10))

# ------------------------------------------------------------
# กราฟ
# ------------------------------------------------------------

top_languages = language_summary.head(10).sort_values("Count")

plt.figure(figsize=(10, 6))

plt.barh(
    top_languages["Language"],
    top_languages["Count"]
)

plt.xlabel("Number of posts")
plt.ylabel("Language")
plt.title("Top 10 Languages Used in Bluesky Posts")

plt.tight_layout()

plt.savefig(
    "./data/Q2_top_languages.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ----------------- Your code here -----------------
# ============================================================
# CELL 4 — Q3: คนพูดถึงอะไร?
# ============================================================

print("=" * 60)
print("Q3: คนพูดถึงอะไร?")
print("=" * 60)


# ============================================================
# ส่วนที่ 1 — HASHTAG
# ============================================================

def extract_hashtags(text):

    return re.findall(
        r"#\w+",
        text.lower()
    )


# เตรียมข้อความ
df["text"] = (
    df["text"]
    .fillna("")
    .astype(str)
)


# ดึง Hashtag จากข้อมูลจริง
all_hashtags = []

for text in df["text"]:

    hashtags = extract_hashtags(text)

    all_hashtags.extend(hashtags)


# นับความถี่
hashtag_counts = Counter(
    all_hashtags
)


hashtag_df = pd.DataFrame(
    hashtag_counts.most_common(20),
    columns=[
        "hashtag",
        "count"
    ]
)


# ------------------------------------------------------------
# แสดง Hashtag
# ------------------------------------------------------------

print("\n20 Hashtags ที่พบมากที่สุด")

if len(hashtag_df) > 0:

    display(hashtag_df)

else:

    print("ไม่พบ Hashtag ในข้อมูลตัวอย่าง")


# ------------------------------------------------------------
# กราฟ Top 10 Hashtags
# ------------------------------------------------------------

if len(hashtag_df) > 0:

    top_hashtags = (
        hashtag_df
        .head(10)
        .sort_values("count")
    )

    plt.figure(figsize=(10, 6))

    sns.barplot(
        data=top_hashtags,
        x="count",
        y="hashtag"
    )

    plt.title(
        "Top 10 Hashtags in Bluesky Posts",
        fontsize=16,
        fontweight="bold"
    )

    plt.xlabel("Number of Posts")
    plt.ylabel("Hashtag")

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            DATA_DIR,
            "Q3_top_hashtags.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# ส่วนที่ 2 — WORD FREQUENCY
# ============================================================

import nltk

nltk.download(
    "stopwords",
    quiet=True
)

from nltk.corpus import stopwords


english_stopwords = set(
    stopwords.words("english")
)


# คำทั่วไปที่ไม่ช่วยในการตีความ
additional_stopwords = {
    "im",
    "ive",
    "dont",
    "didnt",
    "cant",
    "just",
    "like",
    "really",
    "got",
    "one",
    "get",
    "also"
}

english_stopwords.update(
    additional_stopwords
)


def clean_text(text):

    # lowercase
    text = text.lower()

    # ลบ URL
    text = re.sub(
        r"http\S+|www\S+",
        " ",
        text
    )

    # ลบ mention
    text = re.sub(
        r"@\w+",
        " ",
        text
    )

    # ลบเครื่องหมาย hashtag
    text = re.sub(
        r"#",
        "",
        text
    )

    # เก็บเฉพาะภาษาอังกฤษ
    text = re.sub(
        r"[^a-z\s]",
        " ",
        text
    )

    words = text.split()

    # ตัด stopwords และคำที่สั้นเกินไป
    words = [
        word
        for word in words
        if (
            word not in english_stopwords
            and len(word) >= 3
        )
    ]

    return words


# ------------------------------------------------------------
# นับคำ
# ------------------------------------------------------------

word_counter = Counter()

for text in df["text"]:

    words = clean_text(text)

    word_counter.update(words)


word_df = pd.DataFrame(
    word_counter.most_common(20),
    columns=[
        "word",
        "count"
    ]
)


print("\n20 คำที่พบมากที่สุด")
display(word_df)


# ============================================================
# ส่วนที่ 3 — ข้อความจริงประกอบการวิเคราะห์
# ============================================================

print("\n")
print("=" * 60)
print("ตัวอย่างข้อความจริง")
print("=" * 60)


if len(hashtag_df) > 0:

    selected_hashtag = (
        hashtag_df.iloc[0]["hashtag"]
    )

    print(
        f"\nตัวอย่างโพสต์ที่มี "
        f"{selected_hashtag}"
    )

    example_posts = df[
        df["text"]
        .str.lower()
        .str.contains(
            re.escape(
                selected_hashtag
            ),
            na=False
        )
    ][
        [
            "created_at",
            "text"
        ]
    ].head(5)

    display(example_posts)

else:

    print(
        "ไม่มี Hashtag สำหรับใช้เลือกข้อความตัวอย่าง"
    )
```
