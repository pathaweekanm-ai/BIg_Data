## บทบาทของนักศึกษาในงานนี้

ทีมของคุณคือ **ทีม Data and Insights** ที่ถูกจ้างมาช่วยองค์กรหนึ่งตัดสินใจ
ลูกค้าไม่ได้ต้องการรายงานสถิติ แต่ต้องการคำตอบว่า **"แล้วเราควรทำอะไรต่อ"**

ผลงานที่ส่งคือ **ชุด Insight ที่นำไปทำต่อได้จริง** พร้อมหลักฐานที่เชื่อถือได้ ไม่ใช่การไล่ทำเทคนิคให้ครบทุกอย่าง
เทคนิคทั้งหลาย (frequency, topic model, sentiment) เป็นเพียงเครื่องมือ **คะแนนอยู่ที่สิ่งที่คุณค้นพบและข้อเสนอที่ตามมา**

**สิ่งที่งานนี้ต้องการจริง ๆ**
* คิดเองว่าจะมองข้อมูลจากมุมไหนถึงจะเจอสิ่งที่คนอื่นไม่เห็น

* แปลงสิ่งที่เจอเป็นข้อเสนอที่มีเจ้าของงาน มีวิธีวัดผล และมีข้อควรระวัง

**สิ่งที่งานนี้ไม่ต้องการ**
* รายงานที่ไล่บรรยายว่า "คำที่พบบ่อยที่สุดคือ..." โดยไม่บอกว่าแล้วยังไงต่อ
* การสร้างโมเดลทำนายที่ซับซ้อนโดยไม่มีคนใช้ผลลัพธ์
* การทำครบทุกเทคนิคแบบผิวเผินแทนที่จะทำน้อยอย่างแต่ลึก

## เทคนิคที่อาจใช้ — เลือกเท่าที่จำเป็น

ไม่ต้องใช้ทั้งหมด โดยทั่วไป **2–4 เทคนิคก็เพียงพอ** ให้เลือกตามคำถามของทีม

| เทคนิค | เหมาะเมื่ออยากรู้ |
|---|---|
| **Word / n-gram frequency** | คนพูดถึงอะไรบ่อย |
| **TF-IDF / distinctive words** | สองกลุ่มพูดต่างกันอย่างไร |
| **Sentiment** | ข้อความเป็นบวก/ลบเพียงใด |
| **Aspect analysis** | คนพูดถึงราคา คุณภาพ การส่งของ ฯลฯ อย่างไร |
| **Topic modeling** | มีประเด็นอะไรที่เราอาจยังไม่รู้ |
| **Trend / segmentation** | ปัญหาเปลี่ยนตามเวลา หรือหนักในกลุ่มไหน |

> เทคนิคขั้นสูง เช่น embedding, clustering, regression, classification หรือ anomaly detection ทำได้ แต่ **ไม่บังคับ** และเหมาะสำหรับ Bonus เมื่อใช้ถูกและอธิบายได้

ถ้ายังไม่เคยใช้เทคนิคใด ให้ทดลองกับข้อมูลชุดเล็กก่อน แล้วตรวจด้วยตาว่าผลสมเหตุสมผลหรือไม่


## แบบฟอร์ม Insight Card

ทุก insight เขียนในรูปแบบนี้ 5 บรรทัด ไม่ต้องยาว

> **Insight:** ประโยคเดียวที่บอกสิ่งที่พบ ต้องมีตัวเลขอยู่ในประโยค \
> **หลักฐาน:** ตัวเลขพร้อม n และชี้ว่าดูได้จากกราฟหรือตารางไหน \
> **So what:** กระทบธุรกิจอย่างไร ถ้าประมาณขนาดผลกระทบเป็นตัวเลขได้ยิ่งดี \
> **Action:** ทีมไหนควรทำอะไร และจะรู้ได้อย่างไรว่าได้ผล \
> **ความมั่นใจ:** สูง กลาง หรือ ต่ำ พร้อมเหตุผลสั้น ๆ

**ตัวอย่างที่ยังไม่ใช่ insight**
> ลูกค้าพูดถึงเรื่องราคาบ่อยที่สุด คิดเป็น 32% ของรีวิว

เป็นเพียงข้อสังเกต ไม่ได้บอกว่าใครต้องทำอะไร และ 32% นั้นมากหรือน้อยก็ไม่รู้เพราะไม่มีอะไรให้เทียบ

**ตัวอย่างที่ใช้ได้**
> **Insight:** รีวิวที่พูดถึงบรรจุภัณฑ์รั่วมีเพียง 6.2% ของทั้งหมด (n = 3,100) แต่ได้ดาวเฉลี่ย 2.1 เทียบกับค่าเฉลี่ยรวม 4.0 และกระจุกอยู่ในสินค้า 12 ชิ้นที่ใช้ขวดแบบปั๊ม \
> **หลักฐาน:** ตารางค่าเฉลี่ยดาวแยกตามด้าน และกราฟแท่งสินค้า 12 ชิ้นในผลวิเคราะห์ \
> **So what:** สินค้า 12 ชิ้นนี้คิดเป็น 18% ของยอดรีวิวทั้งหมวด ถ้าแก้ปัญหาฝาขวดได้ คาดว่าคะแนนเฉลี่ยของหมวดจะขยับขึ้นราว 0.15 ดาว \
> **Action:** ทีม Product ตรวจสอบซัพพลายเออร์ฝาขวดของสินค้า 12 รายการนี้ภายในไตรมาสหน้า แล้วติดตามสัดส่วนรีวิวที่มีคำว่า leak, spill, broken cap ซึ่งตอนนี้อยู่ที่ 14% ต่อเดือน \
> **ความมั่นใจ:** กลาง ยืนยันได้จากข้อความ แต่ยังไม่มีข้อมูลการคืนสินค้ามาช่วยยืนยัน

### 3.1 เลือกแหล่งข้อมูลและวิธีเก็บ

เลือกแหล่งข้อมูลที่สัมพันธ์กับคำถามของทีม แล้วใช้ **เพียง 1 วิธีเก็บ**

ตัวอย่าง:
* Hacker News API — post + score + comments
* RSS — ข่าว/บทความ
* Traffy Fondue Open Data — เรื่องร้องเรียน
* เว็บไซต์สาธารณะที่อนุญาต crawler — ข่าว/บทความ/รายการต่าง ๆ

**กรอกก่อนเขียนโค้ด**
* แหล่งข้อมูล: ...
* คำถาม: ...
* ผู้ใช้ผลวิเคราะห์: ...
* **1 record คือ:** ...
* วิธีเก็บ: API / RSS / Robot
* field สำคัญที่จะเก็บ: ...

ด้านล่างมี
* **ตัวอย่างที่ 1:** API แบบสั้น
* **ตัวอย่างที่ 2–3:** ทางเลือกอื่นแบบสรุป
* **ตัวอย่างที่ 4:** Web Robot แบบอธิบายทีละขั้น


````python
# ===== ตัวอย่างที่ 1: Hacker News API (JSON, ไม่ต้องใช้ key) =====
# ได้หัวข้อข่าวพร้อม score และจำนวนคอมเมนต์ เหมาะกับการดูว่าเรื่องแบบไหนคนสนใจ
import urllib.request, json, time
import pandas as pd

def hn_get(url, timeout=30):
    with urllib.request.urlopen(url, timeout=timeout) as r:
        return json.loads(r.read().decode("utf-8"))

def fetch_hackernews(n=200, sleep=0.1):
    base = "https://hacker-news.firebaseio.com/v0"
    # ตัวอย่างนี้เก็บสูงสุด 200 stories ก็เพียงพอสำหรับ Part 3 เวอร์ชันลด workload
    ids = hn_get(f"{base}/topstories.json")[:n]
    rows = []
    for i, item_id in enumerate(ids):
        try:
            it = hn_get(f"{base}/item/{item_id}.json")
        except Exception as e:
            print("ข้าม", item_id, e)
            continue
        if not it or it.get("type") != "story" or not it.get("title"):
            continue
        rows.append({
            "doc_id": str(it["id"]),
            "text": it["title"] + (" " + it.get("text", "") if it.get("text") else ""),
            "score": it.get("score", 0),
            "n_comments": it.get("descendants", 0),
            "author": it.get("by"),
            "published_at": pd.to_datetime(it.get("time", 0), unit="s").isoformat(),
            "url": f"https://news.ycombinator.com/item?id={it['id']}",
            "source": "hackernews",
        })
        time.sleep(sleep)
        if (i + 1) % 100 == 0:
            print(f"ดึงแล้ว {i+1} รายการ")
    return rows

# rows = fetch_hackernews(50)
# print(len(rows)); rows[0]
````

### ตัวอย่างที่ 4: Article Collection Robot — อธิบายทีละขั้น

ตัวอย่างนี้ใช้เว็บไซต์ข่าวเพื่อให้เห็นแนวคิดของ Web Robot ชัดเจน

**เป้าหมาย:** ต้องการสร้างชุดข้อมูลที่

> **1 row = 1 บทความข่าว**

ดังนั้น Robot จะไม่ crawl ทุก link ในเว็บไซต์ เพราะจะได้หน้าเมนู หน้า category หน้า search หรือ URL ซ้ำ ๆ ปะปนมา

Robot จะทำงานเป็น 2 ช่วง:

`หน้ารวมข่าว → หา URL ของบทความ → เปิดบทความแต่ละ URL → เก็บ title/text/metadata → 1 ข่าว = 1 row`

ตัวอย่างด้านล่างใช้รูปแบบ URL ของ Thai PBS (`/news/content/<id>`) เพื่อสาธิตหลักการเท่านั้น  
ก่อนใช้จริงต้องตรวจ `robots.txt`, Terms of Service และปรับ URL pattern / selector ให้ตรงกับเว็บไซต์ที่ทีมเลือก

> **ข้อสังเกต:** ตัวอย่างตั้งจำนวนข่าวไว้ไม่มากเพื่อทดลองก่อน เมื่อโค้ดถูกแล้วจึงค่อยเพิ่มจำนวนหรือทำ pagination/archive


#### Step 1 — กำหนดเว็บไซต์และสิ่งที่ต้องการเก็บ

เริ่มจากกำหนดให้ชัดก่อนว่าเราจะเก็บอะไร

ในตัวอย่างนี้
* `LISTING_URL` = หน้ารวมข่าว ใช้สำหรับ **หา link**
* `ARTICLE_PATH` = รูปแบบ URL ที่ถือว่าเป็น **บทความจริง**
* `MAX_ARTICLES` = จำนวนข่าวที่ทดลองเก็บ
* `DELAY` = เวลาพักระหว่างการเรียกแต่ละหน้า เพื่อไม่ยิง request ถี่เกินไป

จุดสำคัญคือ **หน้ารวมข่าวไม่ใช่ record ที่จะเอาไปวิเคราะห์** แต่เป็นเพียงทางเข้าไปหา article URLs


````python
# ===== Step 1: ตั้งค่า =====
# หากยังไม่มี package:
# !pip install -q requests beautifulsoup4

import re
import time
import requests
import pandas as pd

from bs4 import BeautifulSoup
from urllib.parse import urljoin, urlparse, urldefrag
from urllib.robotparser import RobotFileParser

BASE_URL = "https://www.thaipbs.or.th"
LISTING_URL = "https://www.thaipbs.or.th/news"

# Thai PBS ใช้ URL ข่าวลักษณะ /news/content/509281
ARTICLE_PATH_PATTERN = r"^/news/content/\d+$"

MAX_ARTICLES = 20     # เริ่มน้อย ๆ เพื่อทดสอบก่อน
DELAY = 1.5           # พัก 1.5 วินาทีระหว่างหน้า
BOT_USER_AGENT = "KKU-TextAnalytics-StudentBot/1.0"

session = requests.Session()
session.headers.update({"User-Agent": BOT_USER_AGENT})

print("Listing page:", LISTING_URL)
print("เป้าหมาย: 1 row = 1 news article")
````

#### Step 2 — ทำความสะอาด URL และตรวจ `robots.txt`

URL เดียวกันอาจเขียนได้หลายแบบ เช่น

`https://www.thaipbs.or.th/news`  
`https://www.thaipbs.or.th/news#main-content`

ส่วนหลัง `#...` เรียกว่า **fragment** และโดยทั่วไปไม่ได้หมายถึงหน้าใหม่ จึงควรตัดออกก่อนตรวจ URL ซ้ำ

จากนั้นตรวจ `robots.txt` ก่อนเข้าแต่ละส่วนของเว็บ ถ้าไม่อนุญาตให้ bot เข้า path นั้น ให้ **เปลี่ยนแหล่งข้อมูล** ไม่ใช่หาวิธีหลบ


````python
# ===== Step 2: normalize URL + robots.txt =====

_robots_cache = {}

def normalize_url(url):
    """ตัด #fragment ออกจาก URL เพื่อไม่ให้หน้าเดียวกันถูกนับซ้ำ"""
    clean_url, _ = urldefrag(url)
    return clean_url.rstrip("/")


def robots_allowed(url, user_agent=BOT_USER_AGENT):
    """คืน True เมื่อ robots.txt อนุญาตให้ bot เข้าถึง URL"""
    p = urlparse(url)
    root = f"{p.scheme}://{p.netloc}"

    if root not in _robots_cache:
        rp = RobotFileParser()
        rp.set_url(urljoin(root, "/robots.txt"))
        try:
            rp.read()
            _robots_cache[root] = rp
        except Exception as e:
            print("อ่าน robots.txt ไม่ได้:", e)
            return False

    return _robots_cache[root].can_fetch(user_agent, url)


print("robots.txt อนุญาตหน้ารวมข่าวหรือไม่:", robots_allowed(LISTING_URL))
````

#### Step 3 — เข้า “หน้ารวมข่าว” เพื่อหาเฉพาะ Article URLs

ขั้นนี้ **ยังไม่เก็บเนื้อหาข่าว**

Robot เปิดหน้ารวมข่าว แล้วดู `<a href="...">` ทุก link จากนั้นกรองให้เหลือเฉพาะ URL ที่ตรงกับรูปแบบบทความ เช่น

`/news/content/509281`

การกรองนี้คือสิ่งที่ทำให้ Robot ต่างจาก generic crawler ที่วิ่งไปทุกหน้าใน domain


````python
# ===== Step 3: หา link ของ "บทความจริง" =====

def find_article_links(listing_url):
    if not robots_allowed(listing_url):
        raise PermissionError("robots.txt ไม่อนุญาตให้เข้าหน้ารวมข่าวนี้")

    r = session.get(listing_url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")
    article_urls = set()

    for a in soup.find_all("a", href=True):
        url = urljoin(listing_url, a["href"])
        url = normalize_url(url)

        p = urlparse(url)

        # 1) เอาเฉพาะ domain เดียวกับเว็บไซต์หลัก
        same_domain = (p.netloc == urlparse(BASE_URL).netloc)

        # 2) เอาเฉพาะ path ที่มีรูปแบบเป็นบทความข่าว
        is_article = bool(re.match(ARTICLE_PATH_PATTERN, p.path))

        if same_domain and is_article:
            article_urls.add(url)

    return sorted(article_urls)


article_urls = find_article_links(LISTING_URL)

print("พบ article URLs =", len(article_urls))
print("ตัวอย่าง 5 URLs:")
for u in article_urls[:5]:
    print(" -", u)
````

#### Step 4 — เปิดบทความทีละหน้าแล้วดึง “เนื้อหาที่ต้องการ”

ตอนนี้เรามี URL ของบทความแล้ว จึงเปิดทีละข่าวและแยก field ที่จะใช้วิเคราะห์ เช่น

* `url`
* `title`
* `published_at`
* `text`
* `word_count`

หลักสำคัญคือ **พยายามดึงเฉพาะตัวบทความ** ไม่ใช่ menu/footer/ข่าวแนะนำทั้งหน้า

ตัวอย่างใช้ `<article>` ก่อน ถ้าเว็บไซต์ไม่มี tag นี้จึง fallback ไปที่ `<main>` และนำข้อความจาก `<p>` มาต่อกัน  
เว็บไซต์อื่นอาจต้องเปลี่ยน selector เช่น `.article-body`, `.post-content` หรือ selector เฉพาะของเว็บนั้น


````python
# ===== Step 4: ดึงข้อมูลจากบทความ 1 หน้า =====

def extract_article(url):
    if not robots_allowed(url):
        print("robots.txt ไม่อนุญาต ข้าม:", url)
        return None

    r = session.get(url, timeout=20)
    r.raise_for_status()

    soup = BeautifulSoup(r.text, "html.parser")

    # ---- title ----
    h1 = soup.find("h1")
    title = h1.get_text(" ", strip=True) if h1 else ""

    # ---- published date: พยายามอ่านจาก meta หรือ time ----
    published_at = ""

    meta_time = soup.find("meta", attrs={"property": "article:published_time"})
    if meta_time and meta_time.get("content"):
        published_at = meta_time["content"]
    else:
        time_tag = soup.find("time")
        if time_tag:
            published_at = time_tag.get("datetime") or time_tag.get_text(" ", strip=True)

    # ---- article text ----
    container = soup.find("article") or soup.find("main")

    if container is None:
        return None

    paragraphs = []
    for p in container.find_all("p"):
        txt = p.get_text(" ", strip=True)

        # ตัดบรรทัดสั้น ๆ ที่มักไม่ใช่เนื้อหาหลัก
        if len(txt) >= 30:
            paragraphs.append(txt)

    text = " ".join(paragraphs).strip()

    # ถ้าดึงมาแล้วสั้นมาก มีโอกาสว่า selector ไม่ถูก
    if len(text) < 200:
        return None

    return {
        "doc_id": normalize_url(url),
        "url": normalize_url(url),
        "title": title,
        "published_at": published_at,
        "text": text,
        "word_count": len(text.split()),
        "source_domain": urlparse(url).netloc,
    }
````

#### Step 5 — ให้ Robot ทำซ้ำกับหลายบทความ

นี่คือส่วนที่ทำให้มันเป็น “Robot”

แทนที่จะเปิดข่าวด้วยมือทีละหน้า โปรแกรมจะวน URL ที่หาได้ใน Step 3 แล้วเรียก `extract_article()` ซ้ำ ๆ

มี `time.sleep(DELAY)` ทุกครั้งเพื่อไม่ส่ง request ถี่เกินไป

ในช่วงพัฒนาให้เริ่มจาก 5–20 ข่าวก่อน เมื่อดูตัวอย่างแล้วแน่ใจว่า `text` เป็นเนื้อหาข่าวจริง ค่อยเพิ่มจำนวน


````python
# ===== Step 5: เก็บหลายบทความอัตโนมัติ =====

def collect_articles(article_urls, max_articles=20, delay=1.5):
    rows = []

    for i, url in enumerate(article_urls[:max_articles], start=1):
        try:
            item = extract_article(url)

            if item is not None:
                rows.append(item)
                print(f"[{i}] OK   :", item["title"][:70])
            else:
                print(f"[{i}] SKIP :", url)

        except Exception as e:
            print(f"[{i}] ERROR:", url, e)

        time.sleep(delay)

    return rows


rows = collect_articles(
    article_urls,
    max_articles=MAX_ARTICLES,
    delay=DELAY
)

print("\nเก็บบทความที่ใช้ได้ =", len(rows))
````

#### Step 6 — ตรวจว่า “1 row = 1 ข่าว” จริงหรือไม่

อย่าเชื่อเพียงว่าโค้ดรันผ่าน ต้องเปิดดูตัวอย่างข้อมูลด้วย

สิ่งที่ต้องตรวจ:
1. URL แต่ละแถวเป็น article URL จริงหรือไม่
2. `title` ตรงกับข่าวหรือไม่
3. `text` เป็นเนื้อหาข่าว ไม่ใช่ menu/footer
4. มีข่าวซ้ำหรือไม่
5. ข้อความสั้นผิดปกติหรือไม่

หลังผ่านขั้นนี้จึงค่อยบันทึกเป็น corpus และนำไป Text Analytics


````python
# ===== Step 6: Quality Check =====

df_robot = pd.DataFrame(rows)

if len(df_robot) > 0:
    print("จำนวนก่อน clean =", len(df_robot))
    print("URL ซ้ำ =", df_robot["url"].duplicated().sum())
    print("ข้อความซ้ำ =", df_robot["text"].duplicated().sum())

    # ตัด URL ซ้ำและข้อความซ้ำ
    df_robot = (
        df_robot
        .drop_duplicates(subset=["url"])
        .drop_duplicates(subset=["text"])
        .copy()
    )

    # ตัวอย่างเกณฑ์ง่าย ๆ: ข่าวต้องมีอย่างน้อย 80 คำ
    df_robot = df_robot[df_robot["word_count"] >= 80].copy()

    print("จำนวนหลัง clean =", len(df_robot))

    display(
        df_robot[
            ["title", "published_at", "word_count", "url"]
        ].head(10)
    )

    # อ่านเนื้อหาจริง 1 ข่าวเพื่อตรวจด้วยตา
    print("\nตัวอย่าง text:\n")
    print(df_robot.iloc[0]["text"][:1500])
else:
    print("ยังไม่มีข้อมูล ตรวจ article pattern / selector / robots.txt")
````

เมื่อแน่ใจแล้วว่าแต่ละแถวเป็นบทความจริง ให้ save ข้อมูลไว้ก่อน เพื่อไม่ต้องยิงเว็บไซต์ใหม่ทุกครั้งที่เปิด notebook

### ถ้าต้องการมากกว่าข่าวที่อยู่หน้าแรกทำอย่างไร?

ต้องเพิ่ม **pagination / archive pages** ตามโครงสร้างของเว็บไซต์นั้น เช่น

`หน้ารวมวันที่ 1 → article links`  
`หน้ารวมวันที่ 2 → article links`  
`หน้ารวมวันที่ 3 → article links`

แล้วรวม URL ก่อนเก็บ article content

**ไม่ควร** แก้ด้วยการให้ crawler วิ่งตามทุก link บนเว็บไซต์ เพราะจะกลับไปได้ menu/category/บริการอื่นปะปนเหมือน generic crawler

### สิ่งที่นักศึกษาต้องปรับเมื่อใช้เว็บอื่น

โดยทั่วไปแก้เพียง 3 จุด:
1. `LISTING_URL` — หน้าไหนมีรายการสิ่งที่ต้องการ
2. `ARTICLE_PATH_PATTERN` — URL แบบไหนคือ record จริง
3. selector ใน `extract_article()` — เนื้อหาหลักอยู่ใน tag/class ไหน

เป้าหมายสุดท้ายไม่ใช่ “crawler เก็บได้กี่หน้า” แต่คือ

> **ได้ analytical records ที่นิยามชัด สะอาด และตอบคำถามของทีมได้**


### 3.1.1 คำถาม แหล่งข้อมูล และวิธีเก็บ

* **แหล่งข้อมูล:** The Guardian Open Platform Content API (คีย์ Developer ฟรี สำหรับงานไม่เชิงพาณิชย์) บทความที่ติดแท็กปัญญาประดิษฐ์ (AI) ช่วง 1 มิ.ย. - 5 ต.ค. 2569 (ชื่อแท็กที่ใช้จะบันทึกหลังตรวจ)
* **คำถาม:** ข่าว AI ใน The Guardian ถูกเล่าด้วยกรอบใด (ความเสี่ยง/ความปลอดภัย, กฎหมายและการกำกับดูแล, การแข่งขันสหรัฐ-จีน, เศรษฐกิจและแรงงาน, สิทธิและจริยธรรม, นวัตกรรมผลิตภัณฑ์) และสัดส่วนเปลี่ยนไปอย่างไรรายเดือน รวมถึงต่างกันอย่างไรระหว่างข่าวกับบทความความเห็น
* **ผู้ใช้ผลวิเคราะห์:** ทีมสื่อสารของหน่วยงานนโยบาย AI หรือองค์กรด้านความปลอดภัย AI ที่ต้องเลือกประเด็นและจังหวะการสื่อสาร
* **1 record คือ:** 1 บทความที่ไม่ซ้ำ ที่มีเนื้อหาอย่างน้อย 300 คำ
* **วิธีเก็บ:** API (Guardian Open Platform) วิธีเดียวตามกติกา
* **field สำคัญที่จะเก็บ:** id, headline, standfirst, body_text, section, type, published_at, word_count, url (ไม่เก็บชื่อผู้เขียน) และ tone_tags, keyword_tags (รหัสแท็กเท่านั้น)
* **เทคนิควิเคราะห์ (1-2 อย่าง):** จัดหมวดกรอบข่าวด้วยรายการคำที่กำหนดล่วงหน้า + เทียบสัดส่วนรายเดือนและตามประเภทบทความ

**ข้อกำหนดที่ล็อกก่อนดูผล**
* แท็กหรือคำค้นที่ใช้ และรายการคำของแต่ละกรอบ จะบันทึกในโน้ตบุ๊กก่อนวิเคราะห์ หากปรับภายหลังต้องบันทึกเหตุผล
* เกณฑ์ผ่านก่อนวิเคราะห์: อย่างน้อย 200 valid records หลัง clean

**ข้อจำกัดที่ทราบแล้ว**
* เป็นสื่อเดียว (สื่ออังกฤษ) ผลอธิบายการเล่าข่าวใน The Guardian เท่านั้น ไม่ใช่สื่อโลกทั้งหมด
* การจัดหมวดจากคำอาจผิด จะให้สมาชิกในทีมสุ่มตรวจอย่างน้อย 30-50 บทความ
* ข้อมูลจริงจบที่วันที่เก็บข้อมูล (1 ต.ค. 2569) เดือน ต.ค. ไม่ครบ จึงไม่ใส่ในกราฟแนวโน้มรายเดือน

**ความรับผิดชอบ:** ใช้คีย์ของตนเอง ใส่ delay ตามข้อจำกัดของ API ใช้เฉพาะงานเรียน ไม่เผยแพร่เนื้อข่าวซ้ำ และให้เครดิต The Guardian

#### 3.1.2 ตั้งค่า Guardian API และค้นหาแท็ก AI
ใช้ Guardian Open Platform Content API เพียงวิธีเดียว (ไม่ scrape หน้าเว็บ) ขั้นนี้นับและตรวจเท่านั้น ยังไม่เก็บข้อมูลจริงและไม่บันทึกไฟล์

````python
if not RERUN_GUARDIAN_API:
    print("Skipped: set RERUN_GUARDIAN_API = True (authors only, needs a Guardian API key) to run this cell.")
else:
    # ===== Part 3 (Guardian API) - Setup: key, polite request helper =====
    import os
    import time
    import requests

    BASE_URL = "https://content.guardianapis.com"
    REQUEST_DELAY_SEC = 1.2   # polite delay between requests (>= 1 second)
    TIMEOUT_SEC = 20          # per-request timeout
    MAX_RETRIES = 3           # retries after the first attempt

    # Read the API key from Colab Secrets, fall back to an environment variable
    API_KEY = None
    try:
        from google.colab import userdata
        API_KEY = userdata.get("GUARDIAN_API_KEY")
    except Exception:
        API_KEY = None
    if not API_KEY:
        API_KEY = os.environ.get("GUARDIAN_API_KEY")

    # Never print the key itself, only whether it was found
    print("API key loaded." if API_KEY else "API key not found: add GUARDIAN_API_KEY to Colab Secrets.")

    api_calls = 0          # total HTTP requests sent in this session
    _last_call_time = 0.0  # timestamp of the previous request
    last_remaining = {}    # latest remaining allowance from rate-limit headers


    def _print_rate_limits(headers):
        """Print only the remaining daily and per-minute allowance, if the headers exist."""
        global last_remaining
        rem = {k: v for k, v in headers.items()
               if "ratelimit" in k.lower() and "remaining" in k.lower()}
        if not rem:
            print("[rate limit] no rate-limit headers returned")
            return
        last_remaining = rem
        day = next((v for k, v in rem.items() if "day" in k.lower()), "n/a")
        minute = next((v for k, v in rem.items() if "minute" in k.lower()), "n/a")
        print(f"[rate limit] remaining today: {day} | remaining this minute: {minute}")


    def guardian_get(endpoint, params):
        """GET one Guardian endpoint with delay, timeout and retries. Returns the 'response' dict or None."""
        global api_calls, _last_call_time
        if not API_KEY:
            print("No API key, request skipped.")
            return None
        query = dict(params)
        query["api-key"] = API_KEY   # key goes only through the params argument
        for attempt in range(1, MAX_RETRIES + 2):
            # Polite delay since the previous request
            wait = REQUEST_DELAY_SEC - (time.time() - _last_call_time)
            if wait > 0:
                time.sleep(wait)
            _last_call_time = time.time()
            api_calls += 1
            try:
                resp = requests.get(f"{BASE_URL}/{endpoint}", params=query, timeout=TIMEOUT_SEC)
            except requests.RequestException:
                # Do not print the exception (it may contain the request URL)
                print(f"Attempt {attempt}: network error or timeout, retrying" if attempt <= MAX_RETRIES
                      else f"Attempt {attempt}: network error or timeout, giving up")
                time.sleep(2 * attempt)
                continue
            _print_rate_limits(resp.headers)
            if resp.status_code == 200:
                try:
                    return resp.json().get("response", {})
                except ValueError:
                    print("HTTP 200 but the response was not valid JSON")
                    return None
            print(f"Attempt {attempt}: HTTP {resp.status_code} - request failed")
            if resp.status_code in (429, 500, 502, 503, 504) and attempt <= MAX_RETRIES:
                time.sleep(2 * attempt)   # back off, then retry
                continue
            return None   # non-retryable error (e.g. 400/401/403) or retries used up
        return None
````

````python
if not RERUN_GUARDIAN_API:
    print("Skipped: set RERUN_GUARDIAN_API = True (authors only, needs a Guardian API key) to run this cell.")
else:
    # ===== Part 3 (Guardian API) - Step 1: list candidate AI tags =====
    calls_before = api_calls
    tags = guardian_get("tags", {"q": "artificial intelligence", "page-size": 20})

    if tags:
        print(f"\nTags matched: {tags.get('total')} (showing up to 20)\n")
        print(f"{'#':>2}  {'id':<45} {'type':<10} {'sectionId':<16} webTitle")
        for i, t in enumerate(tags.get("results", []), 1):
            print(f"{i:>2}  {t.get('id', ''):<45} {t.get('type', ''):<10} "
                  f"{str(t.get('sectionId', '-')):<16} {t.get('webTitle', '')}")

    print(f"\nAPI calls used in Step 1: {api_calls - calls_before}")
````

#### 3.1.3 บันทึกแท็กและช่วงเวลาที่ล็อกไว้
* **แท็กที่ใช้:** `technology/artificialintelligenceai` (webTitle: "AI (artificial intelligence)") เลือกจากผล `/tags` q="artificial intelligence" (พบ 5 แท็ก อีก 4 แท็กไม่เกี่ยวกับ AI) ใช้แท็กนี้แท็กเดียว ไม่เพิ่มแท็กหรือคำค้นอื่น
* **ช่วงเวลา:** from-date = 2026-06-01, to-date = 2026-10-05
* **วิธีเก็บ:** Guardian Open Platform Content API (`/search`) วิธีเดียว

````python
if not RERUN_GUARDIAN_API:
    print("Skipped: set RERUN_GUARDIAN_API = True (authors only, needs a Guardian API key) to run this cell.")
else:
    # ===== Part 3 (Guardian API) - Step 2: cheap counts (page-size=1, read response.total only) =====
    AI_TAG = "technology/artificialintelligenceai"
    FROM_DATE, TO_DATE = "2026-06-01", "2026-10-05"


    def count_total(from_date, to_date, section=None):
        """Return response.total for the AI tag in a date range (optionally one section)."""
        params = {"tag": AI_TAG, "from-date": from_date, "to-date": to_date, "page-size": 1}
        if section:
            params["section"] = section
        res = guardian_get("search", params)
        return res.get("total") if res else None


    calls_before_step2 = api_calls
    counts = []   # (group, label, total)

    # a) whole period
    counts.append(("a) total", f"{FROM_DATE} to {TO_DATE}", count_total(FROM_DATE, TO_DATE)))

    # b) per month
    months = [("2026-06", "2026-06-01", "2026-06-30"),
              ("2026-07", "2026-07-01", "2026-07-31"),
              ("2026-08", "2026-08-01", "2026-08-31"),
              ("2026-09", "2026-09-01", "2026-09-30"),
              ("2026-10 (1-5)", "2026-10-01", "2026-10-05")]
    for label, f, t in months:
        counts.append(("b) month", label, count_total(f, t)))

    # c) by section, whole period
    sections = ["technology", "business", "world", "politics", "commentisfree",
                "science", "us-news", "uk-news", "culture"]
    for s in sections:
        counts.append(("c) section", s, count_total(FROM_DATE, TO_DATE, section=s)))

    # Compact table
    print(f"\n{'group':<12} {'label':<28} {'total':>7}")
    print("-" * 49)
    for g, lab, n in counts:
        print(f"{g:<12} {lab:<28} {str(n):>7}")
    print(f"\nAPI calls used in Step 2: {api_calls - calls_before_step2}")
````

````python
if not RERUN_GUARDIAN_API:
    print("Skipped: set RERUN_GUARDIAN_API = True (authors only, needs a Guardian API key) to run this cell.")
else:
    # ===== Part 3 (Guardian API) - Step 3: format check on ONE page (10 newest articles) =====
    calls_before_step3 = api_calls
    page = guardian_get("search", {
        "tag": AI_TAG,
        "from-date": FROM_DATE,
        "to-date": TO_DATE,
        "order-by": "newest",
        "page-size": 10,
        "page": 1,
        "show-fields": "headline,standfirst,bodyText,wordcount",
    })

    n_ok = 0
    if page:
        print(f"\n{'#':>2}  {'id':<70} {'sectionId':<14} {'type':<9} {'words':>6}  body")
        for i, a in enumerate(page.get("results", []), 1):
            fields = a.get("fields", {})
            body_ok = bool(fields.get("bodyText", "").strip())   # check only, never print body text
            try:
                wc = int(fields.get("wordcount", 0))
            except (TypeError, ValueError):
                wc = 0
            if wc >= 300 and body_ok:
                n_ok += 1
            print(f"{i:>2}  {a.get('id', '')[:70]:<70} {a.get('sectionId', ''):<14} "
                  f"{a.get('type', ''):<9} {wc:>6}  {'yes' if body_ok else 'no'}")
        print(f"\nArticles with wordcount >= 300 and non-empty bodyText: {n_ok} / {len(page.get('results', []))}")

    print(f"API calls used in Step 3: {api_calls - calls_before_step3}")
    print(f"API calls used in Steps 2-3: {api_calls - calls_before_step2}")
    day_left = next((v for k, v in last_remaining.items() if "day" in k.lower()), "n/a")
    print(f"Remaining daily allowance: {day_left}")
````

#### 3.1.4 เก็บบทความ ทำความสะอาด และบันทึก corpus
ดึงทุกหน้าจาก `/search` (แท็กและช่วงวันที่ตามที่ล็อกไว้) ไม่ขอข้อมูลผู้เขียน clean 5 ขั้นตามลำดับ แล้วบันทึก `corpus.jsonl` + `corpus_manifest.json`

````python
if not RERUN_GUARDIAN_API:
    print("Skipped: set RERUN_GUARDIAN_API = True (authors only, needs a Guardian API key) to run this cell.")
else:
    # ===== Part 3 (Guardian API) - Collection: fetch all pages, stop on rate limit =====
    import datetime as dt

    COLLECT_PARAMS = {
        "tag": AI_TAG,
        "from-date": FROM_DATE,
        "to-date": TO_DATE,
        "order-by": "oldest",
        "page-size": 50,
        "show-fields": "headline,standfirst,bodyText,wordcount",
        "show-tags": "tone,keyword",   # no byline, no contributor tags
    }


    class RateLimitStop(Exception):
        """Raised when the API signals a rate limit; collection stops without retrying."""


    def utc_now():
        return dt.datetime.now(dt.timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ")


    def _store_rate_limits(headers):
        """Store (do not print) the remaining allowance from rate-limit headers."""
        global last_remaining
        rem = {k: v for k, v in headers.items()
               if "ratelimit" in k.lower() and "remaining" in k.lower()}
        if rem:
            last_remaining = rem


    def fetch_page(params):
        """One polite request. Returns (status_code, response_dict). Never retries on 429."""
        global api_calls, _last_call_time
        query = dict(params)
        query["api-key"] = API_KEY   # key goes only through the params argument
        for attempt in range(1, MAX_RETRIES + 2):
            wait = REQUEST_DELAY_SEC - (time.time() - _last_call_time)
            if wait > 0:
                time.sleep(wait)
            _last_call_time = time.time()
            api_calls += 1
            try:
                resp = requests.get(f"{BASE_URL}/search", params=query, timeout=TIMEOUT_SEC)
            except requests.RequestException:
                # Do not print the exception (it may contain the request URL)
                print(f"Attempt {attempt}: network error or timeout")
                if attempt <= MAX_RETRIES:
                    time.sleep(3 * attempt)
                    continue
                return None, None
            _store_rate_limits(resp.headers)
            if resp.status_code == 429:
                hint = resp.headers.get("Retry-After")
                print("HTTP 429 - rate limit reached, collection stopped."
                      + (f" Retry hint: {hint}" if hint else " No retry hint provided."))
                raise RateLimitStop()
            if resp.status_code == 200:
                try:
                    return 200, resp.json().get("response", {})
                except ValueError:
                    print("HTTP 200 but the response was not valid JSON")
                    return 200, None
            print(f"Attempt {attempt}: HTTP {resp.status_code} - request failed")
            if resp.status_code in (500, 502, 503, 504) and attempt <= MAX_RETRIES:
                time.sleep(3 * attempt)   # back off, then retry
                continue
            return resp.status_code, None
        return None, None


    calls_before_collect = api_calls
    collect_started_utc = utc_now()
    raw_records = []
    page_size_used = 50
    page_no, total_pages, pages_done = 1, None, 0
    collection_stopped = False

    try:
        while True:
            params = dict(COLLECT_PARAMS, **{"page-size": page_size_used, "page": page_no})
            status, res = fetch_page(params)
            # Fallback: if page-size=50 is rejected on the first page, use 20
            if res is None and status == 400 and page_no == 1 and page_size_used == 50:
                print("page-size=50 rejected (HTTP 400), falling back to 20")
                page_size_used = 20
                continue
            if res is None or res.get("status") != "ok":
                print(f"Stopped at page {page_no}: request failed (HTTP {status})")
                collection_stopped = True
                break
            fetched_at = utc_now()
            for a in res.get("results", []):
                f = a.get("fields", {}) or {}
                tags = a.get("tags", []) or []
                try:
                    wc = int(f.get("wordcount", 0))
                except (TypeError, ValueError):
                    wc = 0
                # Keep only the agreed fields; no author information is stored
                raw_records.append({
                    "id": a.get("id"),
                    "type": a.get("type"),
                    "sectionId": a.get("sectionId"),
                    "webPublicationDate": a.get("webPublicationDate"),
                    "webUrl": a.get("webUrl"),
                    "headline": f.get("headline", ""),
                    "standfirst": f.get("standfirst", ""),
                    "bodyText": f.get("bodyText", ""),
                    "wordcount": wc,
                    "tone_tags": [t.get("id") for t in tags if t.get("type") == "tone"],
                    "keyword_tags": [t.get("id") for t in tags if t.get("type") == "keyword"],
                    "retrieved_at": fetched_at,
                })
            pages_done = page_no
            total_pages = res.get("pages", 0)
            if page_no >= total_pages:
                break
            page_no += 1
    except RateLimitStop:
        collection_stopped = True

    collect_finished_utc = utc_now()
    collect_calls = api_calls - calls_before_collect
    COLLECT_PARAMS_USED = dict(COLLECT_PARAMS, **{"page-size": page_size_used})
    print(f"Pages fetched: {pages_done} / {total_pages} | page-size used: {page_size_used} | "
          f"records: {len(raw_records)} | stopped early: {collection_stopped}")
````

````python
if not RERUN_GUARDIAN_API:
    print("Skipped: set RERUN_GUARDIAN_API = True (authors only, needs a Guardian API key) to run this cell.")
else:
    # ===== Part 3 (Guardian API) - Cleaning: 5 steps in a fixed order =====
    import pandas as pd

    if collection_stopped:
        print("Collection is incomplete - cleaning skipped.")
    else:
        df_raw = pd.DataFrame(raw_records)
        removed = {}

        s1 = df_raw[df_raw["type"] == "article"]
        removed["1) type != article"] = len(df_raw) - len(s1)

        s2 = s1.drop_duplicates(subset="id", keep="first")
        removed["2) duplicate id"] = len(s1) - len(s2)

        s3 = s2.drop_duplicates(subset="bodyText", keep="first")
        removed["3) duplicate identical bodyText"] = len(s2) - len(s3)

        s4 = s3[s3["bodyText"].fillna("").str.strip() != ""]
        removed["4) empty bodyText"] = len(s3) - len(s4)

        s5 = s4[s4["wordcount"] >= 300]
        removed["5) wordcount < 300"] = len(s4) - len(s5)

        # Valid records: pick the original dicts (keeps native Python types for JSON)
        valid_records = [raw_records[i] for i in s5.index]
        df_valid = pd.DataFrame(valid_records)
        print(f"Raw: {len(df_raw)} -> valid: {len(df_valid)}")
````

````python
if not RERUN_GUARDIAN_API:
    print("Skipped: set RERUN_GUARDIAN_API = True (authors only, needs a Guardian API key) to run this cell.")
else:
    # ===== Part 3 (Guardian API) - Save corpus.jsonl + corpus_manifest.json, then verify =====
    import json

    if collection_stopped:
        print("Collection is incomplete - nothing saved.")
    else:
        with open("corpus.jsonl", "w", encoding="utf-8") as f:
            for rec in valid_records:
                f.write(json.dumps(rec, ensure_ascii=False) + "\n")

        manifest = {
            "source": "Guardian Open Platform Content API",
            "endpoint": "/search",
            "query_parameters": COLLECT_PARAMS_USED,   # api-key is not included
            "tag_id": AI_TAG,
            "date_range": {"from": FROM_DATE, "to": TO_DATE},
            "retrieved_at_utc": {"started": collect_started_utc, "finished": collect_finished_utc},
            "api_calls": collect_calls,
            "pages_fetched": pages_done,
            "raw_count": len(raw_records),
            "valid_count": len(valid_records),
        }
        with open("corpus_manifest.json", "w", encoding="utf-8") as f:
            json.dump(manifest, f, ensure_ascii=False, indent=2)

        # Verify that the corpus loads back
        corp = pd.read_json("corpus.jsonl", lines=True)
        print("corpus.jsonl loaded back, shape:", corp.shape)
````

````python
if not RERUN_GUARDIAN_API:
    print("Skipped: set RERUN_GUARDIAN_API = True (authors only, needs a Guardian API key) to run this cell.")
else:
    # ===== Part 3 (Guardian API) - Collection report (no article text is printed) =====
    if collection_stopped:
        print("Collection is incomplete - report skipped.")
    else:
        # a) raw records and counts by type
        print(f"a) Raw records fetched: {len(df_raw)}")
        print(df_raw["type"].value_counts().to_string(), "\n")

        # b) removed at each cleaning step
        print("b) Removed at each cleaning step:")
        for k, v in removed.items():
            print(f"   {k:<35} {v:>5}")
        print(f"   {'Final valid records':<35} {len(df_valid):>5}\n")

        # c) valid records per month and per section
        dates = pd.to_datetime(df_valid["webPublicationDate"], utc=True)
        month_labels = {"2026-06": "Jun", "2026-07": "Jul", "2026-08": "Aug",
                        "2026-09": "Sep", "2026-10": "Oct (partial)"}
        per_month = dates.dt.strftime("%Y-%m").value_counts()
        print("c) Valid records per month:")
        for k, lab in month_labels.items():
            print(f"   {lab:<15} {int(per_month.get(k, 0)):>5}")
        print("   Valid records per section:")
        print(df_valid["sectionId"].value_counts().to_string(), "\n")

        # d) tone tags (top 10) and records with no tone tag
        tones = df_valid["tone_tags"].explode().dropna()
        print("d) Valid records per tone tag (top 10):")
        print(tones.value_counts().head(10).to_string())
        print(f"   Records with no tone tag: {int((df_valid['tone_tags'].str.len() == 0).sum())}\n")

        # e) word count summary
        wc = df_valid["wordcount"]
        print(f"e) Word count - min: {wc.min()}, median: {wc.median():.0f}, max: {wc.max()}\n")

        # f) date range actually observed
        print(f"f) Earliest webPublicationDate: {df_valid['webPublicationDate'].min()}")
        print(f"   Latest webPublicationDate:   {df_valid['webPublicationDate'].max()}\n")

    # g) API calls in this step
    print(f"g) Total API calls in this step: {collect_calls}")
    day_left = next((v for k, v in last_remaining.items() if "day" in k.lower()), "n/a")
    print(f"   Remaining daily allowance (header): {day_left}")
````

````python
# ----------------- Your code here -----------------

# [moved from team cell hMiBrYNyIPKN]
# ===== Part 3.3 (team) - Frame dictionary v1 (frozen before looking at results) =====
import re

FRAME_VERSION = "v1"
LEDE_WORDS = 200        # lede = headline + standfirst + first 200 words of bodyText
FRAME_THRESHOLD = 2     # a frame is flagged when total matches in the lede >= 2

# Items containing \w* are prefix patterns; everything else is an exact word or phrase
FRAME_DICT = {
    "risk_safety": [
        "risk", "risks", "risky", "danger", "dangers", "dangerous", "harm", "harms", "harmful",
        "threat", "threats", "threaten", "threatens", "safety", "unsafe", "existential",
        r"catastroph\w*", "misuse", r"hallucinat\w*", r"deepfake\w*", "disinformation",
        "misinformation", "loss of control", "extinction", r"jailbreak\w*", "scam", "scams", "fraud",
    ],
    "law_regulation": [
        r"regulat\w*", r"legislat\w*", r"lawmaker\w*", "law", "laws", r"lawsuit\w*", "sued", "court",
        "courts", "judge", "judges", "ruling", "probe", "investigation", "investigations",
        "antitrust", "FTC", "executive order", "compliance", "oversight", "ban", "banned", "bans",
        "EU AI Act", "Senate", "Congress", "parliament", "attorney general",
    ],
    "geopolitics": [
        "China", "Chinese", "Beijing", "arms race", "AI race", "rival", "rivals",
        r"export control\w*", r"geopolit\w*", "national security", "Pentagon", "military",
        r"superpower\w*", "sovereign", "DeepSeek", "Huawei", "Taiwan", "Xi Jinping", "cold war",
    ],
    "economy_work": [
        "job", "jobs", "worker", "workers", "employment", "employer", "employers", "employees",
        r"unemploy\w*", r"layoff\w*", "productivity", "economy", "economic", "economist",
        "economists", "investor", "investors", "investment", "investments", "valuation", "bubble",
        "billion", "trillion", "profit", "profits", "revenue", "shares", "stock", "stocks",
        "data centre", "data centres", "data center", "data centers", "electricity", "power grid",
        "energy", "startup", "startups", "funding",
    ],
    "rights_society": [
        "copyright", "copyrighted", "privacy", "surveillance", "bias", "biased", "discrimination",
        "consent", "artists", "creators", "intellectual property", "ethic", "ethics", "ethical",
        "human rights", "personal data", "facial recognition", "mental health", "children",
        "child", "teenagers", "teens", "students", "schools", "teachers", "creative", "writers",
        "actors",
    ],
    "innovation_product": [
        "launch", "launched", "launches", "unveil", "unveiled", "unveils", "release", "released",
        "releases", "rollout", "rolls out", "rolled out", "new model", "latest model",
        "new feature", "new features", "app", "apps", "assistant", "assistants", "breakthrough",
        "upgrade", "upgraded", "AI agent", "AI agents", "chatbot", "chatbots",
    ],
}


def item_to_regex(item):
    r"""Exact word/phrase -> escaped regex; a trailing \w* is kept as a prefix wildcard."""
    is_prefix = item.endswith(r"\w*")
    core = item[:-3] if is_prefix else item
    rx = r"\s+".join(re.escape(w) for w in core.split())   # phrases allow any whitespace
    return rx + (r"\w*" if is_prefix else "")


# One case-insensitive, whole-word pattern per frame (longest alternatives first)
FRAME_PATTERNS = {
    fr: re.compile(r"\b(?:" + "|".join(sorted((item_to_regex(i) for i in items), key=len, reverse=True)) + r")\b",
                   re.IGNORECASE)
    for fr, items in FRAME_DICT.items()
}

print(f"Frame dictionary {FRAME_VERSION} | threshold >= {FRAME_THRESHOLD} | lede words = {LEDE_WORDS}")
for fr, items in FRAME_DICT.items():
    print(f"  {fr:<20} {len(items):>3} items")

# [moved from team cell uMkv1pjQIPUx]
# ===== Part 3.3 (team) - Apply frame dictionary v1 to the lede =====
import os
import html
import pandas as pd

CORPUS_OK = os.path.exists("corpus.jsonl")
if not CORPUS_OK:
    print("corpus.jsonl not found (runtime may have been reset). Stopped - please re-run the collection cells first.")
else:
    corpus = pd.read_json("corpus.jsonl", lines=True)

    def clean_html(s):
        """Strip HTML tags, unescape HTML entities, normalise whitespace."""
        s = "" if s is None else str(s)
        s = re.sub(r"<[^>]+>", " ", s)
        s = html.unescape(s)
        return re.sub(r"\s+", " ", s).strip()

    standfirst_clean = corpus["standfirst"].map(clean_html)
    body_clean = corpus["bodyText"].map(clean_html)
    first_words = body_clean.map(lambda t: " ".join(t.split()[:LEDE_WORDS]))
    corpus["lede"] = corpus["headline"].fillna("").astype(str) + " " + standfirst_clean + " " + first_words

    # Match counts and flags per frame
    FLAG_COLS = []
    for fr, pat in FRAME_PATTERNS.items():
        corpus[f"count_{fr}"] = corpus["lede"].map(lambda t, p=pat: len(p.findall(t)))
        corpus[f"flag_{fr}"] = corpus[f"count_{fr}"] >= FRAME_THRESHOLD
        FLAG_COLS.append(f"flag_{fr}")
    corpus["n_frames"] = corpus[FLAG_COLS].sum(axis=1)
    corpus["no_frame"] = corpus["n_frames"] == 0

    # Content group from tone tags (precedence: opinion > news > other > unknown)
    OPINION_TONES = {"tone/comment", "tone/editorials", "tone/letters"}

    def content_group(tags):
        tags = set(tags) if isinstance(tags, list) else set()
        if tags & OPINION_TONES:
            return "opinion"
        if "tone/news" in tags:
            return "news"
        if tags:
            return "other"
        return "unknown"

    corpus["content_group"] = corpus["tone_tags"].map(content_group)

    # Overall summary only (no breakdown by month or group)
    n = len(corpus)
    print(f"Articles: {n}\n")
    print("Overall prevalence (% of articles flagged):")
    for fr in FRAME_PATTERNS:
        print(f"  {fr:<20} {corpus[f'flag_{fr}'].mean() * 100:5.1f}%")
    print(f"  {'no frame':<20} {corpus['no_frame'].mean() * 100:5.1f}%\n")
    print("Distribution of n_frames:")
    print(corpus["n_frames"].value_counts().sort_index().to_string(), "\n")
    print("Counts per content_group:")
    print(corpus["content_group"].value_counts().to_string())

# [moved from team cell kTmXsHtOh6YT]
# ===== Part 3.3b (team) - A1: analysis data (v1 lede flags, Jun-Sep 2026) =====
import os
import numpy as np
import pandas as pd
from statsmodels.stats.proportion import proportion_confint, confint_proportions_2indep
from statsmodels.stats.multitest import multipletests
from scipy.stats import chi2_contingency, fisher_exact

PRELIM = "PRELIMINARY (v1 dictionary, frames not yet validated by humans)"


def ptitle(name):
    """Print a table title that always carries the preliminary prefix."""
    print(f"\n=== {PRELIM} | {name} ===")


def wilson(k, n):
    """Wilson 95% CI in percent."""
    lo, hi = proportion_confint(k, n, alpha=0.05, method="wilson")
    return lo * 100, hi * 100


def diff_ci(k1, n1, k2, n2):
    """Difference p1 - p2 in percentage points with Newcombe 95% CI."""
    lo, hi = confint_proportions_2indep(k1, n1, k2, n2, method="newcomb", compare="diff", alpha=0.05)
    return (k1 / n1 - k2 / n2) * 100, lo * 100, hi * 100


def two_sample_p(k1, n1, k2, n2):
    """Two-sided p: Pearson chi-square (no continuity correction), Fisher exact if any expected count < 5."""
    table = np.array([[k1, n1 - k1], [k2, n2 - k2]])
    if table[:, 0].sum() == 0 or table[:, 1].sum() == 0:
        return 1.0, "none"   # no variation, test not informative
    chi2, p, dof, expected = chi2_contingency(table, correction=False)
    if (expected < 5).any():
        return fisher_exact(table)[1], "fisher"
    return p, "chi2"


A_OK = False
FRAMES = list(FRAME_DICT) if "FRAME_DICT" in globals() else []
needed = ["corpus", "FRAME_PATTERNS", "clean_html", "FLAG_COLS", "FRAME_DICT"]
if not os.path.exists("corpus.jsonl"):
    print("corpus.jsonl not found (runtime may have been reset). Stopped.")
elif not all(v in globals() for v in needed):
    print("Step C objects are not in memory. Re-run only our cells FRAME_DICT (hMiBrYNyIPKN) and Step C (uMkv1pjQIPUx), then this cell.")
else:
    # Reuse the Step C dataframe (lede + v1 flags + content_group) without changing it
    df_all = corpus.copy()
    df_all["month"] = pd.to_datetime(df_all["webPublicationDate"], utc=True).dt.strftime("%Y-%m")
    ANALYSIS_MONTHS = ["2026-06", "2026-07", "2026-08", "2026-09"]
    pop = df_all[df_all["month"].isin(ANALYSIS_MONTHS)].copy()
    pop["period"] = np.where(pop["month"] == "2026-09", "Sep", "Jun-Aug")
    A_OK = True

    print(f"Analysis population: {len(pop)} articles (excluded {len(df_all) - len(pop)} from 2026-10)")
    ptitle("n per month and content_group (Jun-Sep 2026)")
    print(pd.crosstab(pop["month"], pop["content_group"], margins=True, margins_name="total").to_string())

# [moved from team cell c0cOavk3h6jG]
# ===== Part 3.3b (team) - A2: monthly frame shares with Wilson 95% CI =====
if not A_OK:
    print("A1 did not complete - A2 skipped.")
else:
    rows = []
    for fr in FRAMES:
        row = {"frame": fr}
        for m in ANALYSIS_MONTHS + ["pooled"]:
            sub = pop if m == "pooled" else pop[pop["month"] == m]
            k, n = int(sub[f"flag_{fr}"].sum()), len(sub)
            lo, hi = wilson(k, n)
            row[m] = f"{k}/{n} {k / n * 100:4.1f}% [{lo:4.1f}, {hi:4.1f}]"
        rows.append(row)
    ptitle("monthly share per frame: flagged/n share% [Wilson 95% CI]")
    with pd.option_context("display.width", 300, "display.max_colwidth", None):
        print(pd.DataFrame(rows).to_string(index=False))

# [moved from team cell FAW6wycTh6uV]
# ===== Part 3.3b (team) - A3: September vs June-August =====
if not A_OK:
    print("A1 did not complete - A3 skipped.")
else:
    sep, jja = pop[pop["period"] == "Sep"], pop[pop["period"] == "Jun-Aug"]
    rows = []
    for fr in FRAMES:
        k1, n1 = int(sep[f"flag_{fr}"].sum()), len(sep)
        k2, n2 = int(jja[f"flag_{fr}"].sum()), len(jja)
        d, lo, hi = diff_ci(k1, n1, k2, n2)
        p, test = two_sample_p(k1, n1, k2, n2)
        rows.append({"frame": fr, "Sep_%": k1 / n1 * 100, "JunAug_%": k2 / n2 * 100,
                     "diff_pp": d, "CI_low": lo, "CI_high": hi, "p": p, "test": test})
    a3_res = pd.DataFrame(rows)
    a3_res["p_holm"] = multipletests(a3_res["p"], method="holm")[1]
    a3_res = a3_res.reindex(a3_res["diff_pp"].abs().sort_values(ascending=False).index)
    ptitle(f"September (n={len(sep)}) vs June-August (n={len(jja)}), diff = Sep - JunAug, Newcombe 95% CI")
    print(a3_res.to_string(index=False, float_format=lambda x: f"{x:.3f}" if abs(x) < 1 else f"{x:.1f}"))

    ptitle("composition: content_group share (%) by period")
    print((pd.crosstab(pop["content_group"], pop["period"], normalize="columns") * 100).round(1).to_string())
    for per, sub in [("Sep", sep), ("Jun-Aug", jja)]:
        ptitle(f"composition: top 5 sections, {per} (% of period)")
        print((sub["sectionId"].value_counts(normalize=True).head(5) * 100).round(1).to_string())

# [moved from team cell 4tEcIoS-h652]
# ===== Part 3.3b (team) - A4: stratified check of Sep vs Jun-Aug (no p-values) =====
if not A_OK:
    print("A1 did not complete - A4 skipped.")
else:
    all_sign = {r.frame: np.sign(round(r.diff_pp, 1)) for r in a3_res.itertuples()}
    for grp in ["news", "opinion"]:
        g = pop[pop["content_group"] == grp]
        gs, gj = g[g["period"] == "Sep"], g[g["period"] == "Jun-Aug"]
        ptitle(f"stratified Sep vs Jun-Aug within content_group = {grp} (Sep n={len(gs)}, Jun-Aug n={len(gj)})")
        if len(gs) < 30 or len(gj) < 30:
            print("n too small")
            continue
        rows = []
        for fr in FRAMES:
            d, lo, hi = diff_ci(int(gs[f"flag_{fr}"].sum()), len(gs), int(gj[f"flag_{fr}"].sum()), len(gj))
            rows.append({"frame": fr, "Sep_%": gs[f"flag_{fr}"].mean() * 100, "JunAug_%": gj[f"flag_{fr}"].mean() * 100,
                         "diff_pp": d, "CI_low": lo, "CI_high": hi,
                         "same_sign_as_all": "yes" if np.sign(round(d, 1)) == all_sign[fr] else "no"})
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))

# [moved from team cell wPJP5iTNh7FP]
# ===== Part 3.3b (team) - A5: news vs opinion (months pooled) =====
if not A_OK:
    print("A1 did not complete - A5 skipped.")
else:
    nw, op = pop[pop["content_group"] == "news"], pop[pop["content_group"] == "opinion"]
    rows = []
    for fr in FRAMES:
        k1, n1 = int(nw[f"flag_{fr}"].sum()), len(nw)
        k2, n2 = int(op[f"flag_{fr}"].sum()), len(op)
        d, lo, hi = diff_ci(k1, n1, k2, n2)
        p, test = two_sample_p(k1, n1, k2, n2)
        rows.append({"frame": fr, "news_%": k1 / n1 * 100, "opinion_%": k2 / n2 * 100,
                     "diff_pp": d, "CI_low": lo, "CI_high": hi, "p": p, "test": test})
    a5_res = pd.DataFrame(rows)
    a5_res["p_holm"] = multipletests(a5_res["p"], method="holm")[1]
    ptitle(f"news (n={len(nw)}) vs opinion (n={len(op)}), diff = news - opinion, Newcombe 95% CI")
    print(a5_res.to_string(index=False, float_format=lambda x: f"{x:.3f}" if abs(x) < 1 else f"{x:.1f}"))

    ptitle("monthly share of opinion articles among all articles")
    op_m = pop.groupby("month")["content_group"].agg(n="size", opinion=lambda s: int((s == "opinion").sum()))
    op_m["opinion_%"] = (op_m["opinion"] / op_m["n"] * 100).round(1)
    print(op_m.to_string())

# [moved from team cell Kxsp2nvah7Qr]
# ===== Part 3.3b (team) - A6: sensitivity checks planned in v1 =====
import re

if not A_OK:
    print("A1 did not complete - A6 skipped.")
else:
    AI_TERMS = ["AI", "A.I.", "artificial intelligence", "chatbot", "chatbots", "ChatGPT", "OpenAI",
                "Anthropic", "Gemini", "Grok", "Claude", "large language model", "LLM",
                "algorithm", "algorithms"]
    # Whole-word match that also works for terms ending in "." (A.I.)
    AI_PATTERN = re.compile(r"(?<!\w)(?:" + "|".join(
        r"\s+".join(re.escape(w) for w in t.split()) for t in sorted(AI_TERMS, key=len, reverse=True)
    ) + r")(?!\w)", re.IGNORECASE)

    # Full body text (cleaned the same way as in Step C)
    body_full = pop["bodyText"].map(clean_html)
    head_stand = pop["headline"].fillna("").astype(str) + " " + pop["standfirst"].map(clean_html)
    pop["has_ai_term"] = head_stand.map(lambda t: bool(AI_PATTERN.search(t)))

    def summarize(flags, mask=None):
        """Pooled, Sep and Jun-Aug share (%) and Sep - JunAug difference (pp)."""
        d = pop if mask is None else pop[mask]
        f = flags if mask is None else flags[mask]
        s_sep, s_jja = f[d["period"] == "Sep"].mean() * 100, f[d["period"] == "Jun-Aug"].mean() * 100
        diff = s_sep - s_jja
        direction = "0" if round(diff, 1) == 0 else ("+" if diff > 0 else "-")
        return f.mean() * 100, s_sep, s_jja, diff, direction

    rows = []
    for fr in FRAMES:
        body_counts = body_full.map(lambda t, p=FRAME_PATTERNS[fr]: len(p.findall(t)))
        variants = {
            "v1 (baseline: lede >= 2)": (pop[f"flag_{fr}"], None),
            "(a) lede >= 1": (pop[f"count_{fr}"] >= 1, None),
            "(b) lede >= 3": (pop[f"count_{fr}"] >= 3, None),
            "(c) full body >= 2": (body_counts >= 2, None),
            "(d) v1, AI term in headline/standfirst": (pop[f"flag_{fr}"], pop["has_ai_term"]),
        }
        for vname, (flags, mask) in variants.items():
            pooled, s_sep, s_jja, diff, direction = summarize(flags, mask)
            rows.append({"frame": fr, "variant": vname, "pooled_%": pooled, "Sep_%": s_sep,
                         "JunAug_%": s_jja, "diff_pp": diff, "dir": direction})
    ptitle("sensitivity: share (%) per frame and variant, diff = Sep - JunAug")
    with pd.option_context("display.width", 250):
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))

    ptitle(f"variant (d): articles with an AI term in headline/standfirst, n = {int(pop['has_ai_term'].sum())} of {len(pop)}")
    d_m = pop.groupby("month")["has_ai_term"].agg(n="size", with_ai_term="sum")
    d_m["share_%"] = (d_m["with_ai_term"] / d_m["n"] * 100).round(1)
    print(d_m.to_string())

````

### 3.3.1 กติกาจัดกรอบข่าว 6 กรอบ (ล็อกก่อนดูผล)

**นิยามกรอบข่าว 6 กรอบ**
* `risk_safety` = ความเสี่ยงและความปลอดภัย: เน้นอันตราย ภัยคุกคาม การใช้ในทางที่ผิด ข้อมูลเท็จ หรือความเสียหายจาก AI
* `law_regulation` = กฎหมายและการกำกับดูแล: เน้นกฎหมาย การออกกฎ คดีความ การสอบสวน หรือบทบาทของศาลและสภา
* `geopolitics` = ภูมิรัฐศาสตร์และการแข่งขัน: เน้นการแข่งขันระหว่างประเทศ (โดยเฉพาะสหรัฐ-จีน) ความมั่นคง หรือการทหาร
* `economy_work` = เศรษฐกิจและแรงงาน: เน้นเงินลงทุน มูลค่า ตำแหน่งงาน การจ้างงาน ศูนย์ข้อมูลและพลังงาน
* `rights_society` = สิทธิ จริยธรรม และผลกระทบต่อสังคม: เน้นลิขสิทธิ์ ความเป็นส่วนตัว อคติ เด็ก/นักเรียน สุขภาพจิต
* `innovation_product` = นวัตกรรมและผลิตภัณฑ์: เน้นการเปิดตัวผลิตภัณฑ์ โมเดลใหม่ ฟีเจอร์ หรือความก้าวหน้าทางเทคนิค

**กติกา**
1. บทความหนึ่งมีได้หลายกรอบ (multi-label)
2. ประเมินกรอบจาก "lede" = headline + standfirst + 200 คำแรกของ bodyText
3. นับว่าบทความมีกรอบนั้น เมื่อจำนวนคำที่ตรงกับพจนานุกรมใน lede รวมกัน >= 2
4. พจนานุกรมและเกณฑ์ล็อกไว้ที่ v1 หากเปลี่ยนภายหลังต้องบันทึกเหตุผลและเลขเวอร์ชันใหม่
5. การตรวจความไว (sensitivity) ที่วางแผนไว้: เกณฑ์ 1 และ 3 และใช้ bodyText เต็มแทน lede

ล็อกเมื่อ: 2026-10-01 11:02:05 UTC

````python
# ===== Part 3.3 (team) - Blind validation sample (50 articles) =====
if not CORPUS_OK:
    print("corpus.jsonl not found - sampling skipped.")
else:
    seed = globals().get("RANDOM_SEED", 42)   # fallback 42 if no seed was defined in setup
    news_pool = corpus[corpus["content_group"] == "news"]
    other_pool = corpus[corpus["content_group"] != "news"]
    val_sample = pd.concat([
        news_pool.sample(n=25, replace=False, random_state=seed),
        other_pool.sample(n=25, replace=False, random_state=seed),
    ])

    # Keep program flags for later comparison (not printed, to keep labelling blind)
    val_program_flags = val_sample.set_index("id")[FLAG_COLS + ["n_frames", "no_frame"]].copy()

    print(f"Seed used: {seed}\n")
    with pd.option_context("display.max_colwidth", None, "display.width", 300):
        print(val_sample[["id", "content_group", "sectionId", "headline", "webUrl"]].to_string(index=False))

    val_ids = val_sample["id"].tolist()
    print("\nval_ids =", val_ids)
````

````python
# ===== Part 3.3 (team) - F1: read the two labeled sheets and validate =====
import io
import contextlib
import numpy as np
import pandas as pd

LABEL7 = ["risk", "law", "geo", "econ", "rights", "prod", "ai_central"]
EXPECTED_COLS = ["id", "headline", "webUrl"] + LABEL7
F1_OK = False

if LABELS_FROM_NOTEBOOK:
    # Offline path: use the AI-rater labels stored in the notebook (LABELS_AI); no Google services
    if "LABELS_AI" not in globals() or "val_ids" not in globals():
        print("LABELS_AI or val_ids not found. Run the stored-labels cell and Step D first.")
    else:
        problems = []
        ids_ai = LABELS_AI["id"].astype(str).str.strip()
        if len(LABELS_AI) != 50:
            problems.append(f"LABELS_AI: expected 50 rows, found {len(LABELS_AI)}")
        if ids_ai.duplicated().any():
            problems.append(f"LABELS_AI: column id has {int(ids_ai.duplicated().sum())} duplicated rows")
        if set(ids_ai) != set(val_ids):
            problems.append(f"LABELS_AI: column id differs from val_ids "
                            f"({len(set(ids_ai) - set(val_ids))} unexpected, {len(set(val_ids) - set(ids_ai))} missing)")
        for key in ("A", "B"):
            for col in LABEL7:
                name = f"{key}_{col}"
                if name not in LABELS_AI.columns:
                    problems.append(f"rater {key}: column {col} is missing")
                    continue
                invalid = int((~LABELS_AI[name].isin([0, 1])).sum())
                if invalid:
                    problems.append(f"rater {key}: column {col} has {invalid} invalid rows")
        if problems:
            print("Validation FAILED - stopped:")
            for p in problems:
                print("  -", p)
        else:
            # Same structure as the Google path: lab["A"], lab["B"] with the 7 label columns in val_ids order
            ai = LABELS_AI.assign(id=ids_ai).set_index("id").loc[val_ids]
            lab = {key: ai[[f"{key}_{col}" for col in LABEL7]].set_axis(LABEL7, axis=1).astype(int)
                   for key in ("A", "B")}
            F1_OK = True
            print("Validation passed for both sheets (50 rows each, ids match val_ids, all 7 labels are 0/1).")
            print("(labels source: LABELS_AI stored in the notebook)")
elif not RERUN_GOOGLE_SHEETS:
    print("Skipped: set LABELS_FROM_NOTEBOOK = True (stored labels) or RERUN_GOOGLE_SHEETS = True (authors only, needs Google authorization).")
else:
    if "created" not in globals() or "val_ids" not in globals():
        print("Sheet objects or val_ids not found (runtime may have been reset). Stopped - please tell me the sheet links.")
    else:
        from google.colab import sheets
        sheet_urls = {title: sh.url for title, sh in created}   # kept in a variable, never printed

        def read_sheet(url):
            """Re-open a sheet by URL and read it fresh, silencing any library output."""
            with contextlib.redirect_stdout(io.StringIO()):
                s = sheets.InteractiveSheet(url=url, display=False)
                df = s.as_df()
            df.columns = [str(c).strip() for c in df.columns]
            return df

        def to01(v):
            """Return 0/1 for valid values (int or '0'/'1' string), otherwise None."""
            if isinstance(v, (bool, np.bool_)):
                return None
            if isinstance(v, (int, np.integer)) and int(v) in (0, 1):
                return int(v)
            if isinstance(v, (float, np.floating)) and not np.isnan(v) and float(v) in (0.0, 1.0):
                return int(v)
            if isinstance(v, str) and v.strip() in ("0", "1"):
                return int(v.strip())
            return None

        raw_labels, problems = {}, []
        for title in ["labeling_rater_A", "labeling_rater_B"]:
            df = read_sheet(sheet_urls[title])
            raw_labels[title] = df
            missing_cols = [c for c in EXPECTED_COLS if c not in df.columns]
            if missing_cols:
                problems.append(f"{title}: missing columns {missing_cols}")
                continue
            if len(df) != 50:
                problems.append(f"{title}: expected 50 rows, found {len(df)}")
            ids = df["id"].astype(str).str.strip()
            if ids.duplicated().any():
                problems.append(f"{title}: column id has {int(ids.duplicated().sum())} duplicated rows")
            if set(ids) != set(val_ids):
                problems.append(f"{title}: column id differs from val_ids "
                                f"({len(set(ids) - set(val_ids))} unexpected, {len(set(val_ids) - set(ids))} missing)")
            for col in LABEL7:
                vals = df[col]
                blank = vals.map(lambda v: v is None or (isinstance(v, float) and np.isnan(v))
                                 or (isinstance(v, str) and v.strip().upper() in ("", "NA", "N/A", "NAN"))).sum()
                invalid = vals.map(lambda v: to01(v) is None).sum()
                if invalid:
                    problems.append(f"{title}: column {col} has {int(invalid)} invalid rows "
                                    f"({int(blank)} blank/NA, {int(invalid - blank)} other values)")

        if problems:
            print("Validation FAILED - stopped:")
            for p in problems:
                print("  -", p)
        else:
            # Aligned 0/1 label tables in val_ids order
            lab = {}
            for key, title in [("A", "labeling_rater_A"), ("B", "labeling_rater_B")]:
                df = raw_labels[title].copy()
                df["id"] = df["id"].astype(str).str.strip()
                df = df.set_index("id").loc[val_ids]
                lab[key] = df[LABEL7].apply(lambda s: s.map(to01)).astype(int)
            F1_OK = True
            print("Validation passed for both sheets (50 rows each, ids match val_ids, all 7 labels are 0/1).")
````

````python
# ===== Part 3.3 (team) - F2: inter-rater reliability (A vs B) =====
import warnings
from sklearn.metrics import cohen_kappa_score

if not F1_OK:
    print("F1 validation did not pass - F2 skipped.")
else:
    A, B = lab["A"], lab["B"]
    rows = []
    for col in LABEL7:
        a, b = A[col].values, B[col].values
        agree = (a == b).mean() * 100
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            k = cohen_kappa_score(a, b)
        kappa = "undefined" if (k is None or np.isnan(k)) else f"{k:.3f}"
        rows.append({"label": col, "agreement_%": round(agree, 1), "kappa": kappa,
                     "A=1": int(a.sum()), "B=1": int(b.sum()), "either=1": int(((a == 1) | (b == 1)).sum())})
    reliability = pd.DataFrame(rows)
    print(reliability.to_string(index=False))

    print(f"\nShare ai_central=1 - A: {A['ai_central'].mean() * 100:.1f}% | B: {B['ai_central'].mean() * 100:.1f}%")
    print(f"Articles where A and B differ on ai_central: {int((A['ai_central'] != B['ai_central']).sum())}")
````

````python
if not RERUN_GOOGLE_SHEETS:
    print("Skipped: set RERUN_GOOGLE_SHEETS = True (authors only, needs Google authorization) to run this cell.")
else:
    from google.colab import sheets   # needed here when F1 used the stored labels
    # ===== Part 3.3 (team) - F3: create the consensus sheet =====
    if not F1_OK:
        print("F1 validation did not pass - F3 skipped.")
    else:
        A, B = lab["A"], lab["B"]
        cons = corpus.set_index("id").loc[val_ids, ["headline", "webUrl"]].reset_index()
        a_vals, b_vals = A.reset_index(drop=True), B.reset_index(drop=True)

        # Number of the 7 labels where A and B differ
        cons["n_disagreements"] = (a_vals[LABEL7] != b_vals[LABEL7]).sum(axis=1).values

        # Spot check: 10 fully agreed articles (seed 42, without replacement)
        agreed_ids = cons.loc[cons["n_disagreements"] == 0, "id"]
        n_spot = min(10, len(agreed_ids))
        spot_ids = set(agreed_ids.sample(n=n_spot, replace=False, random_state=42))
        cons["spot_check"] = cons["id"].isin(spot_ids).astype(int)

        # A_, B_, C_ columns; C_ is prefilled only when A == B and the row is not a spot check
        for col in LABEL7:
            cons[f"A_{col}"] = a_vals[col].values
            cons[f"B_{col}"] = b_vals[col].values
            cons[f"C_{col}"] = [int(x) if (x == y and s == 0) else ""
                                for x, y, s in zip(a_vals[col], b_vals[col], cons["spot_check"])]

        ordered_cols = ["id", "headline", "webUrl"]
        for col in LABEL7:
            ordered_cols += [f"A_{col}", f"B_{col}", f"C_{col}"]
        ordered_cols += ["n_disagreements", "spot_check"]
        cons = cons[ordered_cols].sort_values(["n_disagreements", "spot_check"],
                                              ascending=False, kind="mergesort").reset_index(drop=True)

        # Create the Google Sheet (library output silenced; URL printed once below)
        with contextlib.redirect_stdout(io.StringIO()):
            cons_sheet = sheets.InteractiveSheet(title="labeling_consensus", df=cons.copy(), display=False)

        print(f"Articles with at least one disagreement: {int((cons['n_disagreements'] > 0).sum())}")
        print(f"Spot-check articles: {int(cons['spot_check'].sum())}"
              + ("" if n_spot == 10 else f" (only {n_spot} fully agreed articles available)"))
        print(f"Consensus sheet: {cons_sheet.url}")
````

````python
# ===== Part 3.3b (team) - A7: program vs the two AI raters (NOT human labels) =====
import io
import contextlib

MAP7 = {"risk": "risk_safety", "law": "law_regulation", "geo": "geopolitics",
        "econ": "economy_work", "rights": "rights_society", "prod": "innovation_product"}

ab = None
if "val_program_flags" not in globals():
    print("val_program_flags is missing. Re-run only our cells FRAME_DICT, Step C and Step D first.")
elif list(val_program_flags.index) != list(val_ids):
    print("val_program_flags id order does not match val_ids - stopped.")
elif LABELS_FROM_NOTEBOOK:
    # Offline path: A_ and B_ labels stored in the notebook (LABELS_AI); no Google services
    ab = (LABELS_AI.assign(id=LABELS_AI["id"].astype(str).str.strip()).set_index("id")
          .loc[val_ids, [f"{r}_{k}" for k in MAP7 for r in ("A", "B")]].astype(int))
elif not RERUN_GOOGLE_SHEETS or "cons_sheet" not in globals():
    print("Skipped: set LABELS_FROM_NOTEBOOK = True (stored labels) or RERUN_GOOGLE_SHEETS = True (authors only, needs Google authorization).")
else:
    from google.colab import sheets
    with contextlib.redirect_stdout(io.StringIO()):
        cons_read = sheets.InteractiveSheet(url=cons_sheet.url, display=False).as_df()
    cons_read.columns = [str(c).strip() for c in cons_read.columns]
    cons_read["id"] = cons_read["id"].astype(str).str.strip()
    # Use only A_ and B_ columns; every C_ column is ignored
    ab = cons_read.set_index("id").loc[val_ids, [f"{r}_{k}" for k in MAP7 for r in ("A", "B")]].astype(int)

if ab is not None:
    def prf(pred, truth):
        """Precision, recall, F1 of the program (pred) against a rater (truth)."""
        tp = int(((pred == 1) & (truth == 1)).sum())
        fp = int(((pred == 1) & (truth == 0)).sum())
        fn = int(((pred == 0) & (truth == 1)).sum())
        prec = tp / (tp + fp) if (tp + fp) else np.nan
        rec = tp / (tp + fn) if (tp + fn) else np.nan
        f1 = 2 * prec * rec / (prec + rec) if (tp + fp) and (tp + fn) and (prec + rec) else np.nan
        return prec, rec, f1

    rows = []
    for short, fr in MAP7.items():
        pred = val_program_flags[f"flag_{fr}"].astype(int).values
        row = {"frame": fr}
        for r in ("A", "B"):
            p_, r_, f_ = prf(pred, ab[f"{r}_{short}"].values)
            row[f"precision_{r}"], row[f"recall_{r}"], row[f"F1_{r}"] = p_, r_, f_
        rows.append(row)
    ptitle("program vs AI raters (NOT human labels, do NOT use to edit the dictionary)")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.2f}", na_rep="undefined"))
````

### 3.3.5 กรอบความเสี่ยงที่เพิ่มใน ก.ย. มาจากอะไร (เชิงสำรวจ)
การตรวจในส่วนนี้ **เพิ่มขึ้นหลังจากเห็นผล A3** (risk_safety ใน ก.ย. สูงกว่า มิ.ย.-ส.ค.) จึงเป็นการวิเคราะห์เชิงสำรวจ ไม่ได้วางแผนไว้ล่วงหน้า และรายงานผลของทุก variant ที่ลองไว้ทั้งหมด ไม่เลือกรายงานเฉพาะบางแบบ (พจนานุกรม v1 และกติกา >= 2 ไม่เปลี่ยน)

````python
# ===== Part 3.3c (team) - B1: which risk_safety terms drive the flags =====
import re
import numpy as np
import pandas as pd

EXPL = "EXPLORATORY (not pre-registered; v1 dictionary not human-validated)"


def etitle(name):
    """Print a table title that always carries the exploratory prefix."""
    print(f"\n=== {EXPL} | {name} ===")


B_OK = all(v in globals() for v in ["pop", "FRAME_DICT", "item_to_regex", "diff_ci", "corpus"])
if not B_OK:
    print("Objects from earlier steps are missing. Re-run only our cells FRAME_DICT, Step C and A1 first.")
else:
    RISK_TERMS = FRAME_DICT["risk_safety"]
    # One whole-word, case-insensitive pattern per term (same construction as v1)
    term_rx = {t: re.compile(r"\b(?:" + item_to_regex(t) + r")\b", re.IGNORECASE) for t in RISK_TERMS}
    term_counts = pd.DataFrame({t: pop["lede"].map(lambda s, p=p: len(p.findall(s))) for t, p in term_rx.items()},
                               index=pop.index)
    same = bool((term_counts.sum(axis=1) == pop["count_risk_safety"]).all())
    print(f"Check: per-term counts add up to the v1 risk_safety count for every article: {same}")

    is_sep = pop["period"] == "Sep"
    flagged = pop["flag_risk_safety"]
    n_flag_sep, n_flag_jja = int((flagged & is_sep).sum()), int((flagged & ~is_sep).sum())
    rows = []
    for t in RISK_TERMS:
        has = term_counts[t] > 0
        rows.append({
            "term": t,
            "Sep_articles": int((has & is_sep).sum()),
            "JunAug_articles": int((has & ~is_sep).sum()),
            "Sep_%_of_flagged": (has & flagged & is_sep).sum() / n_flag_sep * 100,
            "JunAug_%_of_flagged": (has & flagged & ~is_sep).sum() / n_flag_jja * 100,
        })
    b1_res = pd.DataFrame(rows).sort_values(["Sep_articles", "JunAug_articles"], ascending=False)
    etitle(f"risk_safety terms in the lede (Sep n=216, flagged {n_flag_sep}; Jun-Aug n=415, flagged {n_flag_jja})")
    print(b1_res.to_string(index=False, float_format=lambda x: f"{x:.1f}"))
````

````python
# ===== Part 3.3c (team) - B2: leave-one-group-out for risk_safety =====
if not B_OK:
    print("Earlier objects missing - B2 skipped.")
else:
    RISK_GROUPS = {
        "generic_hazard": ["risk", "risks", "risky", "danger", "dangers", "dangerous", "harm", "harms",
                           "harmful", "threat", "threats", "threaten", "threatens", "unsafe", "safety"],
        "extreme": ["existential", r"catastroph\w*", "loss of control", "extinction"],
        "tech_failure": ["misuse", r"hallucinat\w*", r"jailbreak\w*"],
        "info_fraud": [r"deepfake\w*", "disinformation", "misinformation", "scam", "scams", "fraud"],
    }
    grouped = [t for g in RISK_GROUPS.values() for t in g]
    assert sorted(grouped) == sorted(RISK_TERMS), "RISK_GROUPS must cover every risk_safety term exactly once"

    variants = {"all groups": grouped}
    for g in RISK_GROUPS:
        variants[f"without {g}"] = [t for t in grouped if t not in RISK_GROUPS[g]]
    for g, terms in RISK_GROUPS.items():
        variants[f"only {g}"] = terms

    rows = []
    for vname, terms in variants.items():
        flag = term_counts[terms].sum(axis=1) >= FRAME_THRESHOLD   # same v1 rule: >= 2 matches in the lede
        k1, n1 = int((flag & is_sep).sum()), int(is_sep.sum())
        k2, n2 = int((flag & ~is_sep).sum()), int((~is_sep).sum())
        d, lo, hi = diff_ci(k1, n1, k2, n2)
        rows.append({"variant": vname, "n_terms": len(terms), "Sep_%": k1 / n1 * 100, "JunAug_%": k2 / n2 * 100,
                     "diff_pp": d, "CI_low": lo, "CI_high": hi})
    etitle("risk_safety leave-one-group-out (rule: >= 2 matches in the lede), diff = Sep - JunAug, Newcombe 95% CI")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
````

````python
# ===== Part 3.3c (team) - B3: concentration of risk_safety over time =====
if not B_OK:
    print("Earlier objects missing - B3 skipped.")
else:
    pub = pd.to_datetime(pop["webPublicationDate"], utc=True)
    iso = pub.dt.isocalendar()
    week_key = iso["year"].astype(str) + "-W" + iso["week"].astype(str).str.zfill(2)
    wk = pd.DataFrame({"week": week_key, "date": pub.dt.date, "flag": pop["flag_risk_safety"]})
    weekly = wk.groupby("week").agg(first_day=("date", "min"), last_day=("date", "max"),
                                    articles=("flag", "size"), risk_flagged=("flag", "sum"))
    weekly["share_%"] = (weekly["risk_flagged"] / weekly["articles"] * 100).round(1)
    etitle("weekly risk_safety (ISO weeks, 2026-06-01 to 2026-09-30; first/last day = observed publication dates)")
    print(weekly.to_string())

    # Top 15 September flagged articles by number of risk-term matches (headline only)
    sep_flag = pop[is_sep & pop["flag_risk_safety"]].copy()
    sep_flag["matches"] = sep_flag["count_risk_safety"]
    sep_flag["matched_terms"] = [
        ", ".join(f"{t}x{int(term_counts.at[i, t])}" for t in RISK_TERMS if term_counts.at[i, t] > 0)
        for i in sep_flag.index
    ]
    sep_flag["date"] = pd.to_datetime(sep_flag["webPublicationDate"], utc=True).dt.strftime("%Y-%m-%d")
    top15 = sep_flag.sort_values(["matches", "date"], ascending=[False, True]).head(15)
    etitle("top 15 September risk-flagged articles by number of risk-term matches")
    with pd.option_context("display.max_colwidth", 90, "display.width", 300):
        print(top15[["date", "matches", "headline", "content_group", "matched_terms"]].to_string(index=False))

    in_window = sep_flag["date"].between("2026-09-10", "2026-09-24")
    print(f"\nSeptember risk-flagged articles published 2026-09-10 to 2026-09-24: "
          f"{int(in_window.sum())} of {len(sep_flag)}")
````

````python
# ===== Part 3.3c (team) - B4: composition check by section group (Mantel-Haenszel) =====
from statsmodels.stats.contingency_tables import StratifiedTable

if not B_OK:
    print("Earlier objects missing - B4 skipped.")
else:
    sec_group = pop["sectionId"].where(pop["sectionId"].isin(["technology", "commentisfree", "us-news"]), "other")
    rows, tables = [], []
    for g in ["technology", "commentisfree", "us-news", "other"]:
        m = sec_group == g
        s, j = pop[m & is_sep], pop[m & ~is_sep]
        k1, n1, k2, n2 = int(s["flag_risk_safety"].sum()), len(s), int(j["flag_risk_safety"].sum()), len(j)
        d, lo, hi = diff_ci(k1, n1, k2, n2)
        rows.append({"section_group": g, "Sep_n": n1, "Sep_%": k1 / n1 * 100, "JunAug_n": n2,
                     "JunAug_%": k2 / n2 * 100, "diff_pp": d, "CI_low": lo, "CI_high": hi})
        # 2x2 per stratum: rows = Sep / Jun-Aug, columns = flagged / not flagged
        tables.append(np.array([[k1, n1 - k1], [k2, n2 - k2]]))
    etitle("risk_safety Sep vs Jun-Aug within section groups, diff = Sep - JunAug, Newcombe 95% CI")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))

    st = StratifiedTable(tables)
    or_lo, or_hi = st.oddsratio_pooled_confint()
    null_test = st.test_null_odds()
    homog = st.test_equal_odds()
    # Crude (unstratified) odds ratio for comparison
    crude = np.sum(tables, axis=0)
    crude_or = (crude[0, 0] * crude[1, 1]) / (crude[0, 1] * crude[1, 0])
    etitle("Mantel-Haenszel pooled odds ratio (Sep vs Jun-Aug), stratified by section group")
    print(pd.DataFrame([{
        "crude_OR": crude_or, "MH_OR": st.oddsratio_pooled, "MH_CI_low": or_lo, "MH_CI_high": or_hi,
        "CMH_p": null_test.pvalue, "Breslow-Day_p (homogeneity)": homog.pvalue,
    }]).to_string(index=False, float_format=lambda x: f"{x:.3f}"))
````

#### 3.3.6 กราฟตรวจกรอบความเสี่ยง (เชิงสำรวจ)
กราฟ 3 รูป: สัดส่วนรายสัปดาห์, ผลต่างตามกลุ่มคำ (B2), และสัดส่วนแยกตาม section (B4)

````python
# ===== Part 3.3c (team) - Chart 1: weekly risk_safety share with Wilson 95% CI =====
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from statsmodels.stats.proportion import proportion_confint

# Shared style: large fonts and one fixed color per item across all charts
plt.rcParams.update({"font.size": 14, "axes.titlesize": 16, "axes.labelsize": 14,
                     "xtick.labelsize": 11, "ytick.labelsize": 12, "legend.fontsize": 12})
COLORS = {
    "risk_safety": "#e6194b", "Jun-Aug": "#4363d8", "Sep": "#f58231", "window": "#ffe119",
    "all terms": "#e6194b", "without generic_hazard": "#911eb4", "generic_hazard only": "#3cb44b",
    "extreme only": "#000075", "tech_failure only": "#42d4f4", "info_fraud only": "#f032e6",
}
TITLE_PREFIX = "EXPLORATORY (v1, not human-validated):"

if not all(v in globals() for v in ["pop", "term_counts", "RISK_GROUPS", "is_sep", "diff_ci"]):
    print("Objects from earlier steps are missing. Re-run only our cells (FRAME_DICT, Step C, A1, B1, B2) first.")
else:
    pub = pd.to_datetime(pop["webPublicationDate"], utc=True).dt.tz_convert(None)
    week_start = (pub.dt.normalize() - pd.to_timedelta(pub.dt.weekday, unit="D"))   # ISO week Monday
    wk = pd.DataFrame({"week": week_start, "flag": pop["flag_risk_safety"].astype(int)})
    weekly = wk.groupby("week").agg(n=("flag", "size"), k=("flag", "sum")).reset_index()
    weekly["share"] = weekly["k"] / weekly["n"] * 100
    lo, hi = proportion_confint(weekly["k"], weekly["n"], alpha=0.05, method="wilson")
    weekly["lo"], weekly["hi"] = lo * 100, hi * 100
    weekly["partial"] = weekly["week"] == pd.Timestamp("2026-09-28")   # Sep 28-30 only

    jja_pooled = pop.loc[~is_sep, "flag_risk_safety"].mean() * 100
    full, part = weekly[~weekly["partial"]], weekly[weekly["partial"]]
    col = COLORS["risk_safety"]

    fig, ax = plt.subplots(figsize=(17, 7.5))
    ax.axvspan(pd.Timestamp("2026-09-10"), pd.Timestamp("2026-09-24"), color=COLORS["window"], alpha=0.35,
               label="Sep 10 - Sep 24 window")
    ax.fill_between(weekly["week"], weekly["lo"], weekly["hi"], color=col, alpha=0.15, label="Wilson 95% CI")
    ax.plot(full["week"], full["share"], "-o", color=col, lw=2.5, ms=9, label="Weekly share (full week)")
    ax.plot(weekly["week"].iloc[-2:], weekly["share"].iloc[-2:], "--", color=col, lw=2)
    ax.plot(part["week"], part["share"], "o", ms=11, markerfacecolor="white", markeredgecolor=col,
            markeredgewidth=2.5, label="Weekly share (partial: Sep 28-30)")
    ax.axhline(jja_pooled, ls="--", color=COLORS["Jun-Aug"], lw=2, label="Jun-Aug pooled share")
    for x, y in zip(weekly["week"], weekly["share"]):
        ax.annotate(f"{y:.0f}%", (x, y), textcoords="offset points", xytext=(0, 12), ha="center",
                    fontsize=12, fontweight="bold", color=col)
    ax.set_xticks(weekly["week"])
    ax.set_xticklabels([f"{d:%b %d}\nn={n}" for d, n in zip(weekly["week"], weekly["n"])])
    ax.set_ylabel("Articles flagged risk_safety (%)")
    ax.set_xlabel("ISO week (Monday date) and number of articles")
    ax.set_ylim(0, max(weekly["hi"].max(), weekly["share"].max()) + 10)
    ax.set_title(f"{TITLE_PREFIX} weekly share of risk_safety flags, Jun 1 - Sep 30, 2026")
    ax.legend(loc="upper left", ncol=2, frameon=True)
    ax.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()
    print("Chart 1 plots the weekly percentage of articles flagged risk_safety (v1, lede >= 2) with Wilson 95% CI, "
          "the Sep 10-24 window, the Jun-Aug pooled share, and the weekly n under each Monday date.")
````

````python
# ===== Part 3.3c (team) - Chart 2: Sep minus Jun-Aug difference by term-group variant =====
if "COLORS" not in globals() or "term_counts" not in globals():
    print("Run Chart 1 and our earlier cells first.")
else:
    all_terms = [t for g in RISK_GROUPS.values() for t in g]
    chart2_variants = {
        "all terms": all_terms,
        "without generic_hazard": [t for t in all_terms if t not in RISK_GROUPS["generic_hazard"]],
        "generic_hazard only": RISK_GROUPS["generic_hazard"],
        "extreme only": RISK_GROUPS["extreme"],
        "tech_failure only": RISK_GROUPS["tech_failure"],
        "info_fraud only": RISK_GROUPS["info_fraud"],
    }
    names, diffs, los, his = [], [], [], []
    for vname, terms in chart2_variants.items():
        flag = term_counts[terms].sum(axis=1) >= FRAME_THRESHOLD   # same v1 rule
        d, lo_, hi_ = diff_ci(int((flag & is_sep).sum()), int(is_sep.sum()),
                              int((flag & ~is_sep).sum()), int((~is_sep).sum()))
        names.append(vname); diffs.append(d); los.append(lo_); his.append(hi_)

    diffs, los, his = np.array(diffs), np.array(los), np.array(his)
    ypos = np.arange(len(names))[::-1]
    fig, ax = plt.subplots(figsize=(13, 7))
    ax.barh(ypos, diffs, color=[COLORS[n] for n in names], height=0.6, alpha=0.9)
    ax.errorbar(diffs, ypos, xerr=[diffs - los, his - diffs], fmt="none", ecolor="black", elinewidth=2, capsize=6)
    ax.axvline(0, color="black", lw=1.5)
    for y, d, h in zip(ypos, diffs, his):
        ax.text(max(h, 0) + 0.6, y, f"{d:+.0f}", va="center", fontsize=13, fontweight="bold")
    ax.set_yticks(ypos)
    ax.set_yticklabels(names)
    ax.set_xlabel("Difference in risk_safety share, Sep minus Jun-Aug (percentage points)")
    ax.set_xlim(min(los.min(), 0) - 3, his.max() + 5)
    ax.set_title(f"{TITLE_PREFIX} Sep vs Jun-Aug difference by risk term group\n(Newcombe 95% CI)")
    ax.grid(axis="x", alpha=0.3)
    plt.tight_layout()
    plt.show()
    print("Chart 2 plots the September minus June-August difference (pp) in the risk_safety flag for six term-group "
          "variants, with Newcombe 95% CI error bars and a zero line.")
````

#### 3.3.7 ตรวจความผันผวนรายสัปดาห์ (เชิงสำรวจ)
ยังเป็นการวิเคราะห์เชิงสำรวจ (ไม่ได้วางแผนไว้ล่วงหน้า) ช่วง 10-24 ก.ย. เลือกหลังจากดูข้อมูลแล้ว

````python
# ===== Part 3.3c (team) - C1: how noisy are the weekly risk_safety shares? =====
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency, binomtest
from statsmodels.stats.multitest import multipletests

EXPL = "EXPLORATORY (not pre-registered; v1 dictionary not human-validated)"
C_OK = all(v in globals() for v in ["pop", "is_sep", "diff_ci", "term_counts", "RISK_GROUPS", "COLORS"])
if not C_OK:
    print("Objects from earlier steps are missing. Re-run only our cells (FRAME_DICT, Step C, A1, B1, B2, Chart 1) first.")
else:
    pub_dt = pd.to_datetime(pop["webPublicationDate"], utc=True).dt.tz_convert(None)
    pop_day = pub_dt.dt.normalize()
    pop_week = pop_day - pd.to_timedelta(pub_dt.dt.weekday, unit="D")   # ISO week Monday
    wk_all = pd.DataFrame({"week": pop_week, "flag": pop["flag_risk_safety"].astype(int)})
    wk_tab = wk_all.groupby("week").agg(flagged=("flag", "sum"), n=("flag", "size"))

    # a) homogeneity across the 13 full weeks Jun 01 - Aug 24 (weeks end Aug 30)
    base_weeks = pd.date_range("2026-06-01", "2026-08-24", freq="7D")
    base = wk_tab.loc[base_weeks]
    table = np.column_stack([base["flagged"], base["n"] - base["flagged"]])
    chi2, p, dof, expected = chi2_contingency(table, correction=False)
    base_share = base["flagged"].sum() / base["n"].sum()
    print(f"\n=== {EXPL} | C1a: chi-square homogeneity of weekly shares, 13 weeks Jun 01 - Aug 24 ===")
    print(f"chi2 = {chi2:.2f}, df = {dof}, p = {p:.3f} | expected counts < 5: {int((expected < 5).sum())} of {expected.size} cells")
    print(f"Pooled share of the 13 weeks: {base['flagged'].sum()}/{base['n'].sum()} = {base_share * 100:.1f}%")

    # b) each week from Aug 31 to Sep 28 vs the 13-week pooled share (exact binomial)
    test_weeks = pd.date_range("2026-08-31", "2026-09-28", freq="7D")
    rows = []
    for w in test_weeks:
        k, n = int(wk_tab.at[w, "flagged"]), int(wk_tab.at[w, "n"])
        end = w + pd.Timedelta(days=6) if w != pd.Timestamp("2026-09-28") else pd.Timestamp("2026-09-30")
        label = f"{w:%b %d} - {end:%b %d}"
        if w.month != end.month:
            label += " (spans Aug/Sep)"
        if w == pd.Timestamp("2026-09-28"):
            label += " (partial)"
        rows.append({"week": label, "flagged": k, "n": n, "share_%": k / n * 100,
                     "p_binom": binomtest(k, n, base_share, alternative="two-sided").pvalue})
    c1b = pd.DataFrame(rows)
    c1b["p_holm"] = multipletests(c1b["p_binom"], method="holm")[1]
    print(f"\n=== {EXPL} | C1b: weekly share vs 13-week pooled share {base_share * 100:.1f}% (exact two-sided binomial, Holm across 5 weeks) ===")
    print(c1b.to_string(index=False, float_format=lambda x: f"{x:.4f}" if x < 1 else f"{x:.1f}"))
````

````python
# ===== Part 3.3c (team) - C2: is the September result driven by one week? =====
if not C_OK:
    print("Earlier objects missing - C2 skipped.")
else:
    jja_mask = ~is_sep   # same comparison group as A3: 2026-06-01 to 2026-08-31
    k2, n2 = int(pop.loc[jja_mask, "flag_risk_safety"].sum()), int(jja_mask.sum())
    versions = {
        "all of September": is_sep,
        "September without Sep 14-20": is_sep & ~pop_day.between("2026-09-14", "2026-09-20"),
        "September without Sep 10-24": is_sep & ~pop_day.between("2026-09-10", "2026-09-24"),
    }
    rows = []
    for vname, m in versions.items():
        k1, n1 = int(pop.loc[m, "flag_risk_safety"].sum()), int(m.sum())
        d, lo, hi = diff_ci(k1, n1, k2, n2)
        rows.append({"version": vname, "Sep_flagged": k1, "Sep_n": n1, "Sep_%": k1 / n1 * 100,
                     "JunAug_%": k2 / n2 * 100, "diff_pp": d, "CI_low": lo, "CI_high": hi})
    print(f"\n=== {EXPL} | C2: Sep minus Jun-Aug (n={n2}) for risk_safety, Newcombe 95% CI ===")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
````

````python
# ===== Part 3.3c (team) - C3: Chart 1 redrawn (wider, larger ticks, span label) =====
import matplotlib.pyplot as plt
from statsmodels.stats.proportion import proportion_confint

if not C_OK:
    print("Earlier objects missing - chart skipped.")
else:
    weekly = wk_tab.reset_index().rename(columns={"flagged": "k"})
    weekly["share"] = weekly["k"] / weekly["n"] * 100
    lo, hi = proportion_confint(weekly["k"], weekly["n"], alpha=0.05, method="wilson")
    weekly["lo"], weekly["hi"] = lo * 100, hi * 100
    weekly["partial"] = weekly["week"] == pd.Timestamp("2026-09-28")
    jja_pooled = pop.loc[~is_sep, "flag_risk_safety"].mean() * 100
    full, part = weekly[~weekly["partial"]], weekly[weekly["partial"]]
    col = COLORS["risk_safety"]

    fig, ax = plt.subplots(figsize=(20, 8))
    ax.axvspan(pd.Timestamp("2026-09-10"), pd.Timestamp("2026-09-24"), color=COLORS["window"], alpha=0.35,
               label="Sep 10-24 (window chosen after viewing the data)")
    ax.fill_between(weekly["week"], weekly["lo"], weekly["hi"], color=col, alpha=0.15, label="Wilson 95% CI")
    ax.plot(full["week"], full["share"], "-o", color=col, lw=2.5, ms=9, label="Weekly share (full week)")
    ax.plot(weekly["week"].iloc[-2:], weekly["share"].iloc[-2:], "--", color=col, lw=2)
    ax.plot(part["week"], part["share"], "o", ms=11, markerfacecolor="white", markeredgecolor=col,
            markeredgewidth=2.5, label="Weekly share (partial: Sep 28-30)")
    ax.axhline(jja_pooled, ls="--", color=COLORS["Jun-Aug"], lw=2, label="Jun-Aug pooled share")
    for x, y in zip(weekly["week"], weekly["share"]):
        ax.annotate(f"{y:.0f}%", (x, y), textcoords="offset points", xytext=(0, 12), ha="center",
                    fontsize=12, fontweight="bold", color=col)
    ax.set_xticks(weekly["week"])
    ax.set_xticklabels([f"{d:%b %d}\nn={n}" for d, n in zip(weekly["week"], weekly["n"])], fontsize=12)
    ax.set_ylabel("Articles flagged risk_safety (%)")
    ax.set_xlabel("ISO week (Monday date) and number of articles")
    ax.set_ylim(0, max(weekly["hi"].max(), weekly["share"].max()) + 10)
    ax.set_title(f"{TITLE_PREFIX} weekly share of risk_safety flags, Jun 1 - Sep 30, 2026")
    ax.legend(loc="upper left", ncol=2, frameon=True)
    ax.grid(axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()
    print("Chart 1 (redrawn) plots the weekly percentage of articles flagged risk_safety with Wilson 95% CI, the "
          "post-hoc Sep 10-24 window, the Jun-Aug pooled share, and the weekly n under each Monday date.")
````

### 3.3.8 ตรวจผลกรอบกฎหมาย: ข่าว vs ความเห็น (เชิงสำรวจ)
การตรวจในส่วนนี้ **เพิ่มขึ้นหลังจากเห็นผล A5** (law_regulation ใน news สูงกว่า opinion) จึงไม่ได้วางแผนไว้ล่วงหน้า และรายงานผลของทุก variant ที่ลองไว้ทั้งหมด ไม่เลือกรายงานเฉพาะบางแบบ (พจนานุกรม v1 และกติกา >= 2 ไม่เปลี่ยน)

````python
# ===== Part 3.3d (team) - D1: which law_regulation terms drive the flag =====
import re
import numpy as np
import pandas as pd
from statsmodels.stats.multitest import multipletests
from statsmodels.stats.contingency_tables import StratifiedTable

ROB = "ROBUSTNESS (added after seeing the news-vs-opinion result; v1 dictionary not human-validated)"


def rtitle(name):
    """Print a table title that always carries the robustness prefix."""
    print(f"\n=== {ROB} | {name} ===")


D_OK = all(v in globals() for v in ["pop", "FRAME_DICT", "item_to_regex", "diff_ci", "two_sample_p",
                                    "body_full", "FRAME_PATTERNS", "FRAME_THRESHOLD"]) and "has_ai_term" in pop.columns
if not D_OK:
    print("Objects from earlier steps are missing. Re-run only our cells (FRAME_DICT, Step C, A1, A5, A6) first.")
else:
    no = pop[pop["content_group"].isin(["news", "opinion"])].copy()
    is_news = no["content_group"] == "news"
    LAW_TERMS = FRAME_DICT["law_regulation"]
    law_rx = {t: re.compile(r"\b(?:" + item_to_regex(t) + r")\b", re.IGNORECASE) for t in LAW_TERMS}
    law_counts = pd.DataFrame({t: no["lede"].map(lambda s, p=p: len(p.findall(s))) for t, p in law_rx.items()},
                              index=no.index)
    print(f"Population: news {int(is_news.sum())}, opinion {int((~is_news).sum())}")
    print("Check: per-term counts add up to the v1 law_regulation count for every article:",
          bool((law_counts.sum(axis=1) == no["count_law_regulation"]).all()))

    flagged = no["flag_law_regulation"]
    nf_news, nf_op = int((flagged & is_news).sum()), int((flagged & ~is_news).sum())
    rows = []
    for t in LAW_TERMS:
        has = law_counts[t] > 0
        rows.append({"term": t, "news_articles": int((has & is_news).sum()), "opinion_articles": int((has & ~is_news).sum()),
                     "total": int(has.sum()),
                     "news_%_of_flagged": (has & flagged & is_news).sum() / nf_news * 100 if nf_news else np.nan,
                     "opinion_%_of_flagged": (has & flagged & ~is_news).sum() / nf_op * 100 if nf_op else np.nan})
    d1_res = pd.DataFrame(rows).sort_values(["total", "news_articles"], ascending=False)
    rtitle(f"law_regulation terms in the lede (news flagged {nf_news}, opinion flagged {nf_op})")
    print(d1_res.to_string(index=False, float_format=lambda x: f"{x:.1f}"))
````

````python
# ===== Part 3.3d (team) - D2: law_regulation dictionary variants (rule unchanged: >= 2 in the lede) =====
def news_vs_opinion(flag, mask=None):
    """Share in news and opinion, news - opinion difference (pp) with Newcombe CI, p-value and n."""
    f = flag if mask is None else flag[mask]
    g = is_news if mask is None else is_news[mask]
    k1, n1 = int((f & g).sum()), int(g.sum())
    k2, n2 = int((f & ~g).sum()), int((~g).sum())
    d, lo, hi = diff_ci(k1, n1, k2, n2)
    p, test = two_sample_p(k1, n1, k2, n2)
    return {"news_%": k1 / n1 * 100, "opinion_%": k2 / n2 * 100, "diff_pp": d, "CI_low": lo, "CI_high": hi,
            "p": p, "test": test, "n_news": n1, "n_opinion": n2}


if not D_OK:
    print("Earlier objects missing - D2 skipped.")
else:
    LAW_VARIANTS = {
        "all_terms (v1)": list(LAW_TERMS),
        "without_generic": [t for t in LAW_TERMS if t not in ["law", "laws", "ban", "banned", "bans"]],
        "without_institutions": [t for t in LAW_TERMS if t not in ["court", "courts", "judge", "judges", "Senate",
                                                                    "Congress", "parliament", "attorney general"]],
        "core_only": [r"regulat\w*", r"legislat\w*", r"lawmaker\w*", r"lawsuit\w*", "sued", "antitrust", "FTC",
                      "executive order", "EU AI Act", "compliance", "oversight", "probe", "investigation",
                      "investigations", "ruling"],
    }
    assert all(t in LAW_TERMS for v in LAW_VARIANTS.values() for t in v), "variant terms must come from the v1 dictionary"
    rows = []
    for vname, terms in LAW_VARIANTS.items():
        flag = law_counts[terms].sum(axis=1) >= FRAME_THRESHOLD
        rows.append({"variant": vname, "n_terms": len(terms), **news_vs_opinion(flag)})
    rtitle("law_regulation dictionary variants, diff = news - opinion, Newcombe 95% CI (unadjusted p, no variant selection)")
    print(pd.DataFrame(rows).drop(columns=["n_news", "n_opinion"]).to_string(
        index=False, float_format=lambda x: f"{x:.3f}" if abs(x) < 1 else f"{x:.1f}"))
````

````python
# ===== Part 3.3d (team) - D3: threshold and text variants (all_terms dictionary) =====
if not D_OK:
    print("Earlier objects missing - D3 skipped.")
else:
    body_law = body_full.loc[no.index].map(lambda t: len(FRAME_PATTERNS["law_regulation"].findall(t)))
    variants = {
        "v1 (lede >= 2)": (no["count_law_regulation"] >= 2, None),
        "lede >= 1": (no["count_law_regulation"] >= 1, None),
        "lede >= 3": (no["count_law_regulation"] >= 3, None),
        "full bodyText >= 2": (body_law >= 2, None),
        "v1, AI term in headline/standfirst": (no["flag_law_regulation"], no["has_ai_term"]),
    }
    rows = []
    for vname, (flag, mask) in variants.items():
        r = news_vs_opinion(flag, mask)
        rows.append({"variant": vname, "news_%": r["news_%"], "opinion_%": r["opinion_%"], "diff_pp": r["diff_pp"],
                     "CI_low": r["CI_low"], "CI_high": r["CI_high"], "n_news": r["n_news"], "n_opinion": r["n_opinion"]})
    rtitle("law_regulation threshold/text variants, diff = news - opinion, Newcombe 95% CI")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
````

````python
# ===== Part 3.3d (team) - D4: composition and stratification =====
if not D_OK:
    print("Earlier objects missing - D4 skipped.")
else:
    sec_grp = no["sectionId"].where(no["sectionId"].isin(["technology", "commentisfree", "us-news"]), "other")
    SEC_ORDER = ["technology", "commentisfree", "us-news", "other"]

    # a) composition
    comp = pd.crosstab(sec_grp, no["content_group"]).reindex(SEC_ORDER).fillna(0).astype(int)
    comp_pct = (comp / comp.sum() * 100).round(1)
    rtitle("D4a: section-group composition of news and opinion (count and % of group)")
    print(pd.concat([comp.add_suffix("_n"), comp_pct.add_suffix("_%")], axis=1).to_string())

    # b) within section groups
    flag = no["flag_law_regulation"]
    rows, tables = [], []
    for g in SEC_ORDER:
        m = sec_grp == g
        n_n, n_o = int((m & is_news).sum()), int((m & ~is_news).sum())
        if n_n >= 20 and n_o >= 20:
            r = news_vs_opinion(flag, m)
            rows.append({"section_group": g, "n_news": n_n, "n_opinion": n_o, "news_%": r["news_%"],
                         "opinion_%": r["opinion_%"], "diff_pp": f"{r['diff_pp']:+.1f} [{r['CI_low']:.1f}, {r['CI_high']:.1f}]"})
            kn, ko = int((flag & m & is_news).sum()), int((flag & m & ~is_news).sum())
            tables.append(np.array([[kn, n_n - kn], [ko, n_o - ko]]))
        else:
            rows.append({"section_group": g, "n_news": n_n, "n_opinion": n_o, "news_%": np.nan,
                         "opinion_%": np.nan, "diff_pp": "n too small"})
    rtitle("D4b: law_regulation news - opinion within section groups (both n >= 20), Newcombe 95% CI")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}", na_rep="-"))

    # c) within periods
    rows = []
    for per in ["Jun-Aug", "Sep"]:
        m = no["period"] == per
        r = news_vs_opinion(flag, m)
        rows.append({"period": per, "n_news": r["n_news"], "n_opinion": r["n_opinion"], "news_%": r["news_%"],
                     "opinion_%": r["opinion_%"], "diff_pp": r["diff_pp"], "CI_low": r["CI_low"], "CI_high": r["CI_high"]})
    rtitle("D4c: law_regulation news - opinion within period, Newcombe 95% CI")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))

    # d) Mantel-Haenszel across eligible section groups
    rtitle(f"D4d: Mantel-Haenszel pooled odds ratio (news vs opinion), {len(tables)} eligible section group(s)")
    if len(tables) >= 2:
        st = StratifiedTable(tables)
        lo, hi = st.oddsratio_pooled_confint()
        print(pd.DataFrame([{"MH_OR": st.oddsratio_pooled, "CI_low": lo, "CI_high": hi,
                             "Breslow-Day_p": st.test_equal_odds().pvalue}]).to_string(index=False, float_format=lambda x: f"{x:.3f}"))
    else:
        print("Fewer than 2 eligible section groups - pooled odds ratio and Breslow-Day test not computed.")
````

````python
# ===== Part 3.3d (team) - D5: six-frame table with the without_generic law variant =====
if not D_OK:
    print("Earlier objects missing - D5 skipped.")
else:
    law_wg = law_counts[LAW_VARIANTS["without_generic"]].sum(axis=1) >= FRAME_THRESHOLD
    rows = []
    for fr in FRAMES:
        flag = law_wg if fr == "law_regulation" else no[f"flag_{fr}"]
        r = news_vs_opinion(flag)
        rows.append({"frame": fr + (" (without_generic)" if fr == "law_regulation" else ""), **r})
    d5_res = pd.DataFrame(rows)
    d5_res["p_holm"] = multipletests(d5_res["p"], method="holm")[1]
    d5_res["diff_pp [95% CI]"] = [f"{d:+.1f} [{l:.1f}, {h:.1f}]" for d, l, h in
                                  zip(d5_res["diff_pp"], d5_res["CI_low"], d5_res["CI_high"])]
    rtitle(f"D5: news (n={int(is_news.sum())}) vs opinion (n={int((~is_news).sum())}), law_regulation = without_generic, Holm across six frames")
    print(d5_res[["frame", "news_%", "opinion_%", "diff_pp [95% CI]", "p", "p_holm"]].to_string(
        index=False, float_format=lambda x: f"{x:.3f}" if x < 1 else f"{x:.1f}"))
````

### 3.3.9 เทคนิคเสริม 1: บริบทคำ คำเด่น sentiment (เชิงสำรวจ)
* เทคนิคในส่วนนี้ **เพิ่มขึ้นหลังจากเห็นผลหลักแล้ว** และรายงานผลของทุก variant ที่รันไว้ทั้งหมด
* ผลทั้งหมดเป็นการ **บรรยายเชิงสำรวจ (descriptive) ไม่ใช่การยืนยันสมมติฐาน (not confirmatory)**
* ข้อความสั้น ๆ ที่พิมพ์ในขั้น KWIC ใช้เพื่อการวิเคราะห์ในชั้นเรียนเท่านั้น ห้ามนำไปเผยแพร่ซ้ำ

````python
# ===== Part 3.3e (team) - E1: preprocessing of the ledes (tokens, stop words, optional lemmas) =====
import re
import numpy as np
import pandas as pd
from collections import Counter
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

# Lemmatization is switched off on purpose so that the table does not depend on whether WordNet is installed; the downstream results (E2-E5) use the original word forms
USE_LEMMAS = False

EXPL2 = "EXPLORATORY (added after seeing the main results; v1 dictionary not human-validated)"


def xtitle(name):
    """Print a table title that always carries the exploratory prefix."""
    print(f"\n=== {EXPL2} | {name} ===")


E_NEEDED = ["pop", "corpus", "FRAMES", "FRAME_PATTERNS", "FRAME_DICT", "clean_html", "diff_ci", "AI_TERMS"]
E_MISSING = [v for v in E_NEEDED if v not in globals()]
E_OK = not E_MISSING
if not E_OK:
    print("Missing variables (runtime may have been reset):", E_MISSING, "- stopped.")
else:
    # Lower-case words made of letters, apostrophes and hyphens, length >= 2 (no leading/trailing ' or -)
    TOKEN_RE = re.compile(r"[a-z][a-z'\-]*[a-z]")
    STOP = set(ENGLISH_STOP_WORDS)

    def tokenize(text):
        text = str(text).lower().replace("’", "'").replace("‘", "'")
        return TOKEN_RE.findall(text)

    # Optional lemmatization: no download here; skip if WordNet data is not available
    LEMMATIZER = None
    if USE_LEMMAS:
        try:
            from nltk.stem import WordNetLemmatizer
            _lm = WordNetLemmatizer()
            _lm.lemmatize("tests")
            LEMMATIZER = _lm
        except Exception:
            print("lemmatization skipped")
    else:
        print("lemmatization skipped")

    ep = pop.copy()   # work on a copy; pop is not modified
    ep["tokens"] = ep["lede"].map(tokenize)
    if USE_LEMMAS and LEMMATIZER is not None:
        ep["lemmas"] = ep["tokens"].map(lambda ts: [LEMMATIZER.lemmatize(t) for t in ts])
    token_col = "tokens"   # the table always uses the original word forms
    is_sep_e = ep["period"] == "Sep"

    xtitle(f"E1: preprocessing summary (n = {len(ep)} ledes, unit = original word forms (lemmatization off by design))")
    print(f"Mean tokens per lede: {ep['tokens'].map(len).mean():.1f}")
    freq_all = Counter(t for ts in ep["tokens"] for t in ts if t not in STOP)
    print("Top 20 non-stop-word tokens (counts):")
    print(pd.DataFrame(freq_all.most_common(20), columns=["token", "count"]).to_string(index=False))
````

````python
# ===== Part 3.3e (team) - E2: broad risk terms - prevalence, collocates and KWIC =====
if not E_OK:
    print("E1 did not complete - E2 skipped.")
else:
    RISK_TERM_GROUPS = {
        "safety": {"safety", "unsafe"},
        "threat": {"threat", "threats", "threaten", "threatens"},
        "danger": {"danger", "dangers", "dangerous"},
        "risk": {"risk", "risks", "risky"},
        "harm": {"harm", "harms", "harmful"},
    }
    flag_r = ep["flag_risk_safety"]
    n_flag_sep, n_flag_jja = int((flag_r & is_sep_e).sum()), int((flag_r & ~is_sep_e).sum())

    # 2a. prevalence of each group
    rows = []
    for g, terms in RISK_TERM_GROUPS.items():
        has = ep["tokens"].map(lambda ts, s=terms: any(t in s for t in ts))
        rows.append({"group": g, "Sep_ledes": int((has & is_sep_e).sum()), "JunAug_ledes": int((has & ~is_sep_e).sum()),
                     "Sep_%_of_flagged": (has & flag_r & is_sep_e).sum() / n_flag_sep * 100,
                     "JunAug_%_of_flagged": (has & flag_r & ~is_sep_e).sum() / n_flag_jja * 100})
    xtitle(f"E2a: ledes containing each term group (Sep n={int(is_sep_e.sum())}, flagged {n_flag_sep}; "
           f"Jun-Aug n={int((~is_sep_e).sum())}, flagged {n_flag_jja})")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))

    # 2b. collocates within +-3 tokens (stop words removed), lift vs corpus unigram frequencies
    uni = Counter(t for ts in ep["tokens"] for t in ts if t not in STOP)
    N_uni = sum(uni.values())
    occurrences = {g: [] for g in RISK_TERM_GROUPS}   # (row index, position, period)
    for idx, ts, per in zip(ep.index, ep["tokens"], ep["period"]):
        for pos, t in enumerate(ts):
            for g, terms in RISK_TERM_GROUPS.items():
                if t in terms:
                    occurrences[g].append((idx, pos, per))
    for g, terms in RISK_TERM_GROUPS.items():
        c_all, c_sep, c_jja, window_total = Counter(), Counter(), Counter(), 0
        for idx, pos, per in occurrences[g]:
            ts = ep.at[idx, "tokens"]
            window = [w for w in ts[max(0, pos - 3):pos] + ts[pos + 1:pos + 4] if w not in STOP and w not in terms]
            window_total += len(window)
            c_all.update(window)
            (c_sep if per == "Sep" else c_jja).update(window)
        rows = []
        for w, cnt in c_all.items():
            if cnt < 4:
                continue
            expected = window_total * uni[w] / N_uni
            rows.append({"collocate": w, "count": cnt, "Sep": c_sep[w], "JunAug": c_jja[w],
                         "lift": cnt / expected if expected else np.nan})
        tab = pd.DataFrame(rows, columns=["collocate", "count", "Sep", "JunAug", "lift"])
        tab = tab.sort_values(["count", "lift"], ascending=False).head(15)
        xtitle(f"E2b: top collocates of '{g}' (+-3 tokens, stop words removed, min total 4; "
               f"{len(occurrences[g])} occurrences)")
        print(tab.to_string(index=False, float_format=lambda x: f"{x:.2f}") if len(tab) else "no collocate with total count >= 4")

    # 2c. KWIC excerpts (max 5 tokens on each side), 5 from Sep and 5 from Jun-Aug per group
    rng_kwic = np.random.default_rng(42)
    for g in RISK_TERM_GROUPS:
        picked = []
        for per in ["Sep", "Jun-Aug"]:
            occ = [o for o in occurrences[g] if o[2] == per]
            k = min(5, len(occ))
            if k:
                picked += [occ[i] for i in rng_kwic.choice(len(occ), size=k, replace=False)]
        rows = []
        for idx, pos, per in picked:
            ts = ep.at[idx, "tokens"]
            excerpt = " ".join(ts[max(0, pos - 5):pos] + [f"[{ts[pos]}]"] + ts[pos + 1:pos + 6])
            rows.append({"id": ep.at[idx, "id"], "date": str(ep.at[idx, "webPublicationDate"])[:10],
                         "period": per, "excerpt": excerpt})
        xtitle(f"E2c: KWIC excerpts for '{g}' (seed 42; classroom analysis only, do not redistribute)")
        with pd.option_context("display.max_colwidth", 120, "display.width", 250):
            print(pd.DataFrame(rows).to_string(index=False) if rows else "no occurrences")
        print("to be read by humans: label the sense of each excerpt")
````

````python
# ===== Part 3.3e (team) - E3: distinctive words (log-odds ratio, informative Dirichlet prior; Monroe et al. 2008) =====
from sklearn.feature_extraction.text import CountVectorizer

if not E_OK:
    print("E1 did not complete - E3 skipped.")
else:
    vec = CountVectorizer(stop_words="english", token_pattern=r"(?u)\b[a-zA-Z]{3,}\b",
                          ngram_range=(1, 2), binary=True, min_df=5)
    X_lede = vec.fit_transform(ep["lede"])
    vocab = np.array(vec.get_feature_names_out())
    print(f"Vocabulary (document-level presence, min_df=5, uni+bigrams): {len(vocab)} terms")

    def log_odds_z(mask_a, mask_b, prior_total=1000.0):
        """Log-odds difference with informative Dirichlet prior (pooled counts scaled to prior_total)."""
        y_a = np.asarray(X_lede[np.asarray(mask_a)].sum(axis=0)).ravel().astype(float)
        y_b = np.asarray(X_lede[np.asarray(mask_b)].sum(axis=0)).ravel().astype(float)
        pooled = y_a + y_b
        keep = pooled > 0
        y_a, y_b, pooled, terms = y_a[keep], y_b[keep], pooled[keep], vocab[keep]
        alpha = pooled / pooled.sum() * prior_total
        a0 = alpha.sum()
        n_a, n_b = y_a.sum(), y_b.sum()
        delta = (np.log((y_a + alpha) / (n_a + a0 - y_a - alpha))
                 - np.log((y_b + alpha) / (n_b + a0 - y_b - alpha)))
        z = delta / np.sqrt(1.0 / (y_a + alpha) + 1.0 / (y_b + alpha))
        return pd.DataFrame({"term": terms, "docs_A": y_a.astype(int), "docs_B": y_b.astype(int), "z": z})

    pub_e = pd.to_datetime(ep["webPublicationDate"], utc=True).dt.tz_convert(None).dt.normalize()
    in_window = pub_e.between("2026-09-10", "2026-09-24")
    comparisons = [
        ("(a) Sep (A) vs Jun-Aug (B), all articles", is_sep_e, ~is_sep_e),
        ("(b) Sep (A) vs Jun-Aug (B), risk_safety-flagged only", is_sep_e & ep["flag_risk_safety"], ~is_sep_e & ep["flag_risk_safety"]),
        ("(c) Sep 10-24 (A; window chosen after viewing the data) vs all other days (B)", in_window, ~in_window),
    ]
    risk_rx = FRAME_PATTERNS["risk_safety"]
    for label, ma, mb in comparisons:
        res_lo = log_odds_z(ma, mb)
        top_a = res_lo.sort_values("z", ascending=False).head(25)
        top_b = res_lo.sort_values("z", ascending=True).head(25)
        for side, tab in [("A", top_a), ("B", top_b)]:
            xtitle(f"E3 {label}: top 25 terms favouring {side} (n_A={int(np.asarray(ma).sum())}, n_B={int(np.asarray(mb).sum())})")
            print(tab.to_string(index=False, float_format=lambda x: f"{x:.2f}"))
        if label.startswith("(b)"):
            print(f"\nTerms matching a risk_safety dictionary pattern: favouring A {int(top_a['term'].map(lambda t: bool(risk_rx.search(t))).sum())}/25, "
                  f"favouring B {int(top_b['term'].map(lambda t: bool(risk_rx.search(t))).sum())}/25")
    print("\nNote: thousands of terms are compared, so these z-scores are descriptive only (no significance threshold).")
````

````python
# ===== Part 3.3e (team) - E4: VADER sentiment on headline + standfirst, with a risk-term masking control =====
from scipy.stats import mannwhitneyu, spearmanr
from statsmodels.stats.multitest import multipletests

SIA = None
if not E_OK:
    print("E1 did not complete - E4 skipped.")
else:
    try:
        from nltk.sentiment.vader import SentimentIntensityAnalyzer
        try:
            SIA = SentimentIntensityAnalyzer()
        except LookupError:
            # Use only a lexicon that is already installed; nothing is downloaded here
            SIA = None
            print("VADER lexicon not found: run the instructor's setup cell 0e864285 first")
    except Exception as e:
        SIA = None
        print(f"VADER lexicon is not available ({type(e).__name__}) - E4 stopped.")

if SIA is not None:
    lex = SIA.lexicon
    xtitle("E4: plain-word dictionary items that are also VADER lexicon entries")
    rows = []
    for fr in FRAMES:
        plain = [t for t in FRAME_DICT[fr] if "\\" not in t and " " not in t]
        in_lex = [t for t in plain if t.lower() in lex]
        rows.append({"frame": fr, "plain_items": len(plain), "in_VADER": len(in_lex)})
        if fr == "risk_safety":
            risk_in_lex = in_lex
    print(pd.DataFrame(rows).to_string(index=False))
    print("risk_safety items in VADER:", ", ".join(risk_in_lex))

    hs = ep["headline"].fillna("").astype(str) + " " + ep["standfirst"].map(clean_html)
    ep["compound_raw"] = hs.map(lambda t: SIA.polarity_scores(t)["compound"])
    ep["compound_masked"] = hs.map(lambda t: SIA.polarity_scores(FRAME_PATTERNS["risk_safety"].sub(" ", t))["compound"])

    rng_bs = np.random.default_rng(42)

    def boot_diff(a, b, n_boot=5000):
        """Percentile bootstrap 95% CI for mean(a) - mean(b)."""
        a, b = np.asarray(a, float), np.asarray(b, float)
        ma = a[rng_bs.integers(0, len(a), size=(n_boot, len(a)))].mean(axis=1)
        mb = b[rng_bs.integers(0, len(b), size=(n_boot, len(b)))].mean(axis=1)
        return np.percentile(ma - mb, [2.5, 97.5])

    splits = [
        ("period: Sep - Jun-Aug", is_sep_e, ~is_sep_e),
        ("risk_safety: flagged - not", ep["flag_risk_safety"], ~ep["flag_risk_safety"]),
        ("content_group: news - opinion", ep["content_group"] == "news", ep["content_group"] == "opinion"),
    ]
    rows = []
    for version in ["compound_raw", "compound_masked"]:
        for label, ma, mb in splits:
            a, b = ep.loc[ma, version], ep.loc[mb, version]
            lo_b, hi_b = boot_diff(a, b)
            rows.append({"version": version, "comparison": label, "n_A": len(a), "n_B": len(b),
                         "mean_A": a.mean(), "mean_B": b.mean(), "median_A": a.median(), "median_B": b.median(),
                         "diff_mean": a.mean() - b.mean(), "CI_low": lo_b, "CI_high": hi_b,
                         "p_MWU": mannwhitneyu(a, b, alternative="two-sided").pvalue})
    e4_res = pd.DataFrame(rows)
    e4_res["p_holm"] = multipletests(e4_res["p_MWU"], method="holm")[1]
    xtitle("E4: VADER compound (headline + standfirst); diff = A - B, percentile bootstrap 95% CI (seed 42, 5000), Holm across 6")
    with pd.option_context("display.width", 250):
        print(e4_res.to_string(index=False, float_format=lambda x: f"{x:.3f}"))

    xtitle("E4: Spearman correlation, number of risk_safety matches in the lede vs compound score")
    for version in ["compound_raw", "compound_masked"]:
        r = spearmanr(ep["count_risk_safety"], ep[version])
        print(f"  {version:<16} rho = {r.statistic:.3f}, p = {r.pvalue:.4f}")

    for label, asc in [("highest", False), ("lowest", True)]:
        top = ep.sort_values("compound_masked", ascending=asc).head(10)
        xtitle(f"E4: 10 headlines with the {label} compound_masked (to be read by a human for face validity)")
        with pd.option_context("display.max_colwidth", 110, "display.width", 250):
            print(top[["compound_masked", "content_group", "headline"]].to_string(index=False, float_format=lambda x: f"{x:.3f}"))
````

#### 3.3.9b ตรวจความสอดคล้องของ sentiment (เชิงสำรวจ)

ค่าความต่างของ sentiment ที่พิมพ์ออกมาในการรันใหม่แบบสะอาด (clean re-run) ต่างจากค่าที่รายงานไว้ในการรันครั้งแรก เซลล์ด้านล่างใช้ตรวจหาสาเหตุ

````python
# ===== Part 3.3e-check (team) - E4b: read-only notes on E4 and on the VADER resources (exploratory) =====
# The notes below come from reading the source of E4 (cell NedcO7MzIV_b); nothing from E4 is executed or printed here.
import os
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer

E4_NOTES = [
    "E4 scores: compound_raw (headline + standfirst as is) and compound_masked (each risk_safety match replaced by a space); every comparison is printed for both.",
    "E4 printed difference (diff_mean): a difference in MEANS (mean_A - mean_B). Group medians are printed, but no median difference.",
    "E4 CI: percentile bootstrap of the mean difference, 5000 resamples, one generator default_rng(42) shared by the six comparisons in order.",
    "E4 p-values: Mann-Whitney U, two-sided, then Holm adjustment across the six tests.",
    "E4 groups: (i) period Sep vs Jun-Aug, (ii) flag_risk_safety True vs False, (iii) content_group news vs opinion.",
    "E4 text unit: headline + ' ' + clean_html(standfirst); population: ep (copy of the 631-article analysis population).",
]
for note in E4_NOTES:
    print(note)

print("nltk version:", nltk.__version__)
lex_ptr = nltk.data.find("sentiment/vader_lexicon.zip")
lex_path = getattr(lex_ptr, "path", str(lex_ptr))
print("VADER lexicon file:", os.path.basename(lex_path), "| size (bytes):", os.path.getsize(lex_path))
print("VADER lexicon entries:", len(SentimentIntensityAnalyzer().lexicon))
````

````python
# ===== Part 3.3e-check (team) - E4c: VADER consistency check (exploratory; no analysis choice is changed) =====
# Recomputes the E4 comparisons with the same analyzer (SIA from E4), the same text unit and the same population (ep).
# ep is not modified: the scores are kept in local Series.
from scipy.stats import mannwhitneyu

if "ep" not in globals() or globals().get("SIA") is None:
    print("E1/E4 objects are missing - run E1 and E4 first. Check skipped.")
else:
    hs_chk = ep["headline"].fillna("").astype(str) + " " + ep["standfirst"].map(clean_html)
    scores_chk = {
        "compound_raw": hs_chk.map(lambda t: SIA.polarity_scores(t)["compound"]),
        "compound_masked": hs_chk.map(lambda t: SIA.polarity_scores(FRAME_PATTERNS["risk_safety"].sub(" ", t))["compound"]),
    }
    splits_chk = [
        ("period", "period: Sep - Jun-Aug", is_sep_e, ~is_sep_e),
        ("risk", "risk_safety: flagged - not", ep["flag_risk_safety"], ~ep["flag_risk_safety"]),
        ("news-opinion", "content_group: news - opinion", ep["content_group"] == "news", ep["content_group"] == "opinion"),
    ]

    def boot_ci_chk(a, b, stat, n_boot=5000, seed=42):
        """Percentile bootstrap 95% CI for stat(a) - stat(b); a fresh default_rng(seed) for each comparison."""
        rng = np.random.default_rng(seed)
        a, b = np.asarray(a, float), np.asarray(b, float)
        ia = rng.integers(0, len(a), size=(n_boot, len(a)))
        ib = rng.integers(0, len(b), size=(n_boot, len(b)))
        f = np.mean if stat == "mean" else np.median
        return np.percentile(f(a[ia], axis=1) - f(b[ib], axis=1), [2.5, 97.5])

    # First-run values (mean differences): copied as literals from the first run
    FIRST_RUN_E4 = {("compound_raw", "period"): -0.006, ("compound_raw", "risk"): -0.204, ("compound_raw", "news-opinion"): -0.102,
                    ("compound_masked", "period"): -0.019, ("compound_masked", "risk"): -0.139, ("compound_masked", "news-opinion"): -0.111}
    # Clean-run values: the diff_mean column printed by E4 in this run, copied as printed
    CLEAN_RUN_E4 = {("compound_raw", "period"): -0.006, ("compound_raw", "risk"): -0.204, ("compound_raw", "news-opinion"): -0.102,
                    ("compound_masked", "period"): -0.019, ("compound_masked", "risk"): -0.139, ("compound_masked", "news-opinion"): -0.111}

    rows_chk = []
    for score, s in scores_chk.items():
        for key, label, mask_a, mask_b in splits_chk:
            a, b = s[mask_a], s[mask_b]
            p_mwu = mannwhitneyu(a, b, alternative="two-sided").pvalue
            for stat in ["mean", "median"]:
                va, vb = (a.mean(), b.mean()) if stat == "mean" else (a.median(), b.median())
                lo, hi = boot_ci_chk(a, b, stat)
                rows_chk.append({"score": score, "comparison": label, "statistic": stat + " difference",
                                 "n_A": len(a), "n_B": len(b), "A": va, "B": vb, "A_minus_B": va - vb,
                                 "CI_low": lo, "CI_high": hi, "p_MWU_2sided": p_mwu,
                                 "first_run": FIRST_RUN_E4[(score, key)] if stat == "mean" else np.nan,
                                 "E4_clean_run": CLEAN_RUN_E4[(score, key)] if stat == "mean" else np.nan})
    chk_e4 = pd.DataFrame(rows_chk)
    print("E4c: VADER compound (headline + standfirst), A - B; percentile bootstrap 95% CI (seed 42, 5000, fresh generator per row); Mann-Whitney two-sided")
    print("first_run and E4_clean_run are mean differences, so they are shown on the mean rows only.")
    with pd.option_context("display.width", 250, "display.max_columns", 20):
        print(chk_e4.to_string(index=False, float_format=lambda x: f"{x:.3f}"))

    gap_first = max(abs(CLEAN_RUN_E4[k] - FIRST_RUN_E4[k]) for k in FIRST_RUN_E4)
    mean_rows = chk_e4[chk_e4["statistic"] == "mean difference"]
    gap_recomp = (mean_rows["A_minus_B"].round(3) - mean_rows["E4_clean_run"]).abs().max()
    print(f"Clean-run E4 values equal the first-run values within 0.005: {gap_first <= 0.005} (largest absolute gap {gap_first:.3f}).")
    print(f"Recomputed mean differences equal the clean-run E4 values within 0.005: {gap_recomp <= 0.005} (largest absolute gap {gap_recomp:.3f}).")

    # Step 3d: list candidate causes (facts only) when the values differ by more than 0.005
    if gap_first > 0.005:
        print("Candidate causes (facts only):")
        print(f"  1. Lexicon: nltk {nltk.__version__}, {len(SIA.lexicon)} entries in the analyzer used by E4.")
        print("  2. Masking: E4 replaces each risk_safety match with a single space (it does not remove it).")
        n_f, n_fs, n_fj = int(ep["flag_risk_safety"].sum()), int(ep.loc[is_sep_e, "flag_risk_safety"].sum()), int(ep.loc[~is_sep_e, "flag_risk_safety"].sum())
        p_f, p_fs = int(pop["flag_risk_safety"].sum()), int(pop.loc[pop["period"] == "Sep", "flag_risk_safety"].sum())
        print(f"  3. Risk flag: E4 uses ep['flag_risk_safety']: {n_f} flagged (Sep {n_fs}, Jun-Aug {n_fj}); A2 uses pop['flag_risk_safety']: {p_f} (Sep {p_fs}); expected 154 (71, 83).")
        print(f"  4. Population: ep has {len(ep)} articles (expected 631; the corpus has 634).")
        print("  5. Text unit: headline + ' ' + clean_html(standfirst) in E4 and in this check.")
    else:
        print("No gap above 0.005 - the cause checks of step 3d are not needed.")
````

````python
# ===== Part 3.3e (team) - E5: frame co-occurrence (six v1 flags, n = 631) =====
if not E_OK:
    print("E1 did not complete - E5 skipped.")
else:
    F = ep[[f"flag_{fr}" for fr in FRAMES]].astype(int)
    F.columns = FRAMES
    phi = F.corr()   # Pearson correlation of 0/1 variables = phi coefficient
    p_marg = F.mean()
    lift_m = pd.DataFrame(index=FRAMES, columns=FRAMES, dtype=float)
    for a in FRAMES:
        for b in FRAMES:
            lift_m.loc[a, b] = np.nan if a == b else ((F[a] & F[b]).mean() / (p_marg[a] * p_marg[b]))
    xtitle("E5: phi coefficients between frame flags")
    print(phi.to_string(float_format=lambda x: f"{x:.2f}"))
    xtitle("E5: lift = P(A and B) / (P(A) P(B)) (diagonal omitted)")
    print(lift_m.to_string(float_format=lambda x: f"{x:.2f}", na_rep="-"))

    risk_f = ep["flag_risk_safety"]
    for scope, m in [("all months", pd.Series(True, index=ep.index)), ("September", is_sep_e), ("June-August", ~is_sep_e)]:
        rows = []
        for fr in FRAMES:
            if fr == "risk_safety":
                continue
            fa, fb = ep.loc[m & risk_f, f"flag_{fr}"], ep.loc[m & ~risk_f, f"flag_{fr}"]
            d, lo, hi = diff_ci(int(fa.sum()), len(fa), int(fb.sum()), len(fb))
            rows.append({"frame": fr, "%_among_risk_flagged": fa.mean() * 100, "%_among_others": fb.mean() * 100,
                         "diff_pp": d, "CI_low": lo, "CI_high": hi, "n_risk": len(fa), "n_other": len(fb)})
        xtitle(f"E5: share of each frame among risk_safety-flagged vs other articles, {scope} (Newcombe 95% CI)")
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
````

### 3.3.10 เทคนิคเสริม 2: regression และกลุ่มเรื่อง (เชิงสำรวจ)
* เทคนิคในส่วนนี้ **เพิ่มขึ้นหลังจากเห็นผลหลักแล้ว** และรายงานทุกโมเดลและทุกค่า k ที่รันไว้ทั้งหมด
* ผลเป็นการบรรยายเชิงสำรวจ (descriptive) ไม่ใช่การยืนยันสมมติฐาน (not confirmatory)
* บทความไม่เป็นอิสระต่อกัน (กระจุกตามช่วงเวลาและรอบเหตุการณ์) ค่า p จึงเป็นค่าประมาณคร่าว ๆ เท่านั้น
* กลุ่ม (clusters) ไม่มีชื่อ ต้องให้คนตั้งชื่อหลังจากอ่าน headline ของแต่ละกลุ่ม

````python
# ===== Part 3.3f (team) - F1: logistic regression (statsmodels GLM, Binomial) =====
import warnings
import numpy as np
import pandas as pd
import statsmodels.api as sm
from scipy.stats import chi2 as chi2_dist

G_NEEDED = ["pop", "term_counts", "RISK_GROUPS", "FRAME_THRESHOLD", "xtitle"]
G_MISSING = [v for v in G_NEEDED if v not in globals()]
G_OK = not G_MISSING
if not G_OK:
    print("Missing variables (runtime may have been reset):", G_MISSING, "- stopped.")
else:
    lg = pop.copy()
    pub_lg = pd.to_datetime(lg["webPublicationDate"], utc=True).dt.tz_convert(None)
    lg["day"] = pub_lg.dt.normalize()
    lg["week"] = lg["day"] - pd.to_timedelta(pub_lg.dt.weekday, unit="D")   # ISO week Monday
    lg["september"] = (lg["month"] == "2026-09").astype(int)
    lg["section_group"] = lg["sectionId"].where(lg["sectionId"].isin(["technology", "commentisfree", "us-news"]), "other")
    lg["cg3"] = lg["content_group"].where(lg["content_group"].isin(["news", "opinion"]), "other/unknown")
    lg["log_words"] = np.log(pd.to_numeric(lg["wordcount"], errors="coerce"))
    # Outcome variants
    no_generic = [t for g, ts in RISK_GROUPS.items() if g != "generic_hazard" for t in ts]
    lg["y_v1"] = lg["flag_risk_safety"].astype(int)
    lg["y_nogeneric"] = (term_counts.loc[lg.index, no_generic].sum(axis=1) >= FRAME_THRESHOLD).astype(int)

    xtitle(f"F1: section_group x content_group (n = {len(lg)} articles)")
    print(pd.crosstab(lg["section_group"], lg["cg3"], margins=True, margins_name="total").to_string())

    SEC_LEVELS = ["technology", "commentisfree", "us-news"]   # reference = other
    CG_LEVELS = ["news", "opinion"]                           # reference = other/unknown

    def design(d, model, interaction=False):
        """Build the design matrix with explicit dummy columns (reference levels omitted)."""
        X = pd.DataFrame({"const": 1.0, "september": d["september"].astype(float)}, index=d.index)
        if model in ("M2", "M3"):
            for s in SEC_LEVELS:
                X[f"sec_{s}"] = (d["section_group"] == s).astype(float)
        if model == "M3":
            for g in CG_LEVELS:
                X[f"cg_{g}"] = (d["cg3"] == g).astype(float)
            X["log_words"] = d["log_words"]
        if interaction:
            for s in SEC_LEVELS:
                X[f"sep_x_{s}"] = X["september"] * X[f"sec_{s}"]
        return X

    def fit_glm(y, X, groups=None):
        """Fit a Binomial GLM; print every warning instead of hiding it."""
        with warnings.catch_warnings(record=True) as wlist:
            warnings.simplefilter("always")
            m = sm.GLM(y, X, family=sm.families.Binomial())
            res = m.fit() if groups is None else m.fit(cov_type="cluster", cov_kwds={"groups": groups})
        for w in wlist:
            print(f"  WARNING ({w.category.__name__}): {str(w.message)[:160]}")
        if hasattr(res, "converged") and not res.converged:
            print("  WARNING: model did not converge")
        return res

    def or_ci(res, name="september"):
        lo, hi = res.conf_int().loc[name]
        return np.exp(res.params[name]), np.exp(lo), np.exp(hi)

    def ame_binary(res, X, name="september"):
        """Average marginal effect of a binary variable (pp) with delta-method 95% CI."""
        b, V = res.params.values, res.cov_params().values
        X1, X0 = X.copy(), X.copy()
        X1[name], X0[name] = 1.0, 0.0
        p1 = 1 / (1 + np.exp(-(X1.values @ b)))
        p0 = 1 / (1 + np.exp(-(X0.values @ b)))
        ame = np.mean(p1 - p0)
        grad = ((p1 * (1 - p1))[:, None] * X1.values - (p0 * (1 - p0))[:, None] * X0.values).mean(axis=0)
        se = np.sqrt(grad @ V @ grad)
        return ame * 100, (ame - 1.96 * se) * 100, (ame + 1.96 * se) * 100

    # Main models on the v1 outcome
    groups_week = lg["week"].factorize()[0]
    n_clusters = len(np.unique(groups_week))
    rows, fits = [], {}
    for mname in ["M1", "M2", "M3"]:
        X = design(lg, mname)
        print(f"Fitting {mname} (ordinary SE):")
        r_ord = fit_glm(lg["y_v1"], X)
        print(f"Fitting {mname} (cluster-robust SE by ISO week):")
        r_cl = fit_glm(lg["y_v1"], X, groups=groups_week)
        fits[mname] = (r_ord, X)
        o, lo, hi = or_ci(r_ord)
        oc, loc_, hic = or_ci(r_cl)
        a, alo, ahi = ame_binary(r_ord, X)
        rows.append({"model": mname, "OR_sep": o, "CI_low": lo, "CI_high": hi,
                     "OR_CI_low_cluster": loc_, "OR_CI_high_cluster": hic,
                     "AME_pp": a, "AME_CI_low": alo, "AME_CI_high": ahi, "n": int(r_ord.nobs)})
    xtitle(f"F1: September odds ratio for risk_safety (v1); cluster-robust CI uses {n_clusters} ISO-week clusters")
    with pd.option_context("display.width", 250):
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.2f}"))
    print(f"Note: with about {n_clusters} clusters the cluster-robust standard errors are unreliable.")

    # Heterogeneity: M3 vs M3 + september x section_group
    X3, X3i = design(lg, "M3"), design(lg, "M3", interaction=True)
    print("Fitting M3 + september:section_group:")
    r3 = fits["M3"][0]
    r3i = fit_glm(lg["y_v1"], X3i)
    lr = 2 * (r3i.llf - r3.llf)
    dfd = X3i.shape[1] - X3.shape[1]
    xtitle("F1: likelihood-ratio test, M3 vs M3 + september:section_group")
    print(f"LR = {lr:.2f}, df = {dfd}, p = {chi2_dist.sf(lr, dfd):.4f}")
    rows = []
    V = r3i.cov_params()
    for s in ["other"] + SEC_LEVELS:
        m = lg["section_group"] == s
        n_sep, n_jja = int((m & (lg["september"] == 1)).sum()), int((m & (lg["september"] == 0)).sum())
        if n_sep < 20 or n_jja < 20:
            rows.append({"section_group": s, "n_Sep": n_sep, "n_JunAug": n_jja, "OR_sep": np.nan,
                         "CI_low": np.nan, "CI_high": np.nan, "note": "n < 20 in a period"})
            continue
        names = ["september"] + ([] if s == "other" else [f"sep_x_{s}"])
        est = sum(r3i.params[n] for n in names)
        var = sum(V.loc[a, b] for a in names for b in names)
        se = np.sqrt(var)
        rows.append({"section_group": s, "n_Sep": n_sep, "n_JunAug": n_jja, "OR_sep": np.exp(est),
                     "CI_low": np.exp(est - 1.96 * se), "CI_high": np.exp(est + 1.96 * se), "note": ""})
    xtitle("F1: September OR within section_group (from the interaction model, ordinary SE)")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.2f}", na_rep="-"))

    # Outcome variants (all reported)
    keep_c = ~lg["day"].between("2026-09-14", "2026-09-20")
    variants = [("(a) v1 flag", lg, "y_v1"),
                ("(b) flag without generic_hazard terms", lg, "y_nogeneric"),
                ("(c) v1 flag, Sep 14-20 removed", lg[keep_c], "y_v1")]
    rows = []
    for vname, d, ycol in variants:
        row = {"variant": vname, "n": len(d), "events": int(d[ycol].sum())}
        for mname in ["M1", "M2", "M3"]:
            print(f"Fitting {vname} / {mname}:")
            r = fit_glm(d[ycol], design(d, mname))
            o, lo, hi = or_ci(r)
            row[mname] = f"{o:.2f} [{lo:.2f}, {hi:.2f}]"
        rows.append(row)
    xtitle("F1: September OR [95% CI, ordinary SE] by outcome variant and model")
    with pd.option_context("display.width", 250):
        print(pd.DataFrame(rows).to_string(index=False))
````

````python
# ===== Part 3.3f (team) - F2a: clustering of the ledes (TF-IDF -> SVD -> k-means), k scan and reported solutions =====
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import normalize
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

if not G_OK:
    print("F1 prerequisites missing - F2 skipped.")
else:
    tfv = TfidfVectorizer(stop_words="english", ngram_range=(1, 2), min_df=3, max_df=0.5, sublinear_tf=True)
    X_tfidf = tfv.fit_transform(lg["lede"])
    tf_terms = np.array(tfv.get_feature_names_out())
    Z = normalize(TruncatedSVD(n_components=100, random_state=42).fit_transform(X_tfidf))   # cosine geometry
    print(f"TF-IDF matrix: {X_tfidf.shape[0]} ledes x {X_tfidf.shape[1]} terms; SVD to 100 dimensions, rows L2-normalized")

    scan, km_fits = [], {}
    for k in range(6, 21):
        km = KMeans(n_clusters=k, n_init=10, random_state=42).fit(Z)
        km_fits[k] = km
        scan.append({"k": k, "inertia": km.inertia_, "silhouette_cosine": silhouette_score(Z, km.labels_, metric="cosine")})
    scan = pd.DataFrame(scan)
    xtitle("F2: k-means scan, k = 6..20 (n_init=10, random_state=42)")
    print(scan.to_string(index=False, float_format=lambda x: f"{x:.4f}"))
    K_CHOSEN = int(scan[scan["k"].between(8, 16)].sort_values("silhouette_cosine", ascending=False)["k"].iloc[0])
    print(f"Chosen k (highest cosine silhouette within 8..16): {K_CHOSEN}")

    REPORT_K = list(dict.fromkeys([K_CHOSEN, 8, 12]))   # unique, chosen first
    sep_flag = lg["september"] == 1
    for k in REPORT_K:
        km = km_fits[k]
        lab_k = km.labels_
        tag = " (chosen)" if k == K_CHOSEN else ""
        xtitle(f"F2: solution k = {k}{tag}; clusters are unlabeled (to be named by humans after reading the headlines)")
        for c in range(k):
            m = lab_k == c
            mean_vec = np.asarray(X_tfidf[m].mean(axis=0)).ravel()
            top_terms = ", ".join(tf_terms[np.argsort(mean_vec)[::-1][:10]])
            centre = km.cluster_centers_[c] / np.linalg.norm(km.cluster_centers_[c])
            idx_c = np.where(m)[0]
            closest = idx_c[np.argsort(Z[idx_c] @ centre)[::-1][:5]]
            print(f"\n[cluster {c}] size={int(m.sum())} | Sep share={sep_flag[m].mean() * 100:.1f}% | "
                  f"risk_safety share={lg['y_v1'][m].mean() * 100:.1f}%")
            print(f"  top terms: {top_terms}")
            for i in closest:
                print(f"  - {str(lg['headline'].iloc[i])[:120]}")
````

````python
# ===== Part 3.3f (team) - F2b: stability, hierarchical check and concentration of risk-flagged articles =====
from sklearn.metrics import adjusted_rand_score
from sklearn.cluster import AgglomerativeClustering

if not G_OK or "K_CHOSEN" not in globals():
    print("F2a did not complete - F2b skipped.")
else:
    base_labels = km_fits[K_CHOSEN].labels_
    aris = [adjusted_rand_score(base_labels, KMeans(n_clusters=K_CHOSEN, n_init=10, random_state=s).fit(Z).labels_)
            for s in range(10)]
    xtitle(f"F2: stability of k = {K_CHOSEN} (seeds 0..9 vs seed 42)")
    print(f"Adjusted Rand index: mean = {np.mean(aris):.3f}, min = {np.min(aris):.3f}")
    ward = AgglomerativeClustering(n_clusters=K_CHOSEN, linkage="ward").fit(Z)
    print(f"Ward hierarchical clustering (k = {K_CHOSEN}) vs k-means: adjusted Rand index = {adjusted_rand_score(base_labels, ward.labels_):.3f}")

    win = lg["day"].between("2026-09-10", "2026-09-24") & (lg["y_v1"] == 1)
    jja = (lg["september"] == 0) & (lg["y_v1"] == 1)
    lab_s = pd.Series(base_labels, index=lg.index)
    for label, m in [("risk_safety-flagged, Sep 10-24 (window chosen after viewing the data)", win),
                     ("risk_safety-flagged, June-August", jja)]:
        shares = lab_s[m].value_counts(normalize=True)
        hhi = float((shares ** 2).sum())
        xtitle(f"F2: concentration of {label}, n = {int(m.sum())}, k = {K_CHOSEN}")
        print((shares.head(3) * 100).round(1).rename("share_%").to_frame().rename_axis("cluster").to_string())
        print(f"Herfindahl index: {hhi:.3f}")
    top_c = int(lab_s[win].value_counts().idxmax())
    xtitle(f"F2: articles per ISO week in cluster {top_c} (largest share of the Sep 10-24 window's flagged articles)")
    per_week = lg.loc[lab_s == top_c].groupby("week").size()
    per_week.index = per_week.index.strftime("%Y-%m-%d")
    print(per_week.rename("articles").rename_axis("week (Monday)").to_string())
````

### 3.3.11 เทคนิคเสริม 3: NER ชื่อบริษัทและประเทศ (เชิงสำรวจ)

````python
# ===== Part 3.3g (team) - N1: entity normalization rules (fixed BEFORE seeing any NER output) =====
import re

CORP_SUFFIXES = {"inc", "inc.", "ltd", "ltd.", "llc", "plc", "corp", "corp.", "corporation", "co", "co."}


def entity_key(s):
    """Comparison key: lower-case without dots."""
    return str(s).lower().replace(".", "")


_ALIAS_GROUPS = {
    "United States": ["us", "u.s", "usa", "united states", "united states of america", "america"],
    "United Kingdom": ["uk", "u.k", "britain", "united kingdom", "great britain"],
    "European Union": ["eu", "european union"],
    "China": ["china", "prc", "people's republic of china"],
    "Google": ["google", "alphabet"],
    "Meta": ["meta", "meta platforms", "facebook"],
    "OpenAI": ["openai", "open ai"],
    "DeepMind": ["deepmind", "google deepmind"],
    "xAI": ["xai", "x.ai"],
}
ALIASES = {entity_key(a): canon for canon, al in _ALIAS_GROUPS.items() for a in al}


def normalize_entity(text, label):
    """Clean an entity string and map it to its canonical name (rules fixed in advance)."""
    s = re.sub(r"\s+", " ", str(text)).strip()
    s = re.sub(r"^the\s+", "", s, flags=re.IGNORECASE)
    s = re.sub(r"(?:'s|’s)$", "", s).strip()
    if label in ("ORG", "GPE", "NORP"):
        parts = s.split(" ")
        if len(parts) > 1 and parts[-1].lower() in CORP_SUFFIXES:
            s = " ".join(parts[:-1]).rstrip(",").strip()
    s = re.sub(r"\s+", " ", s).strip()
    return ALIASES.get(entity_key(s), s)


def resolve_persons(names):
    """Map a one-token PERSON to the full name if it is the last token of exactly one multi-token PERSON in the same article."""
    multi = [n for n in names if len(n.split()) > 1]
    out = []
    for n in names:
        if len(n.split()) == 1:
            matches = sorted({m for m in multi if m.split()[-1] == n})
            out.append(matches[0] if len(matches) == 1 else n)
        else:
            out.append(n)
    return out


print(f"N1 ready: normalize_entity, resolve_persons and {len(ALIASES)} alias keys for {len(_ALIAS_GROUPS)} canonical names defined.")
````

````python
# ===== Part 3.3g (team) - N2b: run spaCy NER once over the 631 ledes (authors only) =====
import numpy as np
import pandas as pd
from collections import Counter

if not RERUN_NER:
    print("Skipped: set RERUN_NER = True (authors only, needs spaCy and en_core_web_sm) to recompute the entity table.")
else:
    import spacy
    nlp_ner = spacy.load("en_core_web_sm")
    keep_pipes = [p for p in ["tok2vec", "ner"] if p in nlp_ner.pipe_names]
    disabled = [p for p in nlp_ner.pipe_names if p not in keep_pipes]
    NER_LABELS = {"ORG", "PERSON", "GPE", "NORP"}
    ids_ner = pop["id"].tolist()
    ner_fresh, ner_mentions = {}, []   # mentions keep offsets and short context (memory only)
    with nlp_ner.select_pipes(disable=disabled):
        for art_id, doc in zip(ids_ner, nlp_ner.pipe(pop["lede"].tolist(), batch_size=32)):
            ents = [e for e in doc.ents if e.label_ in NER_LABELS]
            canon = [normalize_entity(e.text, e.label_) for e in ents]
            # PERSON rule: resolve one-token names inside the same article
            pers_idx = [i for i, e in enumerate(ents) if e.label_ == "PERSON"]
            resolved = resolve_persons([canon[i] for i in pers_idx])
            for i, r in zip(pers_idx, resolved):
                canon[i] = r
            for e, c in zip(ents, canon):
                side = max(0, min(5, (11 - len(e)) // 2))
                ner_mentions.append({
                    "id": art_id, "label": e.label_, "text": e.text, "canonical": c,
                    "start_char": e.start_char, "end_char": e.end_char,
                    "left": [t.text for t in doc[max(0, e.start - side):e.start]],
                    "right": [t.text for t in doc[e.end:e.end + side]],
                })
            ner_fresh[art_id] = sorted({(c, e.label_) for e, c in zip(ents, canon) if c})
    per_art = pd.Series([len(v) for v in ner_fresh.values()])
    xtitle("N2b: spaCy NER run (document-level unique entities per article)")
    print(f"spaCy {spacy.__version__}, model en_core_web_sm {nlp_ner.meta.get('version')}; components used: {keep_pipes}")
    print(f"Articles: {len(ner_fresh)} | entities per article: mean {per_art.mean():.1f}, median {per_art.median():.0f}, max {per_art.max()}")
    distinct = Counter(lbl for lbl in [x[1] for x in {e for v in ner_fresh.values() for e in v}])
    print("Distinct entities per label:", dict(sorted(distinct.items())))
````

````python
# ===== Part 3.3g (team) - N2c: document frequencies and the table to store (authors only) =====
import json

if not RERUN_NER or "ner_fresh" not in globals():
    print("Skipped: needs RERUN_NER = True and the N2b run (authors only).")
else:
    ner_dfreq = Counter(e for v in ner_fresh.values() for e in v)

    def build_store(t_org, t_person):
        keep = {e for e, n in ner_dfreq.items() if n >= (t_person if e[1] == "PERSON" else t_org)}
        return {k: [[n, l] for (n, l) in v if (n, l) in keep] for k, v in ner_fresh.items()}

    NER_T_ORG, NER_T_PERSON = 3, 5
    ner_to_store = build_store(NER_T_ORG, NER_T_PERSON)
    size = len(json.dumps(ner_to_store, ensure_ascii=True))
    print(f"Thresholds: df >= {NER_T_ORG} (ORG, GPE, NORP), df >= {NER_T_PERSON} (PERSON); literal size about {size:,} characters")
    if size > 400_000:
        NER_T_ORG, NER_T_PERSON = 5, 8
        ner_to_store = build_store(NER_T_ORG, NER_T_PERSON)
        size = len(json.dumps(ner_to_store, ensure_ascii=True))
        print(f"Size exceeded 400,000 characters: thresholds raised to {NER_T_ORG} / {NER_T_PERSON}; new size about {size:,} characters")
    kept = Counter(l for (n, l) in {tuple(x) for v in ner_to_store.values() for x in v})
    print("Distinct entities kept per label:", dict(sorted(kept.items())))
````

````python
# ===== Part 3.3g (team) - N2 export helper (authors only): prints the literal lines copied into NER_STORED =====
if not RERUN_NER or "ner_to_store" not in globals():
    print("Skipped: export helper is used only by the authors when RERUN_NER = True.")
else:
    print("NER_EXPORT_START")
    for k in sorted(ner_to_store):
        print(f"    {json.dumps(k)}: {json.dumps(ner_to_store[k], ensure_ascii=True)},")
    print("NER_EXPORT_END")
````

````python
# ===== Part 3.3g (team) - N2f: verify the stored literal table and define ner_table =====
if "ner_to_store" in globals():
    rebuilt = {k: [list(x) for x in v] for k, v in NER_STORED.items()}
    expected = {k: [list(x) for x in v] for k, v in ner_to_store.items()}
    print(f"NER_STORED equals ner_to_store: {rebuilt == expected}")
else:
    print("ner_to_store is not in memory (authors only) - verification skipped; using NER_STORED.")

# ner_table: the fresh run (with the same document-frequency thresholds as the stored table) when RERUN_NER
# is True, otherwise the stored table; spaCy is not imported when RERUN_NER is False
if RERUN_NER and "ner_to_store" in globals():
    ner_table = ner_to_store
    print("ner_table = fresh run (same thresholds as NER_STORED)")
else:
    ner_table = NER_STORED
    print("ner_table = NER_STORED")
````

````python
# ===== Part 3.3g (team) - N3: who is in the news (document frequencies; descriptive) =====
NER_NEEDED = ["pop", "ner_table", "diff_ci", "wilson", "xtitle", "FRAMES"]
NER_MISSING = [v for v in NER_NEEDED if v not in globals()]
NER_OK = not NER_MISSING
if not NER_OK:
    print("Missing variables (runtime may have been reset):", NER_MISSING, "- stopped.")
else:
    info = pop.set_index("id")
    info_day = pd.to_datetime(info["webPublicationDate"], utc=True).dt.tz_convert(None).dt.normalize()
    print(f"Articles in ner_table: {len(ner_table)}; ids equal to the population: {set(ner_table) == set(info.index)}")
    ent_long = pd.DataFrame([(k, n, l) for k, v in ner_table.items() for n, l in v], columns=["id", "name", "label"])
    ent_ids = ent_long.groupby(["name", "label"])["id"].apply(set)
    ent_df = ent_ids.map(len).sort_values(ascending=False)
    N_ART = len(info)

    for lbl in ["ORG", "PERSON", "GPE", "NORP"]:
        top = ent_df[(ent_df.index.get_level_values("label") == lbl) & (ent_df >= 5)].head(25)
        tab = pd.DataFrame({"entity": top.index.get_level_values("name"), "df": top.values,
                            "share_%": top.values / N_ART * 100})
        xtitle(f"N3: top {len(tab)} {lbl} entities by document frequency (df >= 5, n = {N_ART})")
        print(tab.to_string(index=False, float_format=lambda x: f"{x:.1f}"))

    flag_risk = info["flag_risk_safety"]
    ids_jja = set(info.index[flag_risk & (info["period"] == "Jun-Aug")])
    ids_sep = set(info.index[flag_risk & (info["period"] == "Sep")])
    ids_win = set(info.index[flag_risk & info_day.between("2026-09-10", "2026-09-24")])
    top_org = [n for n, l in ent_df.index if l == "ORG"][:15]
    top_per = [n for n, l in ent_df.index if l == "PERSON"][:15]

    def compare_entities(ids_a, ids_b, name_a, name_b):
        """Share of articles mentioning each entity in group B vs group A; diff = B - A (pp), Newcombe 95% CI."""
        rows = []
        for lbl, names in [("ORG", top_org), ("PERSON", top_per)]:
            for nm in names:
                s = ent_ids[(nm, lbl)]
                ka, kb = len(s & ids_a), len(s & ids_b)
                la, ha = wilson(ka, len(ids_a))
                lb, hb = wilson(kb, len(ids_b))
                d, lo, hi = diff_ci(kb, len(ids_b), ka, len(ids_a))
                rows.append({"label": lbl, "entity": nm,
                             f"{name_a}_%": ka / len(ids_a) * 100, f"{name_a}_CI": f"[{la:.1f}, {ha:.1f}]",
                             f"{name_b}_%": kb / len(ids_b) * 100, f"{name_b}_CI": f"[{lb:.1f}, {hb:.1f}]",
                             "diff_pp": d, "diff_CI": f"[{lo:.1f}, {hi:.1f}]"})
        return pd.DataFrame(rows)

    xtitle(f"N3: top 15 ORG / PERSON among risk_safety-flagged articles, Jun-Aug (n={len(ids_jja)}) vs Sep (n={len(ids_sep)}); "
           "diff = Sep - JunAug, Wilson / Newcombe 95% CI")
    with pd.option_context("display.width", 250):
        print(compare_entities(ids_jja, ids_sep, "JunAug", "Sep").to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    xtitle(f"N3: same table, risk_safety-flagged Sep 10-24 (window chosen after viewing the data, n={len(ids_win)}) "
           f"vs Jun-Aug (n={len(ids_jja)}); diff = window - JunAug")
    with pd.option_context("display.width", 250):
        print(compare_entities(ids_jja, ids_win, "JunAug", "Window").to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    print("Note: many entities are compared, so no p-values are reported and single differences should not be over-read.")

    def concentration(ids_g, label_g):
        sub = ent_long[ent_long["id"].isin(ids_g)]
        out = {"group": label_g, "n_articles": len(ids_g)}
        for lbl in ["ORG", "PERSON"]:
            cnt = sub[sub["label"] == lbl]["name"].value_counts().head(20)
            sh = cnt / cnt.sum() if cnt.sum() else cnt
            out[f"HHI_top20_{lbl}"] = float((sh ** 2).sum())
        top5 = set(sub[sub["label"] == "ORG"]["name"].value_counts().head(5).index)
        with_top5 = sub[(sub["label"] == "ORG") & sub["name"].isin(top5)]["id"].nunique()
        out["share_with_top5_ORG_%"] = with_top5 / len(ids_g) * 100
        out["top5_ORG"] = ", ".join(sorted(top5))
        return out

    xtitle("N3: concentration of entities, risk_safety-flagged Sep 10-24 window (chosen after viewing the data) vs Jun-Aug")
    with pd.option_context("display.width", 250, "display.max_colwidth", 80):
        print(pd.DataFrame([concentration(ids_win, "Sep 10-24 window"), concentration(ids_jja, "Jun-Aug")]).to_string(
            index=False, float_format=lambda x: f"{x:.3f}"))
````

````python
# ===== Part 3.3g (team) - N4: geopolitics cross-check with rival-state mentions (descriptive) =====
if not NER_OK:
    print("N3 prerequisites missing - N4 skipped.")
else:
    RIVALS = {"China", "Russia", "Iran", "Taiwan", "North Korea"}
    RIVAL_ADJ = {"Chinese", "Russian", "Iranian"}
    rival_ids = set(ent_long[((ent_long["label"] == "GPE") & ent_long["name"].isin(RIVALS)) |
                             ((ent_long["label"] == "NORP") & ent_long["name"].isin(RIVAL_ADJ))]["id"])
    present = sorted(set(ent_long[(ent_long["label"] == "GPE") & ent_long["name"].isin(RIVALS)]["name"]) |
                     set(ent_long[(ent_long["label"] == "NORP") & ent_long["name"].isin(RIVAL_ADJ)]["name"]))
    print(f"Rival entities present in ner_table (df thresholds apply): {present}")
    rv = pd.Series(info.index.isin(rival_ids), index=info.index, name="rival_mention")
    geo = info["flag_geopolitics"].rename("flag_geopolitics")
    for scope, m in [("all months", pd.Series(True, index=info.index)), ("September", info["period"] == "Sep"),
                     ("June-August", info["period"] == "Jun-Aug")]:
        xtitle(f"N4: rival_mention x flag_geopolitics, {scope} (n = {int(m.sum())})")
        print(pd.crosstab(rv[m], geo[m], margins=True, margins_name="total").to_string())
        r_n, g_n = int(rv[m].sum()), int(geo[m].sum())
        both = int((rv[m] & geo[m]).sum())
        print(f"Share of rival_mention articles flagged geopolitics: {both}/{r_n} = {both / r_n * 100 if r_n else float('nan'):.1f}%")
        print(f"Share of flag_geopolitics articles with rival_mention: {both}/{g_n} = {both / g_n * 100 if g_n else float('nan'):.1f}%")
    print("\nNote: the geopolitics dictionary itself contains China, Chinese and Beijing, so this check is partly circular; "
          "it shows how the threshold of two matches treats articles that mention a rival state; it is NOT a human validation")
    cand = info.index[rv & ~geo].tolist()
    pick = np.random.default_rng(42).choice(len(cand), size=min(10, len(cand)), replace=False) if cand else []
    xtitle(f"N4: 10 random headlines (seed 42) with rival_mention but NOT flagged geopolitics (of {len(cand)})")
    for i in pick:
        print(f"  - {str(info.at[cand[i], 'headline'])[:120]}")
````

````python
# ===== Part 3.3g (team) - N5: companies and frames (descriptive) =====
if not NER_OK:
    print("N3 prerequisites missing - N5 skipped.")
else:
    top10_org = [n for n, l in ent_df.index if l == "ORG"][:10]
    rows = []
    for nm in top10_org:
        has = info.index.isin(ent_ids[(nm, "ORG")])
        for fr in FRAMES:
            f = info[f"flag_{fr}"]
            ka, na = int(f[has].sum()), int(has.sum())
            kb, nb_ = int(f[~has].sum()), int((~has).sum())
            d, lo, hi = diff_ci(ka, na, kb, nb_)
            rows.append({"ORG": nm, "n_mention": na, "frame": fr, "%_flagged_mention": ka / na * 100,
                         "%_flagged_other": kb / nb_ * 100, "diff_pp": d, "CI_low": lo, "CI_high": hi})
    xtitle("N5: frame shares among articles mentioning each top-10 ORG vs not mentioning it (diff = mention - other, Newcombe 95% CI)")
    with pd.option_context("display.width", 250):
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    print("Note: 60 comparisons are shown (10 ORG x 6 frames); no p-values; expect some differences by chance.")

    months = ["2026-06", "2026-07", "2026-08", "2026-09"]
    n_month = info["month"].value_counts()
    rows = []
    for nm in top10_org:
        has = pd.Series(info.index.isin(ent_ids[(nm, "ORG")]), index=info.index)
        row = {"ORG": nm}
        for mo in months:
            m = info["month"] == mo
            row[mo] = has[m].mean() * 100
        rows.append(row)
    xtitle("N5: monthly share (%) of articles mentioning each top-10 ORG; n per month: "
           + ", ".join(f"{mo} n={int(n_month.get(mo, 0))}" for mo in months))
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
````

````python
# ===== Part 3.3g (team) - N6: human check of the NER labels (print only) =====
if "ner_mentions" not in globals():
    print("This step needs the raw mention list of N2b: set RERUN_NER = True (authors only, needs spaCy) - skipped.")
else:
    mentions = pd.DataFrame(ner_mentions)
    kept_person = {n for v in ner_table.values() for n, l in v if l == "PERSON"}
    rng_n6 = np.random.default_rng(42)
    for lbl in ["ORG", "PERSON", "GPE"]:
        pool = mentions[mentions["label"] == lbl]
        if lbl == "PERSON":
            pool = pool[pool["canonical"].isin(kept_person)]
        pick = rng_n6.choice(len(pool), size=min(20, len(pool)), replace=False)
        xtitle(f"N6: 20 random {lbl} mentions (seed 42; classroom analysis only, do not redistribute)")
        for i in pick:
            r = pool.iloc[i]
            window = " ".join(r["left"] + [f"[{r['text']}]"] + r["right"])
            print(f"  {lbl} | {r['text']} | {r['id']} | {window}")
    for lbl in ["PERSON", "ORG"]:
        xtitle(f"N6: 30 most frequent strings tagged {lbl} (mention counts)")
        print(mentions[mentions["label"] == lbl]["text"].value_counts().head(30).rename("count").rename_axis("string").to_string())
    print("\nto be read by humans: mark each sample line as correct/incorrect label; compute precision per label afterwards")
````

#### 3.3.12 แก้ป้าย NER ที่ผิด (เพิ่มหลังเห็นผล)
* โมเดล spaCy ขนาดเล็กติดป้ายผิดหลายคำ เช่น "AI" ถูกติดเป็นสถานที่ (GPE) และองค์กร (ORG), บริษัท AI และแชตบอตถูกติดเป็นบุคคล (PERSON), และมีข้อความ boilerplate ของ Guardian ปนอยู่
* ตาราง NER ชุดแรก (cells MR7VU5OeW06t, BqZOWr2PW1FA, NgOujh6eW1Pu) เก็บไว้เพื่อความโปร่งใส แต่ **ถูกแทนที่ (SUPERSEDED)** ด้วยตารางที่แก้แล้วด้านล่าง สำหรับองค์กรและบุคคล
* กฎด้านล่าง **เขียนขึ้นหลังจากเห็นผลแล้ว** และใช้แบบเดียวกันกับทุกบทความ
* รายการแก้ไขฉบับสุดท้ายจะทำให้ครบหลังจากคนอ่านตัวอย่างที่พิมพ์ไว้ (N6)
* ผลเรื่องประเทศคู่แข่ง (N4 ใช้เฉพาะสถานที่ GPE/NORP) ไม่ได้รับผลจากกฎเหล่านี้

````python
# ===== Part 3.3g-2 (team) - Q2: NER correction rules v2-preliminary (written after seeing the output) =====
import numpy as np
import pandas as pd
from collections import Counter, defaultdict

QX = "EXPLORATORY (added after seeing the NER output; v1 dictionary not human-validated)"


def qtitle(name):
    """Print a table title that always carries the prefix of this step."""
    print(f"\n=== {QX} | {name} ===")


Q_NEEDED = ["pop", "ner_table", "NER_STORED", "diff_ci", "wilson", "FRAMES"]
Q_MISSING = [v for v in Q_NEEDED if v not in globals()]
Q_OK = not Q_MISSING
if not Q_OK:
    print("Missing variables (runtime may have been reset):", Q_MISSING, "- stopped.")
else:
    print(f"ner_table is NER_STORED: {ner_table is NER_STORED}")
    DROP = {"ai", "a.i.", "guardian", "the guardian", "breaking news australia", "pfas"}   # comparison key: lower-case name
    ORG_MERGE = {"openai", "anthropic", "palantir", "spacex", "meta", "google", "deepmind", "xai",
                 "microsoft", "nvidia", "amazon", "apple", "tesla"}   # forced to label ORG whatever the model said
    PRODUCT_MERGE = {"claude", "grok", "chatgpt", "gemini", "copilot"}   # forced to the new label PRODUCT

    def new_label(key, label):
        if key in ORG_MERGE:
            return "ORG"
        if key in PRODUCT_MERGE:
            return "PRODUCT"
        return label

    # First spelling seen (articles in sorted id order) becomes the canonical name of each (key, new label)
    canon_name, dropped = {}, Counter()
    ner_clean = {}
    for art_id in sorted(ner_table):
        seen, out = set(), []
        for name, label in ner_table[art_id]:
            key = name.lower()
            if key in DROP:
                dropped[key] += 1
                continue
            lbl = new_label(key, label)
            canon_name.setdefault((key, lbl), name)
            if (key, lbl) in seen:
                continue   # duplicate after relabelling: one entry per article and entity
            seen.add((key, lbl))
            out.append([canon_name[(key, lbl)], lbl])
        ner_clean[art_id] = sorted(out)

    n_before = sum(len(v) for v in ner_table.values())
    n_after = sum(len(v) for v in ner_clean.values())
    qtitle("Q2: effect of the correction rules")
    print(f"Pairs (article, entity): before {n_before}, after {n_after}")
    print("Pairs dropped per DROP string:", {k: dropped.get(k, 0) for k in sorted(DROP)})
    rows = []
    for key in sorted(ORG_MERGE | PRODUCT_MERGE):
        before = sum(1 for v in ner_table.values() for n, l in v if n.lower() == key)
        target = "ORG" if key in ORG_MERGE else "PRODUCT"
        after = sum(1 for v in ner_clean.values() if any(n.lower() == key and l == target for n, l in v))
        labels_before = sorted({l for v in ner_table.values() for n, l in v if n.lower() == key})
        rows.append({"name": key, "forced_label": target, "articles_before_summed_over_labels": before,
                     "labels_before": "/".join(labels_before) or "-", "articles_after_unique": after})
    print(pd.DataFrame(rows).to_string(index=False))
    print("Note: the 'before' count sums over labels and can count the same article more than once.")
    print(f"Articles with at least one entity: before {sum(1 for v in ner_table.values() if v)}, "
          f"after {sum(1 for v in ner_clean.values() if v)} (of {len(ner_table)})")
````

````python
# ===== Part 3.3g-2 (team) - Q3: gazetteer (regex) cross-check of NER for companies =====
import re

if not Q_OK:
    print("Q2 prerequisites missing - Q3 skipped.")
else:
    GAZ = {   # case-sensitive, whole-word regular expressions applied to the lede
        "OpenAI": r"\bOpenAI\b", "Anthropic": r"\bAnthropic\b", "Google": r"\bGoogle\b", "DeepMind": r"\bDeepMind\b",
        "Meta": r"\bMeta\b", "Microsoft": r"\bMicrosoft\b", "Nvidia": r"\bNvidia\b|\bNVIDIA\b", "Amazon": r"\bAmazon\b",
        "Apple": r"\bApple\b", "xAI": r"\bxAI\b", "Tesla": r"\bTesla\b", "SpaceX": r"\bSpaceX\b", "Palantir": r"\bPalantir\b",
    }
    lede_by_id = pop.set_index("id")["lede"]
    rows = []
    for comp, rx in GAZ.items():
        pat = re.compile(rx)
        by_rx = set(lede_by_id.index[lede_by_id.map(lambda t: bool(pat.search(t)))])
        by_ner = {k for k, v in ner_clean.items() if any(n.lower() == comp.lower() and l == "ORG" for n, l in v)}
        rows.append({"company": comp, "regex": len(by_rx), "ner_clean_ORG": len(by_ner), "both": len(by_rx & by_ner),
                     "regex_only": len(by_rx - by_ner), "ner_only": len(by_ner - by_rx)})
    qtitle("Q3: articles per company - case-sensitive regex vs ner_clean (ORG)")
    print(pd.DataFrame(rows).to_string(index=False))
    print("Note: Amazon, Apple and Meta are common words (Amazon rainforest, apple fruit, meta-analysis); the regex is "
          "case-sensitive but not disambiguated; this table compares two imperfect methods and is not a human validation")
    print("Note: NER counts come from the stored table, which keeps only entities with document frequency >= 3 (ORG).")
````

````python
# ===== Part 3.3g-2 (team) - Q4a: N3-clean (same logic as N3, using ner_clean) =====
if not Q_OK:
    print("Q2 prerequisites missing - Q4a skipped.")
else:
    info_q = pop.set_index("id")
    day_q = pd.to_datetime(info_q["webPublicationDate"], utc=True).dt.tz_convert(None).dt.normalize()
    N_Q = len(info_q)

    def long_table(table):
        return pd.DataFrame([(k, n, l) for k, v in table.items() for n, l in v], columns=["id", "name", "label"])

    cl_long = long_table(ner_clean)
    cl_ids = cl_long.groupby(["name", "label"])["id"].apply(set)
    cl_df = cl_ids.map(len).sort_values(ascending=False)

    for lbl in ["ORG", "PERSON", "GPE", "NORP", "PRODUCT"]:
        top = cl_df[(cl_df.index.get_level_values("label") == lbl) & (cl_df >= 5)].head(25)
        tab = pd.DataFrame({"entity": top.index.get_level_values("name"), "df": top.values, "share_%": top.values / N_Q * 100})
        qtitle(f"Q4a: top {len(tab)} {lbl} entities by document frequency in ner_clean (df >= 5, n = {N_Q})")
        print(tab.to_string(index=False, float_format=lambda x: f"{x:.1f}") if len(tab) else "none with df >= 5")

    flag_q = info_q["flag_risk_safety"]
    q_jja = set(info_q.index[flag_q & (info_q["period"] == "Jun-Aug")])
    q_sep = set(info_q.index[flag_q & (info_q["period"] == "Sep")])
    q_win = set(info_q.index[flag_q & day_q.between("2026-09-10", "2026-09-24")])
    q_top_org = [n for n, l in cl_df.index if l == "ORG"][:15]
    q_top_prod = [n for n, l in cl_df.index if l == "PRODUCT"][:10]

    def compare_clean(ids_a, ids_b, name_a, name_b):
        """Share of articles mentioning each entity in B vs A; diff = B - A (pp), Wilson / Newcombe 95% CI."""
        rows = []
        for lbl, names in [("ORG", q_top_org), ("PRODUCT", q_top_prod)]:
            for nm in names:
                s = cl_ids[(nm, lbl)]
                ka, kb = len(s & ids_a), len(s & ids_b)
                la, ha = wilson(ka, len(ids_a))
                lb, hb = wilson(kb, len(ids_b))
                d, lo, hi = diff_ci(kb, len(ids_b), ka, len(ids_a))
                rows.append({"label": lbl, "entity": nm,
                             f"{name_a}_%": ka / len(ids_a) * 100, f"{name_a}_CI": f"[{la:.1f}, {ha:.1f}]",
                             f"{name_b}_%": kb / len(ids_b) * 100, f"{name_b}_CI": f"[{lb:.1f}, {hb:.1f}]",
                             "diff_pp": d, "diff_CI": f"[{lo:.1f}, {hi:.1f}]"})
        return pd.DataFrame(rows)

    qtitle(f"Q4a: top 15 ORG / top PRODUCT among risk_safety-flagged articles, Jun-Aug (n={len(q_jja)}) vs Sep (n={len(q_sep)}); "
           "diff = Sep - JunAug")
    with pd.option_context("display.width", 250):
        print(compare_clean(q_jja, q_sep, "JunAug", "Sep").to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    qtitle(f"Q4a: same table, risk_safety-flagged Sep 10-24 (window chosen after viewing the data, n={len(q_win)}) "
           f"vs Jun-Aug (n={len(q_jja)}); diff = window - JunAug")
    with pd.option_context("display.width", 250):
        print(compare_clean(q_jja, q_win, "JunAug", "Window").to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    print("Note: many entities are compared, so no p-values are reported and single differences should not be over-read.")

    def concentration_q(long_df, ids_g):
        sub = long_df[long_df["id"].isin(ids_g) & (long_df["label"] == "ORG")]
        cnt = sub["name"].value_counts().head(20)
        sh = cnt / cnt.sum() if cnt.sum() else cnt
        top5 = set(sub["name"].value_counts().head(5).index)
        share5 = sub[sub["name"].isin(top5)]["id"].nunique() / len(ids_g) * 100
        return float((sh ** 2).sum()), share5, ", ".join(sorted(top5))

    old_long = long_table(ner_table)
    rows = []
    for gname, ids_g in [("Sep 10-24 window (chosen after viewing the data)", q_win), ("Jun-Aug", q_jja)]:
        h_old, s_old, t_old = concentration_q(old_long, ids_g)
        h_new, s_new, t_new = concentration_q(cl_long, ids_g)
        rows.append({"group": gname, "n": len(ids_g), "HHI_top20_ORG_old": h_old, "HHI_top20_ORG_new": h_new,
                     "share_top5_ORG_%_old": s_old, "share_top5_ORG_%_new": s_new,
                     "changed": "yes" if (round(h_old, 3) != round(h_new, 3) or round(s_old, 1) != round(s_new, 1)) else "no",
                     "top5_ORG_new": t_new})
    qtitle("Q4a: concentration of ORG mentions among risk_safety-flagged articles, old (ner_table) vs new (ner_clean)")
    with pd.option_context("display.width", 300, "display.max_colwidth", 90):
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.3f}"))
````

````python
# ===== Part 3.3g-2 (team) - Q4b: N5-clean (companies and frames) and PERSON orientation list =====
if not Q_OK or "cl_df" not in globals():
    print("Q4a did not complete - Q4b skipped.")
else:
    q_top10 = [n for n, l in cl_df.index if l == "ORG"][:10]
    rows = []
    for nm in q_top10:
        has = info_q.index.isin(cl_ids[(nm, "ORG")])
        for fr in FRAMES:
            f = info_q[f"flag_{fr}"]
            ka, na = int(f[has].sum()), int(has.sum())
            kb, nb_ = int(f[~has].sum()), int((~has).sum())
            d, lo, hi = diff_ci(ka, na, kb, nb_)
            rows.append({"ORG": nm, "n_mention": na, "frame": fr, "%_flagged_mention": ka / na * 100,
                         "%_flagged_other": kb / nb_ * 100, "diff_pp": d, "CI_low": lo, "CI_high": hi})
    qtitle("Q4b: frame shares among articles mentioning each top-10 ORG (ner_clean) vs not (diff = mention - other, Newcombe 95% CI)")
    with pd.option_context("display.width", 250):
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    print("Note: 60 comparisons are shown (10 ORG x 6 frames); no p-values; expect some differences by chance.")

    months_q = ["2026-06", "2026-07", "2026-08", "2026-09"]
    n_month_q = info_q["month"].value_counts()
    rows = []
    for nm in q_top10:
        has = pd.Series(info_q.index.isin(cl_ids[(nm, "ORG")]), index=info_q.index)
        rows.append({"ORG": nm, **{mo: has[info_q["month"] == mo].mean() * 100 for mo in months_q}})
    qtitle("Q4b: monthly share (%) of articles mentioning each top-10 ORG (ner_clean); n per month: "
           + ", ".join(f"{mo} n={int(n_month_q.get(mo, 0))}" for mo in months_q))
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))

    top_p = cl_df[cl_df.index.get_level_values("label") == "PERSON"].head(15)
    qtitle(f"Q4b: top 15 PERSON names in ner_clean (whole population, n = {N_Q})")
    print(pd.DataFrame({"person": top_p.index.get_level_values("name"), "df": top_p.values,
                        "share_%": top_p.values / N_Q * 100}).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    print("individual names are reported only for orientation; slides should use organisation level")
````

#### 3.3.13 นับบริษัทด้วยรายการชื่อแทน NER
* ผลตรวจเทียบใน Q3 พบว่าโมเดล spaCy ขนาดเล็กพลาดการพบชื่อบริษัทในรายการปิดจำนวนมาก (เช่น Anthropic: NER พบ 52 บทความ เทียบกับ 121 บทความเมื่อจับชื่อแบบตรงตัวพิมพ์)
* สำหรับบริษัท ตารางด้านล่าง **แทนที่ (SUPERSEDE)** ตารางบริษัทใน cells YhONZKI-ZWCc และ kA_iCX-wZWNZ
* รายการชื่อกำหนด **หลังจากเห็นผล Q3** และใช้แบบเดียวกันกับทุกบทความ
* ชื่อที่เป็นคำทั่วไปด้วย (Meta, Amazon, Apple) ไม่นำมาใช้ในตารางกรอบข่าว
* ตัวเลขทั้งหมดเป็นการบรรยาย ยังไม่ผ่านการตรวจโดยคน
* ไม่มีการบรรยายลักษณะของบริษัทใดในข้อความ รายงานเฉพาะจำนวนครั้งที่ถูกกล่าวถึงเท่านั้น

````python
# ===== Part 3.3g-3 (team) - G2: company name list (gazetteer) and mention counts =====
import re
import numpy as np
import pandas as pd

GX = "EXPLORATORY (added after seeing the NER comparison; v1 dictionary not human-validated)"


def gtitle(name):
    """Print a table title that always carries the prefix of this step."""
    print(f"\n=== {GX} | {name} ===")


GZ_NEEDED = ["pop", "clean_html", "diff_ci", "wilson", "FRAMES"]
GZ_MISSING = [v for v in GZ_NEEDED if v not in globals()]
GZ_OK = not GZ_MISSING
if not GZ_OK:
    print("Missing variables (runtime may have been reset):", GZ_MISSING, "- stopped.")
else:
    # Case-sensitive, whole-word patterns; document-level presence in the lede
    GAZ_SAFE = {
        "OpenAI": r"\bOpenAI\b", "Anthropic": r"\bAnthropic\b", "Google": r"\bGoogle\b", "DeepMind": r"\bDeepMind\b",
        "Microsoft": r"\bMicrosoft\b", "Nvidia": r"\bNvidia\b|\bNVIDIA\b", "xAI": r"\bxAI\b", "SpaceX": r"\bSpaceX\b",
        "Palantir": r"\bPalantir\b", "Tesla": r"\bTesla\b",
    }
    GAZ_AMBIGUOUS = {"Meta": r"\bMeta\b", "Amazon": r"\bAmazon\b", "Apple": r"\bApple\b"}   # counts only, never in frame tables
    gz = pop.copy()   # work on a copy; pop is not modified
    gz_body = gz["bodyText"].map(clean_html)
    gz_day = pd.to_datetime(gz["webPublicationDate"], utc=True).dt.tz_convert(None).dt.normalize()
    rows = []
    for name, rx in {**GAZ_SAFE, **GAZ_AMBIGUOUS}.items():
        pat = re.compile(rx)
        gz[f"co_{name}"] = gz["lede"].map(lambda t, p=pat: bool(p.search(t)))
        in_body = gz_body.map(lambda t, p=pat: bool(p.search(t)))
        n_lede, n_body = int(gz[f"co_{name}"].sum()), int(in_body.sum())
        rows.append({"company": name, "list": "safe" if name in GAZ_SAFE else "ambiguous",
                     "lede_all": n_lede, "lede_JunAug": int((gz[f"co_{name}"] & (gz["period"] == "Jun-Aug")).sum()),
                     "lede_Sep": int((gz[f"co_{name}"] & (gz["period"] == "Sep")).sum()),
                     "body_all": n_body, "ratio_lede_body": n_lede / n_body if n_body else np.nan})
    gtitle(f"G2: articles mentioning each name (lede; full bodyText as sensitivity), n = {len(gz)}")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.2f}"))
    any_safe = gz[[f"co_{n}" for n in GAZ_SAFE]].any(axis=1)
    print(f"\nArticles mentioning at least one GAZ_SAFE company in the lede: {int(any_safe.sum())} of {len(gz)}")
````

````python
# ===== Part 3.3g-3 (team) - G3: companies in the risk_safety-flagged articles =====
if not GZ_OK:
    print("G2 prerequisites missing - G3 skipped.")
else:
    flag_g = gz["flag_risk_safety"]
    m_jja = flag_g & (gz["period"] == "Jun-Aug")
    m_sep = flag_g & (gz["period"] == "Sep")
    m_win = flag_g & gz_day.between("2026-09-10", "2026-09-24")
    g3_names = [n for n in GAZ_SAFE if gz[f"co_{n}"].sum() >= 8]

    def comp_table(ma, mb, name_a, name_b):
        rows = []
        for n in g3_names:
            col = gz[f"co_{n}"]
            ka, na = int((col & ma).sum()), int(ma.sum())
            kb, nb_ = int((col & mb).sum()), int(mb.sum())
            la, ha = wilson(ka, na)
            lb, hb = wilson(kb, nb_)
            d, lo, hi = diff_ci(kb, nb_, ka, na)
            rows.append({"company": n, f"{name_a}_%": ka / na * 100, f"{name_a}_CI": f"[{la:.1f}, {ha:.1f}]",
                         f"{name_b}_%": kb / nb_ * 100, f"{name_b}_CI": f"[{lb:.1f}, {hb:.1f}]",
                         "diff_pp": d, "diff_CI": f"[{lo:.1f}, {hi:.1f}]"})
        return pd.DataFrame(rows)

    gtitle(f"G3: share of risk_safety-flagged articles mentioning each company, Jun-Aug (n={int(m_jja.sum())}) vs Sep "
           f"(n={int(m_sep.sum())}); diff = Sep - JunAug, Wilson / Newcombe 95% CI")
    t_sep = comp_table(m_jja, m_sep, "JunAug", "Sep")
    with pd.option_context("display.width", 250):
        print(t_sep.to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    gtitle(f"G3: same, risk_safety-flagged Sep 10-24 (window chosen after viewing the data, n={int(m_win.sum())}) vs Jun-Aug; "
           "diff = window - JunAug")
    t_win = comp_table(m_jja, m_win, "JunAug", "Window")
    with pd.option_context("display.width", 250):
        print(t_win.to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    print(f"Note: {2 * len(g3_names)} comparisons ({len(g3_names)} companies x 2 contrasts); no p-values; "
          "single differences should not be over-read.")

    # Old NER-based rows (ner_clean, cell YhONZKI-ZWCc) for OpenAI and Anthropic, recomputed for comparison
    if "cl_ids" in globals():
        ids_g = gz.set_index("id")
        sets = {"JunAug": set(ids_g.index[m_jja.values]), "Sep": set(ids_g.index[m_sep.values]),
                "Window": set(ids_g.index[m_win.values])}
        rows = []
        for n in ["OpenAI", "Anthropic"]:
            s_ner = cl_ids.get((n, "ORG"), set())
            new_sep = t_sep.set_index("company").loc[n]
            new_win = t_win.set_index("company").loc[n]
            old = {k: len(s_ner & v) / len(v) * 100 for k, v in sets.items()}
            rows.append({"company": n, "NER_JunAug_%": old["JunAug"], "GAZ_JunAug_%": new_sep["JunAug_%"],
                         "NER_Sep_%": old["Sep"], "GAZ_Sep_%": new_sep["Sep_%"],
                         "NER_Window_%": old["Window"], "GAZ_Window_%": new_win["Window_%"]})
        gtitle("G3: old NER-based shares (ner_clean) vs gazetteer shares, OpenAI and Anthropic")
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    else:
        print("ner_clean results (cl_ids) not in memory - old NER rows not shown.")
````

````python
# ===== Part 3.3g-3 (team) - G4: companies and frames, monthly shares =====
if not GZ_OK:
    print("G2 prerequisites missing - G4 skipped.")
else:
    top6 = sorted(GAZ_SAFE, key=lambda n: -int(gz[f"co_{n}"].sum()))[:6]
    rows = []
    for n in top6:
        has = gz[f"co_{n}"]
        for fr in FRAMES:
            f = gz[f"flag_{fr}"]
            ka, na = int(f[has].sum()), int(has.sum())
            kb, nb_ = int(f[~has].sum()), int((~has).sum())
            d, lo, hi = diff_ci(ka, na, kb, nb_)
            rows.append({"company": n, "n_mention": na, "frame": fr, "%_flagged_mention": ka / na * 100,
                         "%_flagged_other": kb / nb_ * 100, "diff_pp": d, "CI_low": lo, "CI_high": hi})
    gtitle("G4: frame shares among articles mentioning each company vs not (diff = mention - other, Newcombe 95% CI)")
    with pd.option_context("display.width", 250):
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    print("Note: 36 comparisons (6 companies x 6 frames); no p-values; expect some differences by chance.")

    months_g = ["2026-06", "2026-07", "2026-08", "2026-09"]
    rows = []
    for n in top6 + ["OpenAI and Anthropic (both)", "OpenAI or Anthropic (either)"]:
        if n.startswith("OpenAI and"):
            col = gz["co_OpenAI"] & gz["co_Anthropic"]
        elif n.startswith("OpenAI or"):
            col = gz["co_OpenAI"] | gz["co_Anthropic"]
        else:
            col = gz[f"co_{n}"]
        row = {"company": n}
        for mo in months_g:
            m = gz["month"] == mo
            k, nn = int(col[m].sum()), int(m.sum())
            lo, hi = wilson(k, nn)
            row[mo] = f"{k / nn * 100:.1f} [{lo:.1f}, {hi:.1f}]"
        rows.append(row)
    n_per = gz["month"].value_counts()
    gtitle("G4: monthly share (%) of articles mentioning each company [Wilson 95% CI]; n per month: "
           + ", ".join(f"{mo} n={int(n_per.get(mo, 0))}" for mo in months_g))
    with pd.option_context("display.width", 250):
        print(pd.DataFrame(rows).to_string(index=False))
````

````python
# ===== Part 3.3g-3 (team) - G5: reading list for humans (headlines only) =====
if not GZ_OK:
    print("G2 prerequisites missing - G5 skipped.")
else:
    oa = gz["co_OpenAI"] | gz["co_Anthropic"]
    rng_g5 = np.random.default_rng(42)
    for label, m in [("risk_safety-flagged, Sep 10-24 (window chosen after viewing the data)", m_win & oa),
                     ("risk_safety-flagged, June-August", m_jja & oa)]:
        sub = gz[m]
        pick = rng_g5.choice(len(sub), size=min(20, len(sub)), replace=False) if len(sub) else []
        gtitle(f"G5: {label}, mentioning OpenAI or Anthropic: {len(sub)} articles; {len(pick)} shown at random (seed 42)")
        for i in pick:
            r = sub.iloc[i]
            print(f"  {gz_day.loc[sub.index[i]]:%Y-%m-%d} | {r['content_group']:<8} | {str(r['headline'])[:120]}")
    print("\nto be read by humans before any interpretation")
````

### 3.3.14 ส่วนเพิ่มของกรอบความเสี่ยงเกี่ยวกับสองบริษัทแค่ไหน (เชิงสำรวจ)
* จากการนับชื่อบริษัท พบว่าการพูดถึง OpenAI และ Anthropic เพิ่มขึ้นใน ก.ย. ใน **บทความทั้งหมด** ไม่ใช่เฉพาะบทความที่ถูก flag risk
* การตรวจด้านล่างถามว่า ส่วนเพิ่มของ flag risk_safety ใน ก.ย. มาจากบทความที่พูดถึงสองบริษัทนี้มากน้อยเพียงใด
* การแบ่งกลุ่มนี้ **กำหนดหลังจากเห็นข้อมูลแล้ว** และผลเป็นการบรรยาย (descriptive)
* การพูดถึงนับจาก lede และการที่ชื่อปรากฏ **ไม่ได้แปลว่าบทความนั้นเป็นเรื่องของบริษัทนั้น**
* ปฏิบัติต่อสองบริษัทแบบเดียวกัน และไม่บรรยายลักษณะของบริษัทใด

````python
# ===== Part 3.3g-4 (team) - H2: grouping variables (defined after seeing the data) =====
import re
import numpy as np
import pandas as pd
from statsmodels.stats.contingency_tables import StratifiedTable

HX = "EXPLORATORY (added after seeing the company counts; v1 dictionary not human-validated)"


def htitle(name):
    """Print a table title that always carries the prefix of this step."""
    print(f"\n=== {HX} | {name} ===")


H_NEEDED = ["gz", "gz_body", "gz_day", "GAZ_SAFE", "diff_ci", "wilson", "lg", "design", "fit_glm", "or_ci", "ame_binary"]
H_MISSING = [v for v in H_NEEDED if v not in globals()]
H_OK = not H_MISSING
if not H_OK:
    print("Missing variables (runtime may have been reset):", H_MISSING, "- stopped.")
else:
    hz = gz.copy()   # copy of the G2 dataframe (same index as pop and lg)
    hz["lab2"] = hz["co_OpenAI"] | hz["co_Anthropic"]
    hz["any_company"] = hz[[f"co_{n}" for n in GAZ_SAFE]].any(axis=1)
    pat_oa = re.compile(GAZ_SAFE["OpenAI"])
    pat_an = re.compile(GAZ_SAFE["Anthropic"])
    hz["lab2_body"] = gz_body.map(lambda t: bool(pat_oa.search(t) or pat_an.search(t)))
    hz["is_sep"] = hz["period"] == "Sep"
    hz["y"] = hz["flag_risk_safety"].astype(int)
    rows = []
    for v in ["lab2", "any_company", "lab2_body"]:
        row = {"variable": v, "n_all": int(hz[v].sum())}
        for per in ["Jun-Aug", "Sep"]:
            m = hz["period"] == per
            row[f"{per}_n"] = int(hz.loc[m, v].sum())
            row[f"{per}_%"] = hz.loc[m, v].mean() * 100
        rows.append(row)
    htitle(f"H2: articles per grouping variable (Jun-Aug n={int((~hz['is_sep']).sum())}, Sep n={int(hz['is_sep'].sum())})")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
````

````python
# ===== Part 3.3g-4 (team) - H3: risk_safety by period within groups, Mantel-Haenszel =====
def stratified_by(var, mask=None, show=True):
    """Period x group table for the risk flag; within-group differences and MH pooled OR (Sep vs Jun-Aug)."""
    d = hz if mask is None else hz[mask]
    rows, tables = [], []
    for g_val, g_name in [(True, "mentions"), (False, "does not mention")]:
        gm = d[var] == g_val
        stats = {}
        for per in ["Jun-Aug", "Sep"]:
            s = d[gm & (d["period"] == per)]
            k, n = int(s["y"].sum()), len(s)
            lo, hi = wilson(k, n) if n else (np.nan, np.nan)
            stats[per] = (k, n)
            rows.append({"group": g_name, "period": per, "n": n, "flagged": k,
                         "share_%": k / n * 100 if n else np.nan, "CI": f"[{lo:.1f}, {hi:.1f}]"})
        (kj, nj), (ks, ns) = stats["Jun-Aug"], stats["Sep"]
        if nj >= 15 and ns >= 15:
            dd, lo, hi = diff_ci(ks, ns, kj, nj)
            rows.append({"group": g_name, "period": "diff Sep - JunAug (pp)", "n": np.nan, "flagged": np.nan,
                         "share_%": dd, "CI": f"[{lo:.1f}, {hi:.1f}]"})
        else:
            rows.append({"group": g_name, "period": "diff Sep - JunAug (pp)", "n": np.nan, "flagged": np.nan,
                         "share_%": np.nan, "CI": "n too small"})
        tables.append(np.array([[ks, ns - ks], [kj, nj - kj]]))
    st = StratifiedTable(tables)
    lo, hi = st.oddsratio_pooled_confint()
    mh = {"MH_OR": st.oddsratio_pooled, "CI_low": lo, "CI_high": hi, "Breslow-Day_p": st.test_equal_odds().pvalue}
    if show:
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}", na_rep="-"))
        print("Mantel-Haenszel pooled OR (Sep vs Jun-Aug) across the two groups: "
              f"{mh['MH_OR']:.2f} [{mh['CI_low']:.2f}, {mh['CI_high']:.2f}], Breslow-Day p = {mh['Breslow-Day_p']:.3f}")
    return mh


if not H_OK:
    print("H2 prerequisites missing - H3 skipped.")
else:
    h3_mh = {}
    for v in ["lab2", "any_company", "lab2_body"]:
        htitle(f"H3: risk_safety share by period within groups of '{v}' (Wilson / Newcombe 95% CI)")
        h3_mh[v] = stratified_by(v)
````

````python
# ===== Part 3.3g-4 (team) - H4: Kitagawa decomposition of the September rise (lab2) =====
if not H_OK:
    print("H2 prerequisites missing - H4 skipped.")
else:
    def kitagawa(g_s, y_s, g_j, y_j):
        """Composition and rate effects (proportions) for a binary grouping g and binary outcome y."""
        out = {"comp": 0.0, "rate": 0.0, "rate_g": {}}
        for gv in (True, False):
            ws, wj = np.mean(g_s == gv), np.mean(g_j == gv)
            rs = y_s[g_s == gv].mean() if (g_s == gv).any() else 0.0
            rj = y_j[g_j == gv].mean() if (g_j == gv).any() else 0.0
            out["comp"] += (ws - wj) * (rs + rj) / 2
            rg = (ws + wj) / 2 * (rs - rj)
            out["rate"] += rg
            out["rate_g"][gv] = rg
        out["total"] = y_s.mean() - y_j.mean()
        return out

    g_s = hz.loc[hz["is_sep"], "lab2"].to_numpy()
    y_s = hz.loc[hz["is_sep"], "y"].to_numpy()
    g_j = hz.loc[~hz["is_sep"], "lab2"].to_numpy()
    y_j = hz.loc[~hz["is_sep"], "y"].to_numpy()
    est = kitagawa(g_s, y_s, g_j, y_j)

    rng_k = np.random.default_rng(42)
    boot = {"total": [], "comp": [], "rate": [], "rate_mention": [], "rate_not": [], "comp_share": [], "rate_share": []}
    for _ in range(5000):
        i_s = rng_k.integers(0, len(y_s), len(y_s))
        i_j = rng_k.integers(0, len(y_j), len(y_j))
        b = kitagawa(g_s[i_s], y_s[i_s], g_j[i_j], y_j[i_j])
        boot["total"].append(b["total"]); boot["comp"].append(b["comp"]); boot["rate"].append(b["rate"])
        boot["rate_mention"].append(b["rate_g"][True]); boot["rate_not"].append(b["rate_g"][False])
        boot["comp_share"].append(b["comp"] / b["total"] if b["total"] != 0 else np.nan)
        boot["rate_share"].append(b["rate"] / b["total"] if b["total"] != 0 else np.nan)

    def ci(key, scale=100):
        a = np.array(boot[key], dtype=float)
        a = a[~np.isnan(a)]
        return np.percentile(a, 2.5) * scale, np.percentile(a, 97.5) * scale

    point = {"total": est["total"], "comp": est["comp"], "rate": est["rate"],
             "rate_mention": est["rate_g"][True], "rate_not": est["rate_g"][False]}
    rows = [{"component": k, "pp": point[k] * 100, "CI_low": ci(k)[0], "CI_high": ci(k)[1]}
            for k in ["total", "comp", "rate", "rate_mention", "rate_not"]]
    rows += [{"component": "composition share of total", "pp": est["comp"] / est["total"] * 100,
              "CI_low": ci("comp_share")[0], "CI_high": ci("comp_share")[1]},
             {"component": "rate share of total", "pp": est["rate"] / est["total"] * 100,
              "CI_low": ci("rate_share")[0], "CI_high": ci("rate_share")[1]}]
    htitle("H4: Kitagawa decomposition of Sep - JunAug in the risk_safety rate, groups = lab2 "
           "(percentage points; shares in %; percentile bootstrap 95% CI, seed 42, 5000 resamples within period)")
    print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.1f}"))
    print("rate_mention / rate_not = rate effect within articles that mention / do not mention OpenAI or Anthropic.")
````

````python
# ===== Part 3.3g-4 (team) - H5: logistic regression M3 and M4 = M3 + lab2 =====
if not H_OK:
    print("H2 prerequisites missing - H5 skipped.")
else:
    def h_models(mask=None, lab_col="lab2"):
        """Fit M3 and M4 (M3 + lab) on the rows in mask; return ordinary and cluster-robust results."""
        d = lg if mask is None else lg[mask]
        X3 = design(d, "M3")
        X4 = X3.copy()
        X4["lab"] = hz.loc[d.index, lab_col].astype(float)
        y = hz.loc[d.index, "y"]
        groups = d["week"].factorize()[0]
        out = {}
        for name, X in [("M3", X3), ("M4", X4)]:
            r_ord = fit_glm(y, X)
            r_cl = fit_glm(y, X, groups=groups)
            out[name] = (r_ord, r_cl, X)
        return out, len(np.unique(groups))

    hm, n_cl = h_models()
    rows = []
    for name in ["M3", "M4"]:
        r_ord, r_cl, X = hm[name]
        o, lo, hi = or_ci(r_ord)
        _, loc_, hic = or_ci(r_cl)
        a, alo, ahi = ame_binary(r_ord, X)
        row = {"model": name, "OR_sep": o, "CI_low": lo, "CI_high": hi, "CI_low_cluster": loc_, "CI_high_cluster": hic,
               "AME_sep_pp": a, "AME_CI_low": alo, "AME_CI_high": ahi}
        if name == "M4":
            ol, oll, olh = or_ci(r_ord, "lab")
            _, cll, clh = or_ci(r_cl, "lab")
            row.update({"OR_lab2": ol, "lab2_CI": f"[{oll:.2f}, {olh:.2f}]", "lab2_CI_cluster": f"[{cll:.2f}, {clh:.2f}]"})
        rows.append(row)
    htitle(f"H5: September OR for risk_safety, M3 vs M4 (= M3 + lab2); cluster-robust CI uses {n_cl} ISO-week clusters")
    with pd.option_context("display.width", 300):
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.2f}", na_rep="-"))
    print(f"Note: with about {n_cl} clusters the cluster-robust standard errors are unreliable.")
````

````python
# ===== Part 3.3g-4 (team) - H6: sensitivity of the H3 / H5 results =====
if not H_OK:
    print("H2 prerequisites missing - H6 skipped.")
else:
    keep = ~gz_day.loc[hz.index].between("2026-09-14", "2026-09-20")
    variants = [("lab2, all days (H3/H5)", None, "lab2"),
                ("(a) lab2, Sep 14-20 removed", keep, "lab2"),
                ("(b) lab2_body, all days", None, "lab2_body"),
                ("(b) lab2_body, Sep 14-20 removed", keep, "lab2_body")]
    rows = []
    for vname, mask, lab_col in variants:
        mh = stratified_by(lab_col, mask=mask, show=False)
        res, _ = h_models(mask=mask, lab_col=lab_col)
        o3 = or_ci(res["M3"][0])
        o4 = or_ci(res["M4"][0])
        ol = or_ci(res["M4"][0], "lab")
        rows.append({"variant": vname, "n": int(len(hz) if mask is None else mask.sum()),
                     "MH_OR_sep": f"{mh['MH_OR']:.2f} [{mh['CI_low']:.2f}, {mh['CI_high']:.2f}]",
                     "BD_p": mh["Breslow-Day_p"],
                     "OR_sep_M3": f"{o3[0]:.2f} [{o3[1]:.2f}, {o3[2]:.2f}]",
                     "OR_sep_M4": f"{o4[0]:.2f} [{o4[1]:.2f}, {o4[2]:.2f}]",
                     "OR_lab_M4": f"{ol[0]:.2f} [{ol[1]:.2f}, {ol[2]:.2f}]"})
    htitle("H6: sensitivity - MH pooled OR (Sep vs Jun-Aug) and GLM September OR (ordinary SE)")
    with pd.option_context("display.width", 300):
        print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.3f}"))
````

### 3.5 Reality Check และข้อจำกัดของ Part 3

**คำอธิบายทางเลือกที่ทดสอบกับผลกรอบความเสี่ยง**
* **ส่วนประกอบของ section/ประเภทบทความเปลี่ยน:** ไม่ใช่คำอธิบายหลัก (odds ratio แบบรวม 1.89 ใกล้ค่าที่ไม่แยก 1.96) แต่ผลไม่เท่ากันในแต่ละ section (Breslow-Day p = 0.012) เพิ่มชัดใน technology (50.0% เทียบ 22.4%) ส่วน section อื่นไม่เพิ่มอย่างชัดเจน
* **คำกว้างขับผล:** ใช่ ตัดกลุ่มคำ generic_hazard (safety, threat, danger ฯลฯ) ออก ส่วนต่างเหลือ +3.1 จุด (CI −0.8 ถึง 7.7)
* **เกิดจากช่วงสั้นช่วงเดียว:** บางส่วน 51 จาก 71 บทความที่ติดกรอบนี้ใน ก.ย. ลงช่วง 10–24 ก.ย. ตัดสัปดาห์ 14–20 ก.ย. ส่วนต่างเหลือ +8.5 จุด (CI 0.7–16.9) ตัดช่วง 10–24 ก.ย. เหลือ +0.8 จุด (ช่วงนี้เลือกหลังเห็นข้อมูล) ความผันผวนรายสัปดาห์ของ มิ.ย.–ส.ค. ไม่มีหลักฐานว่าต่างกัน (chi-square p = 0.134) แต่การทดสอบนี้กำลังต่ำเพราะ n ต่อสัปดาห์เล็ก
* **ผลรอง (เชิงสำรวจ):** กรอบกฎหมายพบในข่าวมากกว่าความเห็น (28.4% เทียบ 11.9%) ผลทนต่อชุดคำและเกณฑ์ แต่ข่าวและความเห็นแทบไม่อยู่ section เดียวกัน อ่อนลงเมื่อใช้เนื้อเต็ม (+9.1 จุด, CI −0.9 ถึง 18.7) และใน ก.ย. (+7.9 จุด, CI −7.9 ถึง 20.0) จึงไม่นำมาเป็นผลหลัก

**ข้อจำกัด**
* ใช้ The Guardian สื่อเดียว ข้อมูลจบที่ 1 ต.ค. 2569 (เดือน ต.ค. ไม่ครบ จึงไม่ใส่ในกราฟแนวโน้ม)
* กรอบข่าวจัดด้วยรายการคำ (v1) ที่ล็อกไว้ก่อนดูผล ความสอดคล้องของ AI สองตัวบนตัวอย่าง 50 บทความอยู่ระดับ kappa 0.43 (risk) ถึง 0.90 (law) ยังไม่ได้ตรวจกับป้ายที่คนติดอิสระ
* โปรแกรมจับบางกรอบได้น้อย (เช่น geopolitics) การเปรียบเทียบ "ระหว่างกรอบ" จึงไม่ถูกต้อง ใช้เปรียบเทียบ "ภายในกรอบเดียวกันข้ามช่วงเวลาหรือกลุ่ม" เท่านั้น ซึ่งสมมติว่าความคลาดเคลื่อนของการวัดคงที่ ข้อสมมติที่อาจไม่จริง
* การตรวจสัปดาห์ การแยกกลุ่มคำ และการตรวจ section เพิ่มหลังเห็นผลเปรียบเทียบหลัก เป็นเชิงสำรวจ และรายงานทุกแบบที่รันแล้ว หลายการทดสอบจึงอาจให้ผลเกินจริง

# เกณฑ์การให้คะแนน (Rubric) — 100 คะแนน + Bonus สูงสุด 5 คะแนน

คะแนนเน้น **ความถูกต้อง + Insight + ประโยชน์ต่อการตัดสินใจ** มากกว่าจำนวนเทคนิค

## Part 1 — Individual Warm-up (10 คะแนน)

| เกณฑ์ | คะแนน |
|---|---:|
| ตอบ 4 คำถามหลักถูกต้อง | 6 |
| Data handling / reproducibility | 2 |
| การตีความและข้อจำกัด | 2 |

## Part 2 — Voice of Customer / Business Insight (50 คะแนน)

| เกณฑ์ | คะแนน | สิ่งที่ดู |
|---|---:|---|
| **Data preparation** | 8 | sampling, join, missing/duplicate และข้อจำกัด |
| **EDA ที่เกี่ยวกับโจทย์** | 7 | สำรวจเท่าที่จำเป็นและใช้ต่อยอดคำถามได้ |
| **Text features + เทคนิค** | 10 | อย่างน้อย 2 features, วิธีถูกและเหมาะกับ brief |
| **Insight & Business usefulness** | **15** | Insight Card 2 ใบ มีหลักฐาน So what และ Action |
| **Reality Check / ความถูกต้องของข้อสรุป** | 6 | ตรวจ alternative explanation อย่างน้อย 1 insight |
| **Visualization + reproducibility** | 4 | กราฟชัด โค้ดและผลตรวจตามได้ |

## Part 3 — Data Collection Robot + External Insight (25 คะแนน)

| เกณฑ์ | คะแนน | สิ่งที่ดู |
|---|---:|---|
| **คำถามและประโยชน์** | 5 | รู้ว่าจะเก็บไปตอบอะไร |
| **แหล่งข้อมูล + วิธีเก็บ** | 5 | Robot/API/RSS เหมาะสมและรับผิดชอบ |
| **Data quality** | 5 | อย่างน้อย 200 valid records และตรวจคุณภาพ |
| **Technique correctness** | 4 | ใช้ 1–2 เทคนิคที่เหมาะและคำนวณถูก |
| **Insight + limitation** | 6 | มีหลักฐาน, Action และไม่ overclaim |

## การนำเสนอและ Q&A (15 คะแนน)

| เกณฑ์ | คะแนน |
|---|---:|
| Storyline + เวลา | 4 |
| Visualization / การสื่อสารผล | 3 |
| อธิบายเหตุผลของวิธีวิเคราะห์ | 3 |
| **Q&A ในส่วนที่ตนรับผิดชอบ (รายบุคคล)** | **5** |

> คะแนน Part 2–3 เป็นคะแนนทีม แต่ Q&A 5 คะแนนคิดเป็นรายบุคคล
